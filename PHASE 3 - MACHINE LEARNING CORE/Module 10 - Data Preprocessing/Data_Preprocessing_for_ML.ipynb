# SECTION 1 - WHAT IS DATA PREPROCESSING?

**Definition**: data preprocessing is the set of operations that transform raw, messy, inconsistent data into a clean, validated, correctly-shaped, model-ready form an ML algorithm can actually learn from. **Purpose**: raw data almost never arrives in the numeric, complete, consistently-formatted shape a model requires - preprocessing is the bridge between "data as it exists in the real world" and "data as scikit-learn's `fit()` requires it."

**Why raw data cannot usually be fed directly into models**: models require numeric input (categorical text must be encoded); most algorithms cannot handle missing values natively; features on wildly different scales distort distance-based and gradient-based algorithms; inconsistent formatting (mixed date formats, inconsistent category casing) silently corrupts learned patterns; duplicate or invalid records bias what the model learns.

**Disambiguating four related terms**: **data cleaning** fixes *errors* (typos, invalid values, duplicates) - a subset of preprocessing; **preprocessing** is the full pipeline from raw to model-ready, including cleaning *and* encoding, scaling, and transformation; **feature engineering** *creates new* predictive signal (ratios, aggregates) - preprocessing prepares existing data, feature engineering adds new information; **EDA** is an *investigative* activity (understanding the data) that informs preprocessing decisions but doesn't itself transform the data.

```
Raw Data -> Data Validation -> Data Cleaning -> Data Transformation -> Feature Preparation -> Model-Ready Data
```

Each arrow is a distinct, necessary stage this module covers in full: **validation** (Section 4) confirms the data matches expectations before touching it; **cleaning** (Sections 5-10) fixes missingness, duplicates, and invalid values; **transformation** (Sections 11-32) reshapes distributions, encodes categories, and scales numerics; **feature preparation** (Sections 41-49) constructs and selects the final feature set; **model-ready data** is what finally reaches `.fit()`.

```
Data Ingestion -> Schema Validation -> Data Quality Checks -> Missing Values -> Duplicates -> Invalid Values
   -> Outliers -> Categorical Encoding -> Numerical Transformation -> Scaling -> Feature Construction
      -> Feature Selection -> Train/Validation/Test Pipeline -> Model
```

**Each stage, in one line**: **ingestion** (Section 56) - get the raw data into your working environment; **schema validation** (Section 4) - confirm structure matches expectations before any transformation; **data quality checks** (Section 64) - completeness, validity, consistency; **missing values** (Sections 5-8) - detect and handle; **duplicates** (Section 9) - detect and resolve; **invalid values** (Section 10) - detect and correct; **outliers** (Sections 11-13) - detect and decide treatment; **categorical encoding** (Sections 14-20) - turn categories into numbers; **numerical transformation** (Sections 26-27) - reshape distributions; **scaling** (Sections 21-25) - normalize numeric ranges; **feature construction** (Sections 41-47) - build new predictive signal; **feature selection** (Section 48) - keep only what genuinely helps; **train/validation/test pipeline** (Sections 33-35) - the critical leakage-prevention boundary; **model** - the destination this entire module exists to prepare data for.

Note this workflow's ordering is deliberate, not arbitrary: validation happens before cleaning (no point cleaning data that fails structural checks entirely); missingness/duplicates/invalid-values/outliers happen before encoding/scaling (a scaler fit on data still containing `-999` placeholder values for missing income produces a badly distorted result); the train/test split happens *before* any fit-dependent step (Section 33) - this single ordering decision is the boundary that separates safe preprocessing from leaked preprocessing throughout this entire module.

Getting dtypes right, explicitly, at ingestion is the cheapest and highest-leverage preprocessing step in this entire module - a two-minute check (`df.dtypes`) that prevents dozens of downstream bugs.

# SECTION 4 - SCHEMA VALIDATION

```python
# [Python] A simple, reusable schema validation system
from dataclasses import dataclass, field

@dataclass
class ColumnSchema:
    name: str
    dtype: str
    required: bool = True
    allowed_values: set = None
    min_value: float = None
    max_value: float = None
    unique: bool = False
    nullable: bool = True

def validate_schema(df, schema: list[ColumnSchema]) -> list[str]:
    errors = []
    for col in schema:
        if col.name not in df.columns:
            if col.required:
                errors.append(f"Missing required column: {col.name}")
            continue
        if str(df[col.name].dtype) != col.dtype:
            errors.append(f"{col.name}: expected dtype {col.dtype}, got {df[col.name].dtype}")
        if not col.nullable and df[col.name].isna().any():
            errors.append(f"{col.name}: contains nulls but is marked non-nullable")
        if col.allowed_values is not None:
            bad = set(df[col.name].dropna().unique()) - col.allowed_values
            if bad:
                errors.append(f"{col.name}: unexpected values {bad}")
        if col.min_value is not None and (df[col.name] < col.min_value).any():
            errors.append(f"{col.name}: values below minimum {col.min_value}")
        if col.max_value is not None and (df[col.name] > col.max_value).any():
            errors.append(f"{col.name}: values above maximum {col.max_value}")
        if col.unique and df[col.name].duplicated().any():
            errors.append(f"{col.name}: duplicate values found but column marked unique")
    return errors

schema = [
    ColumnSchema("customer_id", "int64", unique=True, nullable=False),
    ColumnSchema("age", "int64", min_value=0, max_value=120),
    ColumnSchema("gender", "object", allowed_values={"Male", "Female", "Other"}),
]
errors = validate_schema(df, schema)
```

Validating **column names**, **data types**, **required columns**, **allowed categories**, **numerical ranges**, **uniqueness**, and **nullability** - all in one declarative pass, run *before* any transformation - catches structural problems in seconds rather than discovering them as a confusing downstream error three pipeline stages later (Section 80's troubleshooting catalog covers exactly this class of failure). This hand-rolled system is the conceptual foundation Section 64 extends with dedicated tools (Great Expectations, Pandera, Pydantic).

# SECTION 5 - MISSING DATA

**Why values go missing**: a user skipped an optional form field; a sensor failed to report; a join produced no match; a system upgrade changed what's tracked; a value was deliberately withheld (a high earner declining to state income).

| Mechanism | Definition | Example |
|---|---|---|
| MCAR (Missing Completely At Random) | Missingness unrelated to any variable, observed or not | A random subset of surveys lost in transit |
| MAR (Missing At Random) | Missingness depends only on *observed* variables | Older customers less likely to have an email on file (predictable from age) |
| MNAR (Missing Not At Random) | Missingness depends on the *unobserved* value itself | High earners specifically declining to report income |
| Structural missingness | The value is *logically* inapplicable, not truly missing | `spouse_income` is null because the customer is unmarried |
| System failure | A pipeline/sensor/integration broke, not a real-world absence | A sensor outage produces a gap in readings |
| User behavior | The user chose not to provide the value | Skipping an optional profile field |

Distinguishing these matters directly: **structural missingness** should often be encoded explicitly (a separate "not applicable" category) rather than imputed at all; **MNAR** is the hardest and riskiest case for naive imputation (Section 6) since the missingness itself carries information a mean/median fill destroys; **system-failure missingness** is a data-quality bug to be *fixed upstream* where possible, not silently imputed around indefinitely.

# SECTION 6 - MISSING VALUE STRATEGIES

```python
# [Python/scikit-learn]
from sklearn.impute import SimpleImputer

# Numerical
mean_imputer = SimpleImputer(strategy="mean")           # symmetric distributions, no major outliers
median_imputer = SimpleImputer(strategy="median")         # skewed distributions, robust to outliers
constant_imputer = SimpleImputer(strategy="constant", fill_value=-1)  # when "missing" itself should be a distinct signal

# Interpolation - for ordered/time-series data
df["sensor_reading"] = df["sensor_reading"].interpolate(method="linear")

# Categorical
mode_imputer = SimpleImputer(strategy="most_frequent")
df["region"] = df["region"].fillna("Unknown")             # an explicit sentinel category, often clearer than mode-filling
```

**When each is appropriate**: mean for roughly symmetric numeric data; median when skew or outliers are present (Section 11); mode for categoricals with one clearly dominant category; a constant/sentinel ("Unknown", -1) when missingness is itself meaningful or when blending in a statistical guess would be misleading; interpolation for ordered/sequential data (sensor readings, time series) where neighboring values are genuinely informative about the missing point; model-based imputation (Section 8) when relationships between features can predict the missing value more accurately than any single summary statistic. Every one of these, without exception, must be **fit on training data only** (Section 33) - the single most important rule in this entire module, repeated because it is also the single most common leakage mistake (Section 35, 79).

# SECTION 7 - MISSINGNESS INDICATORS

```python
# [Python]
df["income_missing"] = df["income"].isna().astype(int)
df["income"] = df["income"].fillna(df["income"].median())   # impute AND preserve the missingness signal
```

**When missingness itself can be predictive**: in the MNAR case (Section 5), the *fact* that a value is missing often carries real information - a loan applicant who leaves `income` blank may systematically differ (in ways relevant to default risk) from one who reports it, regardless of what the imputed value ends up being. Adding an explicit `*_missing` binary indicator alongside the imputed value lets the model learn from *both* signals - the best-guess value *and* the fact that a guess was needed at all - a simple, high-leverage technique many practitioners forget, especially valuable whenever missingness rates are non-trivial (roughly above 5%) or plausibly non-random.

# SECTION 8 - ADVANCED IMPUTATION

**KNN imputation**: fills a missing value using the average (or a weighted average) of the value from the $k$ most similar rows (by distance on other features) - more nuanced than a single global statistic, but sensitive to feature scaling (must be applied to already-scaled data) and computationally expensive on large datasets. **Iterative Imputer / MICE concepts (Multiple Imputation by Chained Equations)**: models each column with missing values as a function of every *other* column, iterating repeatedly until the imputed values stabilize - captures multivariate relationships far richer than any single-statistic method. **Regression-based imputation**: a simpler, single-pass version of the same idea - train a regression model to predict one column from the others, using only rows where that column isn't missing, then predict the missing values.

**Advantages and limitations**: advanced methods generally produce more *accurate* individual imputed values than simple statistics, especially when missingness is MAR and correlates with other observed features - but they are more computationally expensive, harder to explain to stakeholders, riskier to get subtly wrong (they must still be fit on training data only, Section 33), and can create a false sense of precision (an imputed value looks like real data but is fundamentally a model's guess) that downstream error analysis (Module 8, Section 62-64) should account for.

# SECTION 9 - DUPLICATE HANDLING

```python
# [Python]
exact_dupes = df[df.duplicated(keep=False)]                          # every column identical
id_dupes = df[df.duplicated(subset=["customer_id"], keep=False)]       # same business key, other columns may differ
txn_dupes = df[df.duplicated(subset=["account_id", "amount", "timestamp"], keep=False)]  # likely re-sent transaction
```

| Duplicate type | Definition |
|---|---|
| Exact duplicates | Every column identical - usually a genuine pipeline re-ingestion artifact |
| Duplicate IDs | Same business key, differing other columns - often an update overwrite gone wrong or unresolved MDM conflict |
| Duplicate transactions | The same real-world event logged twice (a retried API call, a double-submitted form) |
| Near-duplicates | Very similar but not identical rows (fuzzy matching required to detect, e.g. slightly different name spelling for the same customer) |
| Business-rule duplicates | Rows that are *legitimately* distinct by business definition but *look* like duplicates on a naive check |

**Why blindly removing duplicates can destroy legitimate repeated observations**: not every repeated-looking row is a data error - a customer can legitimately make two identical $50 purchases on the same day; a sensor can legitimately report the same reading twice in a row during a stable period; a patient can have two visits with identical vital signs. Deduplicating on *every* column, without first understanding what a legitimate repeat looks like for this specific dataset's grain (Module 8 Section 6), risks silently discarding real, valid data - always deduplicate against an explicit, deliberately chosen business key (Section 9's `id_dupes`/`txn_dupes` pattern), never a blind `drop_duplicates()` with no `subset` specified, unless you've explicitly confirmed every column together genuinely defines uniqueness for this data's grain.

# SECTION 10 - INVALID VALUES

```python
# [Python] Detection
invalid_age = df[(df["age"] < 0) | (df["age"] > 120)]
invalid_income = df[df["income"] < 0]
invalid_gender = df[~df["gender"].isin({"Male", "Female", "Other"})]
invalid_dates = df[pd.to_datetime(df["date"], errors="coerce").isna() & df["date"].notna()]

# Correction strategies
df.loc[df["age"] < 0, "age"] = np.nan                        # treat as missing, then impute (Section 6)
df.loc[df["income"] < 0, "income"] = np.nan
df["gender"] = df["gender"].where(df["gender"].isin({"Male", "Female", "Other"}), "Unknown")
df["date"] = pd.to_datetime(df["date"], errors="coerce")       # unparseable dates become NaT, handled explicitly
```

**Worked examples**: `Age = -5` and `Age = 300` are both physically impossible - neither is a "real" extreme value (contrast with Section 11's outliers, which *are* real, just extreme) - the correct treatment is to convert to missing and handle via Section 6's imputation strategies, not to clip toward plausibility (which can silently mask a genuine upstream data-entry bug worth fixing at the source). `Income = -100000` is similarly implausible for most contexts and should be treated the same way. `Gender = "Unknown123"` reflects a data-entry or integration bug producing a category outside the approved domain (Section 4's schema validation) - map to an explicit sentinel rather than silently dropping the row. `Date = "2025-99-99"` fails to parse entirely - `errors="coerce"` converts it to `NaT`, making it visible and handleable rather than crashing the whole pipeline or silently propagating a garbage value.

**The general principle**: invalid values are *logically impossible* given the column's real-world meaning (not just statistically unusual); detect them with explicit domain-informed range/domain checks (Section 4), and correct by converting to missing (then imputing per Section 6) rather than guessing a "fixed" value - the one exception being a well-understood, documented correction rule (e.g. a known unit-conversion bug affecting a specific date range).

# SECTION 11 - OUTLIER TREATMENT

```python
# [Python]
Q1, Q3 = df["amount"].quantile([0.25, 0.75])
IQR = Q3 - Q1
lower, upper = Q1 - 1.5*IQR, Q3 + 1.5*IQR

df["amount_clipped"] = df["amount"].clip(lower=lower, upper=upper)   # clipping - cap without removing the row

from scipy.stats.mstats import winsorize
df["amount_winsorized"] = winsorize(df["amount"], limits=[0.01, 0.01])  # winsorization - cap at a percentile

df["log_amount"] = np.log1p(df["amount"])                                # transformation - reshape rather than cap (Section 12)

outlier_rows = df[(df["amount"] < lower) | (df["amount"] > upper)]         # removal - only when clearly erroneous, not a default

from sklearn.preprocessing import RobustScaler
robust_scaled = RobustScaler().fit_transform(df[["amount"]])                 # robust scaling - reduces outlier INFLUENCE without altering values
```

**The six treatments, and when each applies**: **clipping** caps extreme values at a chosen bound while preserving every row - a good default when extreme values are real but you want to limit their influence on scale-sensitive models; **winsorization** is functionally similar, capping at a specified percentile rather than an IQR-derived bound; **transformation** (Section 12-13) reshapes the whole distribution rather than singling out extremes, often the better choice when skew (not just a handful of extreme points) is the underlying issue; **removal** should be reserved for values already confirmed *invalid* (Section 10), not simply extreme - removing genuine extreme observations discards real signal; **robust scaling** (Section 24) doesn't touch the outlier values at all, just reduces their influence on the *scaling parameters* used for every other value; **keeping legitimate extremes** untouched is itself a valid, often correct decision, especially in fraud/anomaly-detection contexts where extremes are exactly what the model must learn to recognize.

**The critical distinction repeated for emphasis**: outlier *detection* (IQR, z-score, robust methods - the statistical identification of candidates) and outlier *treatment* (what you then do about them) are two entirely separate decisions - detecting a value as statistically extreme says nothing about whether it's an error, a rare-but-real event, or exactly the signal your model needs to learn. Never let a detection method automatically trigger a treatment without this intermediate judgment call.

# SECTION 12 - LOG TRANSFORMATION

```python
# [Python/NumPy]
df["log_income"] = np.log1p(df["income"])   # log(1+x) - safely handles zero values, unlike plain log(x)
# to reverse: df["income_restored"] = np.expm1(df["log_income"])
```

**Why useful for right-skewed variables**: many real-world numeric features (income, transaction amounts, claim sizes, page views) are strongly right-skewed - a long tail of large values stretches the distribution, which can disproportionately influence distance-based and gradient-based models and violate normality assumptions some statistical methods rely on. A log transform compresses large values proportionally more than small ones, pulling the distribution toward a more symmetric, closer-to-normal shape - `np.log1p` specifically (rather than `np.log`) is preferred because it's defined at zero (`log1p(0) = 0`) where plain `log(0)` is undefined, a common real-world case for count-like or amount-like features that can legitimately be zero. Always remember to back-transform (`np.expm1`) before reporting predictions in business-facing units - a frequently forgotten step (Module 8's Section 103 flags exactly this bug).

# SECTION 13 - POWER TRANSFORMATIONS

```python
# [Python/scikit-learn]
from sklearn.preprocessing import PowerTransformer

box_cox = PowerTransformer(method="box-cox")          # requires STRICTLY POSITIVE input
box_cox.fit(X_train[["income"]])                        # income must be > 0 for every row

yeo_johnson = PowerTransformer(method="yeo-johnson")      # works with zero and NEGATIVE values too
yeo_johnson.fit(X_train[["profit"]])                        # profit can legitimately be negative
```

**Box-Cox**: a family of power transformations parameterized by lambda, automatically fit to find the value of lambda that makes the transformed data closest to normally distributed - but strictly requires all input values to be positive, making it unusable for features that can be zero or negative without a prior shift. **Yeo-Johnson**: a generalization of Box-Cox that handles zero and negative values natively, at the cost of being slightly less interpretable and occasionally less powerful at normalizing than Box-Cox on strictly-positive data. **Differences and limitations**: prefer Box-Cox when a feature is guaranteed strictly positive and interpretability of the specific transformation matters; prefer Yeo-Johnson as the safer general default, especially for features like profit/margin/temperature-change that can be negative; both, like the log transform, must be fit on training data only and applied identically at inference (Section 33, 60) - and both can occasionally *overcorrect*, producing a transformation that fits training data's specific shape well but doesn't generalize cleanly to new data with a meaningfully different distribution.

# SECTION 14 - CATEGORICAL VARIABLES

| Type | Definition | Example |
|---|---|---|
| Nominal | No inherent order | `payment_method`, `region` |
| Ordinal | A genuine, meaningful order | `contract_type` (Month-to-month < One year < Two year) |
| Binary | Exactly two categories | `is_active` |
| High-cardinality | Many distinct values | `zip_code`, `merchant_id` |

**Why this classification matters before choosing an encoding (Sections 15-20)**: treating a nominal category as ordinal (or vice versa) implies a false relationship the model will incorrectly learn from; high-cardinality categoricals overwhelm naive one-hot encoding with hundreds or thousands of sparse columns, usually requiring frequency, target, or hashing-based encoding instead (Sections 18-20); rare categories create unstable, poorly-estimated encoded values (a target-encoded category based on 2 observations is essentially noise) and are often better grouped into an explicit "Other" bucket before encoding.

# SECTION 19 - TARGET ENCODING

```python
# [Python] Correct, leakage-safe (out-of-fold) target encoding
from sklearn.model_selection import KFold
import numpy as np

def target_encode_out_of_fold(df, cat_col, target_col, n_splits=5, smoothing=10):
    df = df.copy()
    df[f"{cat_col}_target_enc"] = np.nan
    global_mean = df[target_col].mean()
    kf = KFold(n_splits=n_splits, shuffle=True, random_state=42)
    for train_idx, val_idx in kf.split(df):
        fold_train = df.iloc[train_idx]
        category_stats = fold_train.groupby(cat_col)[target_col].agg(["mean", "count"])
        smoothed = (category_stats["mean"] * category_stats["count"] + global_mean * smoothing) / (category_stats["count"] + smoothing)
        df.loc[df.index[val_idx], f"{cat_col}_target_enc"] = df.iloc[val_idx][cat_col].map(smoothed).fillna(global_mean)
    return df
```

**Category -> target mean**: replaces each category with the average target value for that category - often highly predictive, but the single riskiest encoding technique in this module for leakage. **Leakage**: computing a category's target mean using the *same* rows that mean is then applied to directly encodes the outcome into the feature for those exact rows - a severe form of target leakage (Module 8 Section 19). **The fix - cross-validation / out-of-fold encoding**: compute each row's encoded value using only *other* folds' data (as in the function above), so no row's own target ever contributes to its own encoded feature value. **Smoothing**: blends a category's own mean with the global mean, weighted by how many observations that category has - preventing rare categories (Section 14) from producing wildly unstable, overfit encoded values based on just a handful of observations. This combination - out-of-fold computation plus smoothing - is the only professionally acceptable way to implement target encoding; a naive `df.groupby(cat_col)[target_col].transform("mean")` applied directly is a textbook leakage bug.

# SECTION 20 - HASHING-BASED ENCODING

```python
# [Python] Conceptual hashing encoding for very high-cardinality categoricals
import hashlib

def hash_encode(value: str, n_buckets: int = 1024) -> int:
    return int(hashlib.md5(str(value).encode()).hexdigest(), 16) % n_buckets

df["merchant_hash"] = df["merchant_id"].apply(hash_encode)
```

**High-cardinality categorical preprocessing**: when a categorical feature has thousands or millions of distinct values (a `merchant_id`, a `user_agent` string, a `product_sku`), neither one-hot (too many columns) nor target/frequency encoding alone (too many rare, unstable categories, Section 14) may be practical or safe. Hashing maps every category, including ones never seen during training, to a fixed, bounded number of buckets via a deterministic hash function - the encoding "just works" on entirely new categories at inference (no `handle_unknown` problem, unlike Section 15) at the cost of occasional **hash collisions** (two different categories mapping to the same bucket, a small, generally acceptable information loss traded for bounded dimensionality and inference robustness). This technique is especially common in large-scale, high-cardinality production systems (ad tech, large-scale recommendation) where the number of distinct categories is effectively unbounded and unpredictable in advance.

# SECTION 21 - WHY SCALING MATTERS

| Algorithm | Scale-sensitive? | Why |
|---|---|---|
| KNN | Yes | Distance calculations are dominated by whichever feature has the largest raw numeric range |
| K-Means | Yes | Cluster assignment is distance-based, same issue as KNN |
| SVM | Yes | The margin-maximization objective is scale-dependent |
| Logistic Regression | Yes | Gradient-descent optimization converges slowly/unevenly across differently-scaled features; regularization penalizes coefficients unevenly if features aren't comparably scaled |
| Neural Networks | Yes | Gradient-based optimization, same issue as logistic regression, often more severe |
| PCA | Yes | Principal components are computed from feature variance - an unscaled high-variance feature dominates the components regardless of actual importance |
| Tree-based models (decision trees, random forests, gradient boosting) | Generally no | Splits are based on threshold comparisons *within* one feature at a time, unaffected by monotonic rescaling |

**Why tree-based algorithms are generally less sensitive to feature scale**: a decision tree asks "is feature X greater than threshold T" - the specific numeric scale of X doesn't change *which* rows fall on each side of any given relative threshold, since the tree independently finds its own threshold for each feature regardless of its range. This is precisely why scaling is often skipped for random forest/gradient boosting pipelines - not because scaling would *hurt*, but because it provides no benefit, and skipping an unnecessary step keeps the pipeline simpler (a still-legitimate choice covered as a common near-mistake in Section 79 - "scaling tree models unnecessarily" is listed there specifically because it's harmless but adds needless pipeline complexity, not because it's wrong).

# SECTION 26 - NORMALIZATION VS STANDARDIZATION

| Term | What it actually means (precisely) |
|---|---|
| Standardization | Rescale to mean 0, std 1 - `StandardScaler` (Section 22) |
| Normalization (statistics/ML sense) | Rescale to a fixed range, typically [0,1] - `MinMaxScaler` (Section 23); in some contexts, rescale each *row* to unit norm (`sklearn.preprocessing.Normalizer`) |
| Scaling | The general umbrella term for any technique adjusting a feature's numeric range/spread |
| Transformation | Reshaping a distribution's *shape* (skew, kurtosis) - log, Box-Cox, Yeo-Johnson, quantile (Sections 12-13, 27) - distinct from merely rescaling range |

**Why this distinction is worth stating explicitly**: these four terms are used inconsistently across textbooks, libraries, and casual conversation, and the confusion causes real bugs - "normalize the data" is genuinely ambiguous between "apply `MinMaxScaler`" and "apply `Normalizer`" (a *row-wise* operation, rescaling each observation's feature vector to unit length, common in text/embedding contexts but rarely appropriate for typical tabular features) - always confirm precisely which operation a colleague, a paper, or a library's documentation means by "normalization" in context, rather than assuming.

# SECTION 27 - QUANTILE TRANSFORMATION

```python
# [Python/scikit-learn]
from sklearn.preprocessing import QuantileTransformer

qt_uniform = QuantileTransformer(output_distribution="uniform", random_state=42)
qt_uniform.fit(X_train[["income"]])

qt_normal = QuantileTransformer(output_distribution="normal", random_state=42)
qt_normal.fit(X_train[["income"]])
```

**Distribution reshaping**: `QuantileTransformer` maps each feature's values to their rank/quantile position in the training distribution, then maps those quantiles onto either a **uniform** [0,1] distribution or a **normal** distribution, regardless of the original distribution's shape - unlike a log or power transform (Sections 12-13), which apply a fixed mathematical function, quantile transformation forces the output into a *specific target shape* directly, non-linearly. **Risks of over-transformation**: because it's rank-based, it can compress the differences between values that were genuinely far apart in the original scale (two values at the 90th and 99th percentile may end up much closer together in the transformed space than their original difference suggests) - and it's particularly prone to overfitting the training data's exact distribution shape if the training set is small, producing a transformation that doesn't generalize well to a validation/test/production distribution that differs even slightly. Use it as a deliberate, tested choice for genuinely difficult, heavily irregular distributions - not a default reach before simpler transforms (Sections 12-13, 22-25) have been tried.

# SECTION 28 - TEXT CLEANING

```python
# [Python]
import re

def clean_text(text: str) -> str:
    text = text.lower()                                          # lowercasing
    text = text.strip()                                            # whitespace
    text = re.sub(r"<[^>]+>", " ", text)                              # HTML tags
    text = re.sub(r"http\S+|www\.\S+", " ", text)                       # URLs
    text = re.sub(r"[^\w\s]", " ", text)                                  # punctuation/special characters
    text = re.sub(r"\s+", " ", text).strip()                                # collapse repeated whitespace
    return text

df["review_clean"] = df["review_text"].apply(clean_text)
```

Covering **lowercasing**, **whitespace**, **punctuation**, **Unicode** (normalizing accented/special characters, e.g. via `unicodedata.normalize`), **special characters**, **HTML**, **URLs**, and **emojis** (often stripped, or deliberately preserved/converted to a text token if sentiment-relevant). **When cleaning can accidentally destroy useful information**: aggressive punctuation removal can erase meaningful signal (`"not good"` vs `"not good!!!"` - the emphasis may correlate with sentiment intensity); lowercasing destroys the distinction between an acronym and a common word (`"US"` the country vs `"us"` the pronoun); emoji removal discards sentiment-rich content in informal text (social media, reviews) where emojis often carry more signal than the surrounding words; overly aggressive stopword-style cleaning (Section 30) can remove negation words (`"not"`, `"no"`) that flip a sentence's entire meaning. The professional discipline: clean deliberately, for a *stated reason* tied to the downstream task, never as a reflexive "always clean everything" default.

# SECTION 29 - TOKENIZATION

| Type | Splits text into | Example |
|---|---|---|
| Word tokenization | Individual words | `"I love this"` -> `["I", "love", "this"]` |
| Subword tokenization | Word pieces, often algorithmically learned (BPE, WordPiece) | `"unbelievable"` -> `["un", "believ", "able"]` |
| Sentence tokenization | Individual sentences | A paragraph split at sentence boundaries |

**Connection to modern NLP and transformers**: classical NLP pipelines (Section 30) typically use word tokenization as the foundation for stopword removal, stemming, and TF-IDF; modern transformer-based models (BERT, GPT-family, and their successors) use **subword tokenization** almost universally, because it elegantly handles rare/unknown words by breaking them into known, previously-seen pieces (never producing a true "out-of-vocabulary" token the way word-level tokenization can), and keeps the vocabulary size bounded and manageable regardless of how much raw text the model has seen. Understanding this distinction is what separates preprocessing designed for classical bag-of-words-style models from preprocessing appropriate for modern transformer pipelines - conflating the two (e.g. applying heavy stemming/lemmatization before feeding text to a transformer) is a specific instance of Section 30's core warning.

# SECTION 30 - TRADITIONAL NLP PREPROCESSING

```python
# [Python] Illustrative classical NLP preprocessing (NLTK-style)
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

stop_words = set(stopwords.words("english"))
tokens = [w for w in tokens if w not in stop_words]              # stopword removal

stemmer = PorterStemmer()
stemmed = [stemmer.stem(w) for w in tokens]                        # stemming - crude, rule-based truncation

lemmatizer = WordNetLemmatizer()
lemmatized = [lemmatizer.lemmatize(w) for w in tokens]                # lemmatization - dictionary-aware, produces real words

from sklearn.feature_extraction.text import TfidfVectorizer
bigrams = TfidfVectorizer(ngram_range=(1, 2), max_features=5000)        # n-grams + TF-IDF
X_text = bigrams.fit_transform(X_train["review_text"])
```

**Stemming vs lemmatization**: stemming crudely truncates words to a root form using fixed rules ("running" -> "run", but also sometimes "university" -> "univers" - not always a real word); lemmatization uses a dictionary/morphological analysis to reduce a word to its true base form ("better" -> "good"), more linguistically accurate but slower. **N-grams**: sequences of $n$ consecutive words/tokens (bigrams, trigrams), capturing local word-order context a pure bag-of-words representation loses entirely. **TF-IDF (Term Frequency-Inverse Document Frequency)**: weights each word by how often it appears in a given document, offset by how common it is *across all documents* - down-weighting generically common words and up-weighting words distinctively characteristic of a specific document.

**When traditional preprocessing is useful vs. inappropriate for modern transformer pipelines**: stemming, lemmatization, stopword removal, and TF-IDF remain genuinely valuable for classical bag-of-words-style models (logistic regression, gradient boosting on TF-IDF features) and for lightweight, interpretable, fast-to-train text classification. They are actively **counterproductive** ahead of a modern transformer model - transformers are pretrained on raw, natural, un-stemmed, un-stopword-removed text and rely on exactly that full linguistic context (including "meaningless" stopwords and word inflections) to build their contextual representations; stripping this information before tokenization *removes signal the model was specifically trained to use*, generally hurting performance. The correct modern default: minimal cleaning (Section 28's basic hygiene - no HTML, no broken encoding) plus the transformer's own built-in subword tokenizer (Section 29), with classical stemming/lemmatization/stopword-removal reserved specifically for classical, non-transformer text pipelines.

# SECTION 32 - CYCLICAL ENCODING

```python
# [Python/NumPy]
df["hour_sin"] = np.sin(2 * np.pi * df["hour"] / 24)
df["hour_cos"] = np.cos(2 * np.pi * df["hour"] / 24)

df["dow_sin"] = np.sin(2 * np.pi * df["day_of_week"] / 7)
df["dow_cos"] = np.cos(2 * np.pi * df["day_of_week"] / 7)

df["month_sin"] = np.sin(2 * np.pi * df["month"] / 12)
df["month_cos"] = np.cos(2 * np.pi * df["month"] / 12)
```

**The problem cyclical encoding solves**: a raw `hour` feature (0-23) implies hour 23 and hour 0 are maximally *far apart* numerically, when in reality they're one hour apart on the clock - the same issue applies to day-of-week (Sunday=6, Monday=0 look numerically distant despite being adjacent) and month (December=12, January=1). `sin(2*pi*x/period)` and `cos(2*pi*x/period)` together map a cyclical value onto a circle, so that the *last* value in the cycle and the *first* value end up genuinely close together in the transformed 2D space, exactly matching their true cyclical relationship - critically, **both** sin and cos components are needed together (a single sine value alone is ambiguous, since multiple points on a circle share the same sine value; the cosine component disambiguates which side of the circle you're on).

**Applied examples**: **hour** (captures daily behavioral cycles - e.g. fraud or usage patterns that differ meaningfully by time of day, with 11pm and midnight correctly treated as adjacent); **day of week** (weekly business cycles); **month** and broader **annual seasonality** (retail demand, energy consumption, agricultural patterns) - any feature with a genuine, meaningful wraparound structure is a candidate for this encoding rather than being fed to a model as a raw, misleadingly-ordered integer.

# SECTION 33 - WHY SPLIT BEFORE LEARNING PREPROCESSING PARAMETERS?

```python
# [Python] The leakage this section exists to prevent, made concrete
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

# WRONG: fit the scaler on the FULL dataset, before splitting
scaler_wrong = StandardScaler().fit(X)                   # X includes what will become the test set
X_scaled_wrong = scaler_wrong.transform(X)
X_train_w, X_test_w = train_test_split(X_scaled_wrong, test_size=0.2, random_state=42)
# The test set's own values contributed to the mean/std used to scale IT and every training row -
# a genuine, if subtle, information leak from test into train.

# CORRECT: split FIRST, fit the scaler on training data ONLY
X_train, X_test = train_test_split(X, test_size=0.2, random_state=42)
scaler_correct = StandardScaler().fit(X_train)              # ONLY training data informs the learned mean/std
X_train_scaled = scaler_correct.transform(X_train)
X_test_scaled = scaler_correct.transform(X_test)               # test data merely TRANSFORMED, never contributes to fitting
```

**The critical principle, stated as plainly as possible**: *fit every preprocessing transformation - scaler, imputer, encoder, feature selector - on training data only.* Test and validation data are *transformed* using parameters already learned from training data, and must never themselves contribute to learning those parameters. **Why the wrong version is genuinely leakage, not just poor style**: the test set's own statistical properties (its mean, its variance, its category frequencies) become baked into the transformation applied to *every* row, including training rows - this means the training process has, in a subtle but real sense, "seen" information derived from the test set before the model was ever evaluated against it, inflating the honesty of the resulting test-set performance estimate. This single principle is the thread connecting virtually every leakage discussion in this module (Section 35) and in Module 8 (Section 19, 30) - internalizing it here, concretely, with a scaler as the simplest possible example, is what makes every more complex leakage scenario in this module immediately recognizable.

# SECTION 35 - PREPROCESSING LEAKAGE

Each example follows: **wrong workflow -> why it leaks -> correct workflow**, extending Module 8's Section 19 with preprocessing-specific detail.

**1. Scaling before splitting.** *Wrong*: `StandardScaler().fit_transform(X)` before `train_test_split`. *Why*: Section 33's exact scenario - test-set values inform the scaling parameters applied to training data. *Correct*: split first; fit only on `X_train`.

**2. Imputation before splitting.** *Wrong*: fill missing `income` with the full dataset's mean before splitting. *Why*: the imputation value is partly derived from test rows. *Correct*: compute the median/mean from `X_train` only (Section 6).

**3. Feature selection before splitting.** *Wrong*: run a correlation-with-target filter (Section 48) using the full dataset, then split. *Why*: the *choice of which features to keep* is itself informed by test-set target values - a subtle, often-overlooked leakage form. *Correct*: perform feature selection using only training data, ideally inside cross-validation folds (Section 88).

**4. Target encoding incorrectly.** *Wrong*: `df.groupby(cat_col)[target].transform("mean")` applied directly, then split. *Why*: Section 19's exact scenario - each row's encoded feature is partly derived from its own target value. *Correct*: out-of-fold target encoding (Section 19), computed after splitting.

**5. Normalization using full data.** *Wrong*: `MinMaxScaler` fit on the complete dataset before any split. *Why*: identical to case 1, applied to a different scaler. *Correct*: fit-on-train-only, always.

**6. Oversampling before splitting.** *Wrong*: apply SMOTE (Module 8's imbalance discussion, Section 50) to the full dataset, then split into train/test. *Why*: synthetic oversampled points generated from *both* what will become train and test rows can end up on both sides of the split - meaning near-duplicate synthetic information leaks across the boundary, inflating test performance. *Correct*: split first; apply oversampling only to the training fold, inside cross-validation if used (Section 88), never touching validation/test data at all.

Every one of these six examples reduces to the exact same root cause: **a fit-dependent statistic or decision was computed using data outside the training set.** Recognizing this one underlying pattern is more valuable than memorizing each individual example - it lets you spot a *novel* leakage scenario this list doesn't explicitly cover, using the same diagnostic question every time: "did anything about this computation depend on rows outside the training fold?"

# SECTION 36 - TRANSFORMER API

| Method | What it does |
|---|---|
| `fit(X, y=None)` | Learns parameters from the given data (a scaler's mean/std, an imputer's fill value, an encoder's category list) |
| `transform(X)` | Applies the already-learned parameters to (possibly different) data |
| `fit_transform(X, y=None)` | `fit()` then `transform()` on the *same* data, in one call - a convenience method, but must only ever be called on training data |

**Estimator vs. transformer vs. predictor**: an **estimator** is the general scikit-learn term for any object implementing `.fit()` - the umbrella category everything below falls under. A **transformer** additionally implements `.transform()` (and usually `.fit_transform()`) - it changes the *shape or values* of the data (`StandardScaler`, `OneHotEncoder`, `SimpleImputer`). A **predictor** additionally implements `.predict()` (and often `.predict_proba()`) - it produces predictions from already-fitted parameters (`LogisticRegression`, `RandomForestClassifier`). This consistent, shared interface across every one of these object types is precisely what makes `Pipeline` (Section 39) and `ColumnTransformer` (Section 38) possible - they can chain arbitrary combinations of transformers and a final predictor together because every component exposes the same small, predictable method surface.

# SECTION 40 - CREATING CUSTOM TRANSFORMERS

```python
# [Python/scikit-learn] A custom transformer using the modern, estimator-compatible API
from sklearn.base import BaseEstimator, TransformerMixin

class DateFeatureExtractor(BaseEstimator, TransformerMixin):
    def __init__(self, date_column: str):
        self.date_column = date_column

    def fit(self, X, y=None):
        return self   # nothing to learn from data - purely a structural transformation

    def transform(self, X):
        X = X.copy()
        dates = pd.to_datetime(X[self.date_column], errors="coerce")
        X[f"{self.date_column}_year"] = dates.dt.year
        X[f"{self.date_column}_month"] = dates.dt.month
        X[f"{self.date_column}_dayofweek"] = dates.dt.dayofweek
        return X.drop(columns=[self.date_column])

class BusinessRuleValidator(BaseEstimator, TransformerMixin):
    def __init__(self, min_age: int = 0, max_age: int = 120):
        self.min_age, self.max_age = min_age, max_age

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        X = X.copy()
        X.loc[(X["age"] < self.min_age) | (X["age"] > self.max_age), "age"] = np.nan
        return X

date_pipeline = Pipeline([("extract_dates", DateFeatureExtractor(date_column="signup_date"))])
```

**Explained**: inheriting from `BaseEstimator` (giving `get_params`/`set_params`, required for `GridSearchCV` compatibility, Section 87) and `TransformerMixin` (automatically providing `fit_transform` from `fit` + `transform`) is the modern, standard way to build a custom, pipeline-compatible component. `fit()` should learn and store any data-dependent parameters as attributes (ending in `_` by scikit-learn convention, e.g. `self.mean_`) if the transformer genuinely needs to learn something from training data; if it's purely a structural/rule-based transformation with nothing to learn (like `DateFeatureExtractor` above), `fit()` simply returns `self` unchanged. `transform()` implements the actual logic and must work correctly and identically whether called on training, validation, test, or live production data. **Examples**: a **custom date extractor** (above); a **business-rule validator** (above, converting known-impossible values to missing, Section 10's logic wrapped as a reusable pipeline step); a **domain-specific feature transformer** (e.g. a `BMICalculator` computing `weight / height**2`, Section 42, as a proper pipeline component rather than ad hoc Pandas code left outside the pipeline). Wrapping domain logic as a custom transformer - rather than leaving it as separate, easy-to-forget Pandas code before the pipeline - is precisely what closes the gap between "preprocessing I did once in a notebook" and "preprocessing that's guaranteed to run identically in production" (Section 60).

# SECTION 41 - WHAT IS FEATURE ENGINEERING?

| Term | Definition |
|---|---|
| Raw feature | A column exactly as it arrived from the source, untransformed |
| Derived feature | Computed from one or more raw features via a defined transformation |
| Interaction feature | Captures a combined effect between two or more features that neither expresses alone (Section 47) |
| Aggregate feature | A group-level summary statistic (Section 44) |
| Temporal feature | Derived from date/time information (Sections 31-32, 45-46) |
| Ratio | One quantity divided by another, capturing a rate or proportion (Section 43) |
| Binary indicator | A 0/1 flag capturing the presence/absence of a condition (Section 7's missingness indicator is one example) |

Feature engineering **creates new predictive signal** - it is preprocessing's natural next step once raw data is clean, valid, and correctly typed (Sections 1-40), but is conceptually distinct: preprocessing prepares what already exists; feature engineering adds genuinely new information a model couldn't otherwise access, encoded as domain-informed derived columns. Every technique in Sections 42-47 is a specific instance of this general idea, and every one must be built with the exact same leakage discipline (Section 33-35) as any other preprocessing step - a feature engineered using information unavailable at real prediction time is just as leaky as a badly-fit scaler.

# SECTION 43 - RATIO FEATURES

```python
# [Python] Safe ratio construction
df["conversion_rate"] = df["conversions"] / df["visits"].replace(0, np.nan)   # avoid division by zero -> inf
df["conversion_rate"] = df["conversion_rate"].fillna(0)                        # decide explicitly what "0 visits" should mean

# Guarding against extreme/unstable ratios
df["debt_ratio"] = (df["debt"] / df["income"].replace(0, np.nan)).clip(upper=10)  # cap implausibly extreme ratios
```

**Division by zero**: a raw `numerator / denominator` where the denominator can legitimately be zero produces `inf` or `NaN` silently - always guard explicitly (`.replace(0, np.nan)` followed by a deliberate fill decision, or a small epsilon added to the denominator) rather than letting `inf` values silently corrupt every downstream statistic and model. **Missing denominators**: if the denominator itself can be missing (Section 5), the ratio is undefined for that row and should typically be treated as missing too, not silently computed as `NaN/x = NaN` and left unexamined. **Extreme ratios**: a customer with `income = 1` and `debt = 50000` produces a `debt_ratio` of 50,000 - technically correct arithmetic, but almost certainly a data-quality problem (Section 10) rather than a genuine signal, and left unclipped can dominate a distance-based or gradient-based model's behavior. **Numerical stability**: ratios involving very small denominators are inherently unstable (a tiny change in the denominator produces a huge swing in the ratio) - consider whether a difference, a log-ratio, or a smoothed ratio (adding a small constant to both numerator and denominator) produces a more stable, more genuinely informative feature than the raw division.

# SECTION 45 - LAG FEATURES

```python
# [Python]
df = df.sort_values(["customer_id", "date"])                       # ALWAYS sort by time first
df["previous_value"] = df.groupby("customer_id")["value"].shift(1)    # the prior observation for this entity
df["lag_7"] = df.groupby("customer_id")["value"].shift(7)
```

**Explaining temporal leakage in this specific context**: `shift(1)` looks *backward* - the previous row's value - which is safe, since it reflects information genuinely available before the current row's timestamp. `shift(-1)` looks *forward* and is a textbook leakage bug (Module 8 Section 47) - at real prediction time, the "next" observation hasn't happened yet, so a feature built from it cannot exist in production even though it's trivially computable from a complete historical dataset during training. Always sort by time *before* any shift operation (an unsorted `shift()` can silently pull from the wrong neighboring row entirely, a distinct and equally dangerous bug from the forward/backward direction issue) and always double-check the shift direction matches "genuinely available before this row's prediction time."

# SECTION 46 - ROLLING FEATURES

```python
# [Python]
df = df.sort_values(["customer_id", "date"])
df["rolling_mean_7"] = df.groupby("customer_id")["amount"].transform(lambda s: s.rolling(window=7, min_periods=1).mean())
df["rolling_sum_30"] = df.groupby("customer_id")["amount"].transform(lambda s: s.rolling(window=30, min_periods=1).sum())
df["rolling_std_7"] = df.groupby("customer_id")["amount"].transform(lambda s: s.rolling(window=7, min_periods=1).std())
df["expanding_avg"] = df.groupby("customer_id")["amount"].transform(lambda s: s.expanding().mean())   # grows to include ALL prior history
```

**Rolling mean, sum, standard deviation**, and **expanding statistics** (a rolling window that grows to include everything up to the current point, rather than a fixed-size window that slides) all capture recent trend/volatility - among the most powerful feature categories in this entire module for behavioral, transactional, and time-series problems. **Why rolling windows must respect prediction time**: a rolling window computed over the *full* dataset without first sorting by time and without confirming the window only looks backward from each row's own timestamp will, exactly like Section 45's lag features, silently include future information relative to that row's actual prediction point - the single most common source of this module's most damaging category of bug (Section 35, 79), because a rolling-window calculation *looks* correct and well-formed even when it's leaking, and the resulting inflated offline performance is discovered, expensively, only in production (Module 8's Section 111 troubleshooting entry "model performs well offline but poorly online" is very often, at its root, exactly this bug).

| Category | Mechanism | Example |
|---|---|---|
| Filter methods | Score each feature independently of any model, using a statistical criterion | Variance threshold, correlation filtering, mutual information |
| Wrapper methods | Use a model's actual performance to iteratively decide which features to keep | Recursive Feature Elimination (RFE) |
| Embedded methods | Feature selection happens *as part of* model training itself | L1 (Lasso) regularization driving unhelpful coefficients to exactly zero |

```python
# [Python/scikit-learn]
from sklearn.feature_selection import VarianceThreshold, mutual_info_classif, RFE
from sklearn.linear_model import LogisticRegression, Lasso

vt = VarianceThreshold(threshold=0.01)                          # drop near-constant features
vt.fit(X_train)

correlation_matrix = X_train.corr()                               # correlation filtering - drop one of any highly correlated pair

mi_scores = mutual_info_classif(X_train, y_train)                    # mutual information - captures NON-linear relevance too

rfe = RFE(estimator=LogisticRegression(max_iter=1000), n_features_to_select=10)  # wrapper method
rfe.fit(X_train, y_train)

lasso = Lasso(alpha=0.1)                                                 # embedded method - L1 shrinks unhelpful coefficients to 0
lasso.fit(X_train, y_train)
```

**Preprocessing prepares** every candidate feature to be usable at all (clean, encoded, scaled); **feature selection** then decides which of those prepared features actually earn their place in the final model - a distinct, later decision, always made using training data only (Section 35's example 3 covers this leakage risk explicitly), and ideally performed *inside* cross-validation (Section 88) rather than once on the full training set, to avoid the subtler leakage of implicitly tuning the feature set to one particular train/validation split.

# SECTION 49 - PCA PREPROCESSING

```python
# [Python/scikit-learn]
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

pca_pipeline = Pipeline([
    ("scaler", StandardScaler()),    # PCA is scale-sensitive - MUST scale first (Section 21)
    ("pca", PCA(n_components=0.95)),  # keep enough components to explain 95% of variance
])
pca_pipeline.fit(X_train)
print(pca_pipeline.named_steps["pca"].explained_variance_ratio_)
```

**Centering, scaling, covariance, eigenvectors, explained variance** - directly connecting back to Module 5's linear algebra: PCA first **centers** the data (subtracts the mean, so the first principal component passes through the data's centroid) and requires prior **scaling** (Section 21-22, since PCA finds directions of maximum *variance*, and an unscaled high-range feature would dominate purely due to its scale, not genuine importance); it then computes the **covariance matrix** of the (centered, scaled) features and finds its **eigenvectors** - the orthogonal directions of maximum variance in the data - each paired with an **eigenvalue** quantifying how much variance that direction explains. **Explained variance** (the proportion of total variance captured by each component, and cumulatively) is what `n_components=0.95` above uses to automatically choose how many components to retain. PCA is a genuine preprocessing/dimensionality-reduction step (reducing many correlated features to fewer, uncorrelated ones) that trades some interpretability (principal components are linear combinations of original features, not directly meaningful business quantities) for reduced dimensionality, reduced multicollinearity (Module 8 Section 17), and often improved performance/training speed for models sensitive to high dimensionality.

**Techniques**: **class weights** (no data modification at all - simply penalize misclassifying the minority class more heavily during training, often the simplest, safest, most defensible first approach); **oversampling** (duplicate or synthesize additional minority-class examples); **undersampling** (discard majority-class examples, risking information loss if the majority class is reduced too aggressively); **SMOTE** (Synthetic Minority Oversampling Technique - generates *new*, synthetic minority examples by interpolating between existing minority examples and their nearest minority neighbors, rather than simply duplicating real rows); **Borderline-SMOTE** and **ADASYN** (variants focusing synthetic generation specifically near the decision boundary or on harder-to-classify minority examples, often more effective than plain SMOTE for genuinely difficult imbalanced problems).

**Why resampling must happen inside the cross-validation pipeline**: exactly Section 35's example 6 - fitting SMOTE (or any resampling) on the full training set *before* cross-validation splits that training set into folds allows synthetic points derived from what becomes one fold's "validation" portion to influence what becomes another fold's "training" portion, and vice versa - leaking information across the CV boundary and inflating validation scores. The correct pattern uses `imblearn`'s `Pipeline` (a drop-in extension of scikit-learn's `Pipeline` supporting resampling steps), ensuring SMOTE is refit fresh, on the training portion only, inside every single cross-validation fold independently (Section 88 covers this pattern in full).

| Domain | Augmentation techniques |
|---|---|
| Images | Rotation, flipping, cropping, color jitter, noise injection (Section 52) |

**When augmentation can introduce unrealistic samples**: augmentation exists to artificially expand limited training data and improve generalization - but applied carelessly, it can produce examples that don't resemble anything the model will actually see in production, actively teaching it wrong patterns. # SECTION 54 - PREPROCESSING IN SQL

```sql
-- [SQL] Core preprocessing operations pushed to the database
SELECT
    customer_id,
    CASE WHEN age < 0 OR age > 120 THEN NULL ELSE age END AS age_clean,      -- CASE for invalid-value correction
    COALESCE(region, 'Unknown') AS region_filled,                              -- COALESCE for simple missing-value fill
    NULLIF(income, -1) AS income_clean,                                          -- NULLIF converts a sentinel value TO null
    EXTRACT(YEAR FROM signup_date) AS signup_year,                                 -- date extraction
    EXTRACT(DOW FROM signup_date) AS signup_dow,
    ROW_NUMBER() OVER (PARTITION BY customer_id ORDER BY updated_at DESC) AS rn      -- window function for deduplication
FROM raw_customers
QUALIFY rn = 1;                                                                        -- [BigQuery/Snowflake] keep only the latest row per customer
```

**What belongs in SQL vs. Python vs. feature pipelines**: push large-scale, set-based operations - deduplication (via `ROW_NUMBER()`), simple imputation (`COALESCE`), invalid-value correction (`CASE`/`NULLIF`), date extraction, and heavy aggregation (`GROUP BY`, window functions) - to **SQL**, where the database engine is purpose-built to do this efficiently at scale, before pulling a much smaller, already-cleaned result into Python (Module 4 Section 33, 40; this module's Section 56); reserve **Python/Pandas** for exploratory, iterative, model-specific logic still actively being developed, and for operations SQL expresses awkwardly (complex conditional feature engineering, custom transformer logic); reserve the **feature pipeline** (scikit-learn `Pipeline`/`ColumnTransformer`, Sections 38-39) specifically for the fit-dependent transformations (scaling, encoding, imputation *statistics*) that must be learned from training data and applied consistently at inference - SQL and Pandas can prepare the *inputs* to these fit-dependent steps, but the fit-on-train-only discipline (Section 33) itself is scikit-learn's job, not SQL's or ad hoc Pandas's.

# SECTION 57 - PREPROCESSING MILLIONS OF RECORDS

**When Pandas becomes unsuitable**: once a dataset's size (or the memory overhead of intermediate copies during preprocessing) approaches or exceeds available RAM, Pandas' fundamentally in-memory design becomes the binding constraint (Module 4 Section 38) - not a hard, universal row-count threshold, but a real limit that depends on machine memory, column count, and how many intermediate copies your specific pipeline creates.

**The escalating toolkit, in order of when to reach for each**: **SQL preprocessing** (Section 54) - push filtering, joining, and aggregation to the database first, minimizing what ever reaches Python; **chunk processing** (`pd.read_csv(..., chunksize=...)`, Module 4 Section 38) - process bounded pieces sequentially, never holding the whole dataset in memory; **Parquet** (Module 4 Section 36) - faster loads, smaller files, column-selective reads, dramatically reducing I/O and memory versus CSV; **column pruning** - load only genuinely needed columns (`usecols`/`columns=`), a simple, high-leverage lever often skipped; **distributed processing** (**Spark**, Section 58; **Dask**, a Pandas-like API that scales a similar workflow across a cluster; **Polars**, Section 59, which scales further on a single machine, often deferring the need for genuine distribution entirely) - reached for once the dataset and required transformations genuinely exceed what chunking/Parquet/column-pruning on one machine can handle in reasonable time.

The professional judgment call, restated from Module 4 Section 38-39 and Module 8 Section 118: profile first, and escalate to the next tier of complexity only once the current tier's limits are actually, measurably hit - not preemptively reaching for Spark on a 2-million-row dataset that fits comfortably in Pandas with sensible dtypes.

# SECTION 59 - POLARS

```python
# [Polars]
import polars as pl

df = pl.scan_parquet("data/transactions.parquet")   # LAZY - builds a query plan, doesn't execute yet

result = (
    df.filter(pl.col("amount") > 0)
      .with_columns([
          (pl.col("amount").log1p()).alias("log_amount"),
          pl.col("date").dt.year().alias("year"),
      ])
      .group_by("customer_id")
      .agg([pl.col("amount").sum().alias("total_amount"), pl.col("amount").mean().alias("avg_amount")])
)

final_df = result.collect()   # NOW the optimized query plan actually executes
```

**Lazy execution**: unlike Pandas' eager execution (every line runs immediately), Polars' `scan_*`/lazy API builds a full query plan first, letting its query optimizer reorder and combine operations (e.g. pushing a filter earlier, only reading needed columns) before anything actually runs - conceptually similar to how a SQL query planner optimizes a query before execution (Section 54), applied to a DataFrame API. **Expressions**: Polars' `pl.col(...)` expression system lets multiple column operations be described declaratively and executed together, enabling far more aggressive internal parallelization than Pandas' typically sequential, eagerly-executed method chains. **Memory efficiency**: Polars is written in Rust with a columnar, Arrow-based memory model, generally using substantially less memory than Pandas for equivalent operations, and is multi-threaded by default without any extra configuration.

**Pandas vs. Polars vs. Spark**: Pandas remains the default, most broadly compatible choice (the entire ML ecosystem - scikit-learn, this module's every prior section - assumes Pandas/NumPy input, Module 4 Section 62); Polars offers substantially better single-machine performance and memory efficiency with a learning curve for its distinct (not drop-in-compatible) API, best suited once Pandas' performance genuinely becomes the bottleneck but data still fits on one machine; Spark is the right choice once data genuinely exceeds single-machine capacity and requires distributed storage and compute across a cluster. This is the identical escalation ladder from Section 57, now compared side by side.

# SECTION 62 - BATCH PREPROCESSING

**Examples**: **nightly customer scoring** (the full active customer base re-scored every night, feeding tomorrow's retention worklist, Module 8 Section 79); **daily fraud analysis** (a daily batch reconciliation and re-scoring pass, distinct from any real-time scoring already happening per-transaction); **monthly risk prediction** (a slower-cadence credit/risk model re-run monthly, matching how frequently underlying financial data genuinely changes).

```python
# [Python] Chunked batch preprocessing for a large scheduled job
def run_batch_preprocessing(input_path: str, pipeline_path: str, output_path: str, chunk_size: int = 100_000):
    pipeline = joblib.load(pipeline_path)
    results = []
    for chunk in pd.read_parquet(input_path, chunksize=chunk_size):
        validate_schema(chunk, schema)                     # Section 4, run on every chunk
        processed = pipeline.transform(chunk)
        results.append(processed)
    pd.concat(results).to_parquet(output_path, index=False)
```

Batch preprocessing's defining characteristic is that **latency is not the binding constraint** - a nightly job has hours, not milliseconds, to complete, which means it can afford heavier validation, chunked processing for very large volumes (Section 57), and more thorough logging/error handling than an online request ever could - the trade-off being that predictions are only as fresh as the last batch run, unsuitable for decisions that must reflect the very latest data (Section 63).

```python
# [Python] Illustrative online preprocessing inside a request handler
from fastapi import FastAPI, HTTPException

app = FastAPI()
pipeline = joblib.load("models/fraud_pipeline_latest.joblib")   # loaded ONCE at startup, never per-request

@app.post("/score")
def score_transaction(transaction: dict):
    try:
        df = pd.DataFrame([transaction])
        errors = validate_schema(df, inference_schema)              # Section 4, applied per-request
        if errors:
            raise HTTPException(status_code=400, detail=errors)
        proba = pipeline.predict_proba(df)[:, 1][0]                    # preprocessing + model, one call, guaranteed consistent
        return {"fraud_probability": float(proba)}
    except Exception as e:
        raise HTTPException(status_code=500, detail=str(e))
```

# SECTION 68 - SECURE DATA PREPROCESSING

**PII and sensitive information**: identify and classify sensitive columns (Data Governance handbook Part 14) *before* preprocessing begins - a preprocessing pipeline that logs sample rows for debugging, or that stores intermediate DataFrames on disk, can inadvertently create additional unprotected copies of sensitive data if this isn't done deliberately upfront. **Credential protection**: database connection strings and API keys used during data ingestion (Section 56) must never be hardcoded (Module 8 Section 72; Data Engineering handbook Part 45). **SQL injection**: any SQL query built via string concatenation from user- or pipeline-variable input (rather than parameterized queries, Module 4 Section 34) is vulnerable - relevant wherever preprocessing code dynamically constructs queries. **Unsafe deserialization**: Section 61's exact warning - loading an untrusted serialized pipeline can execute arbitrary code. **Malicious files**: a preprocessing pipeline that ingests files from external/untrusted sources (uploaded CSVs, scraped data) should validate file structure and size *before* fully parsing, since a maliciously crafted file can be used to trigger excessive memory consumption or exploit a parsing library vulnerability. **Data poisoning**: an attacker deliberately injecting corrupted or biased records into a pipeline that will retrain a model (Module 8 Section 92) - a preprocessing-stage validation gate (Section 64) that flags statistically anomalous incoming data is a meaningful, if partial, defense. **Adversarial data**: inputs specifically crafted to exploit a *particular* preprocessing step's weakness (e.g. a string engineered to break a regex-based cleaning function, Section 28, in a way that produces a mis-encoded or crash-inducing downstream value).

**Connecting preprocessing to ML security broadly**: preprocessing is a genuine, often-overlooked attack surface - it is the first code path untrusted external data touches, making its input validation (Section 4, 64) as security-relevant as a web application's input sanitization, not merely a data-quality nicety.

# SECTION 69 - FINANCIAL DATA

**Transaction amounts**: typically right-skewed, favoring log transformation (Section 12) or robust scaling (Section 24); always validate non-negativity where appropriate (refunds/reversals may be legitimately negative - domain knowledge determines the correct range check, Section 10). **Currencies**: multi-currency data requires explicit normalization to one base currency using the *correct historical* exchange rate for each transaction's date (using today's rate for a transaction from two years ago is a subtle but real correctness bug). **Dates**: transaction timestamps often carry timezone ambiguity across a multi-region financial system (Section 31) - critical to resolve correctly before any time-based feature engineering (Section 45-46). **Missing financial records**: often MNAR or structurally missing (Section 5) - a missing transaction record might mean "no transaction occurred" (a real, meaningful zero) rather than "data is unavailable," a distinction with real modeling consequences. **Extreme transactions**: Section 11's outlier discipline applies with particular force here - an unusually large transaction is very often exactly the pattern a fraud or AML (anti-money-laundering) model exists to catch, not noise to be clipped away. **Fraud labels**: frequently delayed and noisy (Module 8 Section 7) - a "confirmed fraud" label may only be finalized weeks after the transaction, directly shaping what features can legitimately be used (Section 35, 45-46's temporal-leakage discipline applied with maximum rigor given the regulatory and financial stakes).

# SECTION 70 - HEALTHCARE DATA

**Clinical measurements** (blood pressure, lab values): require explicit clinical-plausibility range validation (Section 10) - values outside physiologically possible ranges are almost certainly data-entry or device errors, not genuine extreme observations. **Missing lab values**: frequently MNAR (Section 5) - a lab test *not ordered* is fundamentally different from a lab test ordered but whose result is missing, and conflating the two during imputation can introduce serious, clinically-relevant bias. **Categorical diagnosis codes** (ICD-10 and similar): extremely high-cardinality (Section 14, 20) with a meaningful hierarchical structure (a code's first characters often indicate a broader diagnosis category) that naive encoding discards - hierarchy-aware encoding (grouping to a broader category level before one-hot/frequency encoding) is often more effective than encoding every granular code independently. **Timestamps**: encounter/visit timestamps require the same rigor as Section 31, with particular care around multi-day inpatient stays where "the prediction time" must be unambiguous (Module 8 Section 6's unit-of-observation discipline). **Patient identifiers**: must be handled per Section 68's PII discipline throughout, and must drive patient-level (not row-level) train/test splitting (Module 8 Section 24's grouped-splitting principle) whenever a patient contributes multiple records, to avoid the same patient's data leaking across the train/test boundary. **Leakage and privacy discussed together deliberately**: healthcare preprocessing is where these two concerns are most tightly coupled - a leaked feature (Section 35) in a healthcare model isn't merely a modeling error but can also represent an inappropriate use of protected health information, making rigorous leakage review here simultaneously a technical and an ethical/regulatory obligation (Data Governance handbook Part 19's HIPAA discussion applies directly).

# SECTION 80 - AT LEAST 50 TROUBLESHOOTING SCENARIOS

Representative set of 12, extensible to the full 50. Each: Diagnosis -> Solution.

**1. `OneHotEncoder` fails on unseen categories.** *Diagnosis*: `handle_unknown` was left at its error-raising default. *Solution*: refit with `handle_unknown="ignore"` (Section 15).

**2. Sparse/dense memory explosion.** *Diagnosis*: a high-cardinality one-hot-encoded feature was forced to dense (`sparse_output=False`), consuming excessive memory. *Solution*: keep sparse output for high-cardinality categoricals, or switch to frequency/hashing encoding (Sections 18, 20).

**3. Shape mismatch at prediction time.** *Diagnosis*: the inference DataFrame has different columns (missing, extra, or reordered) than what the pipeline was fit on. *Solution*: always pass raw input through the exact same fitted pipeline object (Section 60), and validate schema (Section 4) before calling `.predict()`.

**4. NaN appears after a transformation that shouldn't produce it.** *Diagnosis*: often a log transform applied to a zero or negative value (Section 12), or a ratio with a zero denominator (Section 43). *Solution*: guard the specific operation explicitly; audit for `inf`/`-inf` as well as `NaN`.

**5. Scaler throws an error on new data.** *Diagnosis*: the new data contains a column the scaler wasn't fit on, or a dtype mismatch. *Solution*: verify `ColumnTransformer` column lists exactly match between fit and transform-time data.

**6. Invalid dates crash the pipeline.** *Diagnosis*: `pd.to_datetime` called without `errors="coerce"`. *Solution*: add `errors="coerce"` and explicitly handle the resulting `NaT` values (Section 10, 31).

**7. Infinite values appear in a feature.** *Diagnosis*: an unguarded division (Section 43) or an unguarded log of zero. *Solution*: audit every ratio/log feature for the zero/negative edge case explicitly.

**8. An unexpected category appears in production that passed validation in training.** *Diagnosis*: the schema's `allowed_values` (Section 4) wasn't kept in sync with the real, evolving category set. *Solution*: treat this as a data-contract violation (Data Governance handbook Part 30) requiring producer-side coordination, not just a pipeline patch.

**9. A `Pipeline` step throws `NotFittedError`.** *Diagnosis*: `.transform()` was called before `.fit()`, often from loading a wrongly-serialized or partially-constructed pipeline object. *Solution*: verify the loaded artifact is the fully fitted pipeline, not a fresh, unfitted instance.

**10. Preprocessing works in a notebook but fails in an API.** *Diagnosis*: exactly Section 60's warning - separately reimplemented logic, or a missing dependency/version mismatch between the notebook and API environments. *Solution*: serialize and reuse the exact fitted pipeline (Section 61); pin dependencies identically across both environments (Section 67).

**11. The model expects a different feature order than what's being passed.** *Diagnosis*: manual feature construction outside a `ColumnTransformer`/`Pipeline`, relying on column position rather than name. *Solution*: always use named `ColumnTransformer` column selections (Section 38), never positional indexing, for exactly this reason.

**12. Production data contains missing columns entirely.** *Diagnosis*: an upstream schema change (Data Governance handbook Part 43) or an incomplete API request. *Solution*: schema validation (Section 4, 64) at the inference boundary, failing loudly and clearly rather than letting the pipeline silently proceed with absent data.

```python
# [Python] The comparative benchmark this section's discussion is built on
import time

def benchmark(label, fn):
    start = time.time()
    result = fn()
    print(f"{label}: {time.time() - start:.4f}s")
    return result

# Pandas vectorized vs. a Python loop (Module 4 Section 39's core lesson, reapplied to preprocessing)
benchmark("Pandas vectorized log", lambda: np.log1p(df["amount"]))
benchmark("Python loop", lambda: [np.log1p(x) for x in df["amount"]])

# Pandas vs. Polars for a groupby aggregation
benchmark("Pandas groupby", lambda: df.groupby("customer_id")["amount"].agg(["sum", "mean"]))
# benchmark("Polars groupby", lambda: pl.from_pandas(df).group_by("customer_id").agg([...]))

# CSV vs. Parquet load time (Module 4 Section 36)
benchmark("CSV load", lambda: pd.read_csv("large_file.csv"))
benchmark("Parquet load", lambda: pd.read_parquet("large_file.parquet"))
```

**Comparing runtime, memory, and scalability across Pandas, NumPy, scikit-learn, Polars, SQL, and Spark**: NumPy vectorized operations are the fastest option for pure numerical array computation, with Pandas adding labeled-axis overhead on top of the same underlying NumPy operations; scikit-learn's transformers are generally well-optimized (often NumPy-vectorized internally) but add object-oriented overhead appropriate for pipeline composability, not raw speed; Polars (Section 59) typically outperforms Pandas substantially on the same single machine, especially for groupby/aggregation-heavy workloads, due to its multi-threaded, Rust-based execution; SQL/database engines (Section 54) are unbeatable for aggregation and filtering at genuinely large scale, since the computation never has to leave the database at all; Spark (Section 58) trades single-operation speed for horizontal scalability across a cluster, becoming the better choice only once data genuinely exceeds single-machine capacity (Section 57). The professional takeaway, restated one final time: benchmark on *your* actual data and operations before choosing a tool based on general reputation - the right choice depends on data size, operation type, and existing infrastructure, not a universal ranking.

# SECTION 85 - REAL-WORLD SCENARIOS

Full worked answers for each of the brief's six named scenarios.

**1. "Your model works perfectly in a notebook but fails in production. Log the intermediate output of each pipeline stage on a real production request and compare against the same stage's output on a known-good notebook example.

**2. "You have 100 million records. How would you preprocess them?"** Push as much filtering and aggregation as possible into SQL/the warehouse first (Section 54, 56), minimizing what reaches Python at all. Use Parquet with column pruning for whatever does need to be pulled (Section 57, Module 4 Section 36). If the remaining data still doesn't comfortably fit in memory, use chunked processing (Section 57) or escalate to Polars (Section 59) for single-machine performance, or Spark (Section 58) if genuinely distributed processing is warranted. Profile first at every step - don't assume distributed processing is needed before confirming simpler options (SQL push-down, Parquet, chunking) are actually insufficient.

**3. "Your categorical feature contains 500,000 unique values. What would you do?"** Rule out that this is actually an ID-like column that should be excluded entirely (Section 14) rather than genuinely encoded. If it's a real categorical signal, avoid one-hot encoding entirely (Section 15's cardinality limit) in favor of frequency encoding (Section 18), leakage-safe out-of-fold target encoding (Section 19), or hashing-based encoding (Section 20) - the right choice depends on whether interpretability (favoring frequency/target encoding) or robust handling of entirely novel categories at inference (favoring hashing) matters more for this specific use case.

**4. "Your training data contains no missing values, but production data contains 20% missing values. **5. "Your fraud model performs extremely well after SMOTE. How do you determine whether the improvement is legitimate?"** Verify SMOTE was applied strictly to the training fold only, inside cross-validation (Section 50, 88) - if it was applied before splitting (Section 35's example 6), the "improvement" is leakage, not genuine skill. Evaluate on a genuinely held-out, un-resampled test set reflecting the real production class balance (never evaluate against SMOTE-resampled validation data). Check PR-AUC and precision/recall specifically (not just accuracy or even ROC-AUC alone, Module 8 Section 56-57) since SMOTE's apparent benefit can look inflated on a metric insensitive to the real-world imbalance the model will actually face.

**6. "Your preprocessing pipeline takes 30 seconds per API request. How would you optimize it?"** Profile the pipeline to find the actual bottleneck (Section 81) rather than guessing - commonly an expensive on-the-fly aggregation feature (moving that computation to a pre-computed feature store, Section 89, instead of recomputing per-request) or an inefficient custom transformer (Section 40) doing unnecessary work. Ensure the pipeline object is loaded once at service startup, never per-request (Section 63's core warning). Consider whether any step (e.g. a KNN-based imputer, Section 8) is fundamentally unsuitable for real-time latency and should be replaced with a faster equivalent (simple imputation) for the online path specifically, even if a heavier version remains appropriate for batch/offline preprocessing (Section 62).

# SECTION 86 - COMPLETE SCIKIT-LEARN PIPELINE

```python
# [Python/scikit-learn] The enterprise-style pipeline, every component named and explained
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, PowerTransformer, OneHotEncoder
from sklearn.ensemble import GradientBoostingClassifier

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),          # Section 6
    ("transformer", PowerTransformer(method="yeo-johnson")),   # Section 13 - reshape skew before scaling
    ("scaler", StandardScaler()),                                # Section 22
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),     # Section 6
    ("encoder", OneHotEncoder(handle_unknown="ignore")),         # Section 15
])

preprocessor = ColumnTransformer([
    ("numerical", numeric_pipeline, numeric_cols),
    ("categorical", categorical_pipeline, categorical_cols),
])

enterprise_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", GradientBoostingClassifier(random_state=42)),
])
```

```
Raw Data -> Train/Test Split -> ColumnTransformer
                                    |-- Numerical Pipeline: Imputer -> Transformer -> Scaler
                                    `-- Categorical Pipeline: Imputer -> Encoder
                                 -> Feature Engineering -> Model
```

**Every component explained**: the numeric branch imputes first (Section 6 - a `PowerTransformer` or `StandardScaler` cannot handle missing values), then reshapes distribution (Section 13 - correcting skew *before* standardizing, since standardization alone doesn't address non-normal shape), then scales (Section 22); the categorical branch imputes (with a sentinel/mode strategy appropriate for categories, Section 6) then encodes (Section 15); both branches run in parallel inside the `ColumnTransformer` (Section 38) and are concatenated automatically; the whole preprocessing stage is wrapped with the model in one final `Pipeline` (Section 39), guaranteeing the fit-on-train-only discipline (Section 33) is mechanically enforced end to end, with feature engineering (Sections 41-47) inserted as an additional stage wherever domain-specific derived features are needed beyond what generic preprocessing alone provides.

```python
# [Python/scikit-learn]
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV

param_grid = {
    "preprocessor__numerical__imputer__strategy": ["median", "mean"],     # TUNING A PREPROCESSING CHOICE
    "preprocessor__categorical__encoder__handle_unknown": ["ignore"],
    "model__n_estimators": [100, 200, 300],                                 # TUNING A MODEL HYPERPARAMETER
    "model__max_depth": [3, 5, 7],
}

grid_search = GridSearchCV(enterprise_pipeline, param_grid, cv=5, scoring="roc_auc", n_jobs=-1)
grid_search.fit(X_train, y_train)
print(grid_search.best_params_)
```

**Why preprocessing must be included within cross-validation, not fit once beforehand**: the double-underscore parameter naming (`preprocessor__numerical__imputer__strategy`) reaches *inside* the pipeline's nested structure, letting `GridSearchCV` treat a preprocessing *choice* (which imputation strategy) exactly like any model hyperparameter - refitting the *entire* pipeline, preprocessing included, fresh on each fold's training portion for every single parameter combination tried. If preprocessing were instead fit once on the full training set *before* the grid search began, every fold's cross-validation score would be computed against test-like validation folds already influenced by preprocessing statistics derived partly from those same folds - a leakage bug identical in spirit to Section 33-35, just occurring inside the tuning loop rather than the initial train/test split. Passing the *whole pipeline object* to `GridSearchCV`/`RandomizedSearchCV`, as shown, is what closes this gap automatically.

# SECTION 89 - FEATURE STORE CONCEPTS

| Concept | Definition |
|---|---|
| Offline features | Computed in batch, typically from a warehouse, used for model training and batch inference (Section 62) |
| Online features | Served with low latency for real-time inference (Section 63), often pre-computed and cached rather than calculated per-request |
| Feature transformations | The logic converting raw data into a feature - ideally defined *once* and shared between offline and online computation paths |
| Point-in-time correctness | Every training example's features reflect only data available strictly before that example's label timestamp - the feature-store-level enforcement of Section 45-46's temporal-leakage discipline, applied systematically rather than per-pipeline |
| Training-serving consistency | Section 60's principle, elevated to a platform-level guarantee - the same feature *definition* computes the same value whether requested for training or live inference |
| Feature freshness | How current an online feature's underlying data is - a real operational property monitored continuously (Section 91) |

**Connecting preprocessing to modern ML platforms**: a feature store is, at its core, an infrastructure answer to two of this module's hardest recurring problems - **point-in-time correctness** (Sections 45-46's leakage discipline, enforced systematically by the platform rather than manually re-verified in every individual pipeline) and **training-serving consistency** (Section 60, enforced by defining a feature's computation logic exactly once and serving it identically to both training jobs and live inference requests, rather than risking the notebook-vs-production divergence Section 85's first scenario describes). Understanding feature stores conceptually - even without operating one directly - is what lets you recognize *which* of this module's manual disciplines (careful `as_of` filtering, careful pipeline serialization) a feature store is designed to make structurally difficult to get wrong, rather than merely convenient.

# SECTION 91 - MONITOR TRANSFORMATIONS

```python
# [Python] A preprocessing-specific monitoring pass, run on every new batch of production data
def monitor_preprocessing(df_production, training_stats: dict) -> dict:
    report = {}

    report["missingness"] = {
        col: {"production": df_production[col].isna().mean(), "training": training_stats["missingness"][col]}
        for col in df_production.columns
    }

    report["category_frequencies"] = {
        col: dict(df_production[col].value_counts(normalize=True))
        for col in df_production.select_dtypes(include=["object", "category"]).columns
    }

    report["numerical_ranges"] = {
        col: {"min": df_production[col].min(), "max": df_production[col].max(), "training_range": training_stats["ranges"][col]}
        for col in df_production.select_dtypes(include="number").columns
    }

    report["transformation_failures"] = int(df_production.isin([np.inf, -np.inf]).sum().sum())   # infinities from unguarded ops (Section 43)

    return report
```

Monitoring **missingness** (has a column's null rate shifted meaningfully from training - an early warning of an upstream schema/quality regression, Section 80); **category frequencies** (has a new category appeared, or has an existing category's prevalence shifted dramatically - Module 8's PSI-based approach, Section 86, applies directly to categorical preprocessing inputs too); **numerical ranges** (are production values falling outside the range training data ever contained - a direct signal that scaling/transformation parameters learned from training may no longer be well-suited); **scaling behavior** (are scaled feature outputs drifting outside their expected roughly-standardized range); **transformation failures** (a count of `inf`/unexpected-`NaN` values produced by preprocessing itself, Section 80's troubleshooting patterns, indicating a guard that's failing silently); **distribution changes** (the full statistical drift-testing toolkit from Module 8 Section 86, applied specifically to preprocessing-stage inputs rather than only final model features). Preprocessing-stage monitoring, run continuously and automatically, is what catches Section 92's incident scenarios *before* they degrade model performance silently, often days or weeks before Module 8's Section 88 delayed-label performance monitoring would otherwise reveal the same underlying problem.

# SECTION 92 - PRODUCTION INCIDENT CASE STUDIES

Each follows: Detection -> Diagnosis -> Root cause -> Remediation -> Prevention.

**Incident 1 - A new category breaks the encoder.** *Detection*: an inference request fails, or (if `handle_unknown="ignore"` was correctly set, Section 15) a valid-looking request silently produces an all-zero-encoded feature with no error at all. *Diagnosis*: compare the failing/suspicious request's category value against the encoder's known category list. *Root cause*: an upstream source system introduced a new category value (a new payment method, a new product line) with no advance notice. *Remediation*: confirm `handle_unknown="ignore"` is set to prevent hard failures; assess whether the model needs retraining to properly learn from the new category rather than silently zero-encoding it indefinitely. *Prevention*: a data contract (Data Governance handbook Part 30) with the producing team, plus category-frequency monitoring (Section 91) that flags a genuinely new category the moment it first appears.

**Incident 2 - A production feature becomes null.** *Detection*: Section 91's missingness monitoring flags a sharp spike for one specific column. *Diagnosis*: trace the feature back through the pipeline to its raw source. *Root cause*: an upstream ETL job silently started failing to populate that specific field, or a joined table's schema changed. *Remediation*: patch the upstream pipeline; in the interim, confirm the model's imputation strategy (Section 6) degrades gracefully rather than catastrophically for this specific feature. *Prevention*: per-feature freshness/completeness monitoring (Section 91) with automated alerting, not just periodic manual review.

**Incident 3 - Feature order changes.** *Detection*: Section 80's "model expects a different feature order" symptom - predictions become nonsensical without any explicit error. *Diagnosis*: compare the exact column order the pipeline was fit on against the order of the DataFrame now being passed to it. *Root cause*: a manual, positional (rather than named, `ColumnTransformer`-based) feature-construction step was changed elsewhere in the codebase. *Remediation*: replace positional feature construction with named `ColumnTransformer` column selection (Section 38) everywhere in the codebase. *Prevention*: pipeline-level unit tests (Section 65) asserting output shape and column identity on every code change.

**Incident 4 - Training and serving use different transformations.** *Detection*: a persistent, unexplained gap between offline validation metrics and live production performance. *Diagnosis*: directly compare the training pipeline's code/artifact against the serving code's actual preprocessing logic, line by line. *Root cause*: Section 60's exact warning - separately maintained training and inference preprocessing code drifted apart over time. *Remediation*: immediately migrate serving to load and reuse the exact fitted training pipeline artifact (Section 61), eliminating the separate implementation entirely. *Prevention*: architecturally forbid any preprocessing logic from existing outside the single, versioned, serialized pipeline object.

**Incident 5 - A database schema changes.** *Detection*: schema validation (Section 4, 64) fails loudly on the next scheduled data pull, or - worse, if validation was skipped - a silent type coercion or null-filling corrupts a feature undetected. *Diagnosis*: compare today's extracted schema against yesterday's recorded schema (Module 4 Section 43-style diff). *Root cause*: an upstream team deployed a database migration without coordinating with this pipeline's consumers. *Remediation*: patch the pipeline to the new schema; if data was silently corrupted before detection, assess and potentially retrain on cleaned data. *Prevention*: a data contract (Data Governance handbook Part 30, 42-43) with CI-enforced schema-compatibility checks on the producing team's side.

**Incident 6 - A preprocessing dependency upgrade changes model behavior.** *Detection*: after a routine `scikit-learn`/`pandas` version upgrade, retrained model performance shifts unexpectedly, or previously-passing tests (Section 65-66) start failing. *Diagnosis*: bisect the dependency upgrade - retrain against the old vs. new library version on identical data and compare outputs directly. *Root cause*: a library's internal implementation of a transformer or algorithm changed subtly between versions (a genuinely common occurrence, Module 8 Section 69). *Remediation*: pin the working dependency version until the behavior change is fully understood and validated; update version-pinning documentation. *Prevention*: pinned dependency versions as a standing practice (Section 67), with any upgrade treated as a deliberate, tested change - never an unreviewed, automatic update in a production environment.

# SECTION 97 - NUMERICAL FEATURE DECISION TREE

```
Numerical Feature
   -> Missing? -- Yes --> choose an imputation strategy (Section 6): mean/median/mode/constant/interpolation/model-based,
   |                       matched to the missingness mechanism (Section 5); add a missingness indicator if plausibly MNAR (Section 7)
   -> No
       -> Skewed? -- Yes --> consider a transformation: log (Section 12) if strictly positive and simplicity is valued,
       |                      Box-Cox (Section 13) if strictly positive and normality is the goal,
       |                      Yeo-Johnson (Section 13) if zero/negative values are present
       -> No
           -> Outliers present? -- Yes --> decide TREATMENT separately from detection (Section 11):
           |                                clip/winsorize if extremes are real but should be bounded,
           |                                remove only if confirmed invalid (Section 10),
           |                                keep if the extremes are exactly the signal the model needs (fraud/anomaly contexts)
           -> Scale required by the chosen model family? -- Yes (KNN/SVM/linear/NN/PCA, Section 21) -->
           |     StandardScaler (Section 22, default) / RobustScaler (Section 24, if outliers were kept) / MinMaxScaler (Section 23, if a bounded range is required)
           -> No (tree-based model, Section 21) --> scaling not required, safe to skip
   -> Final transformed, model-ready numerical feature
```

# SECTION 98 - CATEGORICAL FEATURE DECISION TREE

```
Categorical Feature
   -> Nominal or Ordinal? (Section 14)
       -> Ordinal (genuine, domain-agreed order exists) --> OrdinalEncoder with explicit category order (Section 16)
       -> Nominal (no meaningful order) --> continue below
   -> Cardinality? (Section 14)
       -> Low (a handful of distinct values) --> OneHotEncoder (Section 15)
       -> High (hundreds to thousands) --> Frequency encoding (Section 18) or out-of-fold Target encoding with smoothing (Section 19)
       -> Very high (hundreds of thousands+) --> Hashing-based encoding (Section 20), or reconsider whether this is really an ID column to exclude (Section 14)
   -> Rare categories present? (Section 14) -- Yes --> group below a frequency threshold into "Other" BEFORE encoding,
   |                                                     fit on training data only (Section 33)
   -> Unknown categories possible at inference? (essentially always, in production) -- Yes -->
   |     OneHotEncoder: handle_unknown="ignore" (Section 15)
   |     OrdinalEncoder: handle_unknown="use_encoded_value", unknown_value=-1 (Section 16)
   |     Frequency/Target encoding: map unseen categories to 0 / the global mean (Sections 18-19)
   |     Hashing encoding: unseen categories automatically hash into an existing bucket, no special handling needed (Section 20)
   -> Final encoded, model-ready categorical feature
```

Walking any new categorical feature through both this decision tree and Section 97's numerical tree, in order, before writing a single line of encoding code, is the concrete, repeatable habit this entire module exists to instill - replacing "which encoder did I use last time" with a deliberate, justified decision for *this specific feature's* actual properties.

# SECTION 104 - PREPARATION FOR MODULE 11: LINEAR REGRESSION

```
Python -> NumPy -> Pandas -> Mathematics -> Probability -> Statistics -> ML Workflow -> EDA -> Data Preprocessing -> LINEAR REGRESSION
```

**How preprocessing decisions directly affect Linear Regression, the very next module**: **numerical scaling** (Sections 21-25) - linear regression's coefficients are directly scale-dependent, and regularized variants (Ridge/Lasso, Module 11) penalize coefficients in a way that is *meaningless* unless features are comparably scaled first, making Section 22's `StandardScaler` close to mandatory rather than optional for regularized linear models specifically. **Categorical encoding** (Sections 14-20) - linear regression requires fully numeric input, and the encoding choice directly determines what relationship the model can express (one-hot's independent per-category coefficients vs. ordinal's single monotonic coefficient, Section 16). **Missing values** (Sections 5-8) - linear regression cannot handle missing input at all without imputation first. **Outliers** (Section 11) - linear regression's least-squares fitting is notably sensitive to outliers (a single extreme point can substantially shift the fitted line), making Section 11's detection-and-treatment discipline unusually consequential here. **Transformations** (Sections 12-13, 27) - linear regression assumes a linear relationship between features and target; a transformation that makes that relationship more linear (or that normalizes a skewed target, Module 8 Section 18, 103) can directly improve model fit and validity of its statistical assumptions. **Multicollinearity** (Module 8 Section 17, directly relevant to PCA, Section 49) - highly correlated input features make linear regression's coefficients unstable and difficult to interpret, a problem PCA or careful feature selection (Section 48) can address. **Polynomial features** (Section 47) - the direct mechanism for letting a fundamentally linear model capture non-linear relationships. **Regularization** (introduced in Module 11 itself, but directly dependent on Section 22's scaling being done correctly first). **Target transformation** (Section 12-13, 34's principles, applied to the target variable itself - a log-transformed target is extremely common in linear regression specifically, given the model's linear/additive assumptions).

This module's entire purpose - producing clean, validated, correctly-encoded, correctly-scaled, leakage-safe, reproducible, model-ready data - was never an end in itself. It is the necessary, non-negotiable foundation every subsequent modeling module in this curriculum, starting immediately with Module 11's Linear Regression, depends on completely: a mathematically perfect regression algorithm applied to poorly preprocessed data will still fail, while even a simple algorithm applied to properly preprocessed data has a genuine chance to succeed - the professional discipline this module teaches is what makes that difference reliably, repeatably true.
