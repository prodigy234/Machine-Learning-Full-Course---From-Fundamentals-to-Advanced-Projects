# SECTION 1 - WHAT IS A MACHINE LEARNING WORKFLOW?

**Definition**: an ML workflow is the complete, ordered set of activities - problem framing, data work, modeling, evaluation, deployment, and monitoring - required to turn a business need into a reliable, maintained prediction system in production. It is not a single script; it is a *process* with feedback loops.

**Why ML is a process, not a single model**: a trained model is one artifact produced partway through this process - it has no value in isolation until it's validated, deployed, monitored, and eventually retrained as the world changes. Treating "train a model" as the whole job is the single most common misconception separating a hobbyist from a professional practitioner.

**Training a model vs. building an ML system**: training a model is a bounded technical exercise (fit an estimator to data); building an ML system additionally requires data pipelines, validation, versioning, deployment infrastructure, monitoring, and a retraining strategy - the model is maybe 10-20% of the total engineering effort in a real system.

**Research workflow vs. production workflow**: research prioritizes exploring what's possible - flexible notebooks, many experiments, less concern for latency or maintainability. Production prioritizes reliability, reproducibility, and maintainability - tested modules, versioned artifacts, monitored behavior, and graceful failure. Both are legitimate; confusing them (shipping research-quality code to production, or over-engineering an early exploration) is a recurring professional mistake (Section 110).

**Notebook workflow vs. engineering workflow**: notebooks are excellent for exploration (EDA, quick iteration, visualization) precisely because they're informal and stateful; that same informality (hidden state, out-of-order execution, no tests) is exactly what makes notebooks dangerous as a *deployment* artifact. Section 71 covers the disciplined transition from one to the other in full.

**The complete lifecycle**

```
PROBLEM -> DATA -> UNDERSTANDING -> PREPARATION -> MODELING -> EVALUATION -> DEPLOYMENT -> MONITORING -> RETRAINING
```

This is a **loop**, not a line: monitoring findings (Sections 85-89) feed back into retraining (Section 90), which restarts modeling; a business-problem redefinition can restart the entire chain. Every section in this module maps onto one node in this loop, and by the end you will be able to walk any real project through it explicitly.

# SECTION 2 - BUSINESS PROBLEM DEFINITION

**Concept**: every legitimate ML project starts from a business problem, not an algorithm or a dataset. **Why it matters**: a technically excellent model solving the wrong problem produces zero business value and often gets shipped anyway because "the metrics look good" - the single most expensive failure mode in applied ML.

**Vocabulary**: **business objective** (what the organization is trying to achieve - "reduce customer attrition"); **business KPI** (the measurable metric tied to that objective - "monthly churn rate"); **business constraint** (limits on the solution - budget, latency, interpretability, regulatory requirements); **business stakeholder** (who owns the decision and will act on the prediction); **decision to be supported** (what specific action changes based on the model's output - if no decision changes, the model has no value); **cost of errors** (what a false positive/negative actually costs the business, Section 95); **value of prediction** (the upside when the model is right, weighed against its cost to build and run).

**Worked example**: "Customers are leaving" is a business observation, not a problem statement - it doesn't say who, when, or what action follows. Converted into an actionable ML-adjacent statement: **"Predict which customers are likely to churn within the next 30 days, so the retention team can prioritize outreach to the highest-risk accounts before their contract renewal."** This single sentence now implies a target (churn within 30 days), a unit of observation (customer), a prediction horizon (30 days), and a decision (retention outreach prioritization) - everything Sections 5-7 formalize.

**Why the distinction matters**: without this translation step, a team can spend months building a technically sound model that predicts the wrong thing, at the wrong horizon, for a decision nobody is actually prepared to act on - discovered only after deployment, when it's most expensive to fix.

# SECTION 4 - SHOULD YOU USE MACHINE LEARNING?

**Decision framework** - ask, in order: **Is there enough data?** (ML needs a meaningful volume of historical examples with outcomes; a handful of cases favors rules or expert judgment). **Is the target measurable?** (if you can't define and record ground truth, you can't train or evaluate a model). **Is the problem repetitive?** (ML earns its cost through repeated application; a one-off decision rarely justifies it). **Is prediction actually useful?** (does knowing the answer earlier or more precisely change an outcome). **Can the prediction influence a decision?** (Section 2 - no actionable decision, no value). **Is the cost of ML justified** relative to a simpler alternative? **Can a rule-based system solve it?** (a handful of clear, stable business rules often outperforms ML in cost, transparency, and maintainability for simple, well-understood problems). **Is there a statistical solution** (a simple regression or even a single well-chosen average) that captures most of the value at a fraction of the complexity? **Is human judgment required** for cases involving ethics, rare high-stakes edge cases, or contexts a model cannot reasonably be trained on?

**When ML should NOT be used**: the problem is simple enough for deterministic rules; there's too little historical data to learn a reliable pattern; the underlying relationship changes too fast for any trained model to stay current (Section 87); the cost of an error is catastrophic and unrecoverable with no room for a probabilistic system's inherent error rate; regulatory or ethical requirements demand fully explainable, auditable logic rules alone can provide; or, most simply, nobody can articulate what decision the prediction would actually change.

| Element | Churn example |
|---|---|
| Target variable | `will_churn` (binary: cancels within the outcome window) |
| Prediction horizon | 30 days into the future |
| Unit of prediction | One customer |
| Observation window | The trailing 90 days of behavior used to build features |
| Outcome window | The next 30 days, in which the target is determined |
| Prediction frequency | Weekly, for every active customer |
| Business objective | Reduce attrition through proactive retention outreach |
| ML objective | Rank customers by predicted churn probability within 30 days |

**The four defining questions**: **WHO** is the prediction about (unit of observation, Section 6)? **WHEN** is the prediction made (the point in time features are computed as of)? **WHAT** exactly is being predicted (the precise target definition, Section 7)? **HOW FAR into the future** does the outcome window extend (the prediction horizon)? Every one of these four questions must have a crisp, written-down answer before any data work begins - vague answers here are the root cause of a large share of the leakage problems in Section 19, because "when" and "how far" are exactly what temporal leakage violates.

# SECTION 6 - DEFINING THE UNIT OF OBSERVATION

| Level | Example use case |
|---|---|
| Customer-level | Churn, lifetime value |
| Transaction-level | Fraud detection |
| Account-level | Credit risk on a specific account/loan |
| Device-level | Predictive maintenance, IoT anomaly detection |
| Patient-level | Healthcare risk prediction |
| Product-level | Demand forecasting, pricing |
| Session-level | Web conversion, recommendation |

**Why an incorrect observation unit causes serious problems**: choosing the wrong grain corrupts everything downstream. If a fraud model is trained at *customer* level ("is this customer ever fraudulent") when the real business decision is made per *transaction* ("should this specific transaction be blocked"), the model cannot support the actual decision at all - it answers a different question than the one that matters. Getting the grain wrong also creates data leakage risk (aggregating information across the wrong boundary) and evaluation problems (a model can look excellent at one grain and be useless at the grain decisions are actually made). This is the ML-workflow analog of dimensional modeling's "grain" concept from the Data Warehousing handbook - the discipline is identical: **state the grain explicitly, in one sentence, before doing anything else.**

# SECTION 7 - TARGET VARIABLE DESIGN

| Target type | Example |
|---|---|
| Binary | `is_fraud` (0/1) |
| Multiclass | `issue_category` (billing/technical/other) |
| Continuous | `predicted_revenue` |
| Count | `number_of_support_calls_next_month` |
| Time-to-event | `days_until_equipment_failure` |

**Label quality concerns, each explained**: **label delay** - the true outcome isn't known until well after the prediction is needed (a loan's default status may only be knowable 12+ months later), forcing a choice between waiting for clean labels or training on necessarily incomplete/proxy labels; **label leakage** - the label itself, or how it was recorded, encodes information that wouldn't be available at prediction time (Section 19); **ambiguous labels** - the target's definition has edge cases nobody has resolved (does a customer who downgrades but doesn't cancel count as "churned"?); **noisy labels** - the recorded label is sometimes simply wrong (a fraud label based on a customer dispute that was later proven incorrect). Every one of these must be resolved and documented *before* modeling - a model trained against an ambiguous, delayed, or noisy target cannot be meaningfully evaluated, let alone trusted in production.

# SECTION 8 - DATA REQUIREMENTS

**Determining requirements, systematically**: **required features** (what signals are plausibly related to the target, informed by domain knowledge); **required target** (how and where the ground-truth label is recorded); **required historical period** (enough history to capture the target's natural variation - seasonality, economic cycles, rare-event frequency); **required sample size** (enough positive examples specifically, not just enough rows - a 0.1% fraud rate needs a much larger total dataset than a 30% churn rate to get a usable number of positive cases); **required granularity** (does the decision need daily, hourly, or per-event resolution); **required update frequency** (how often new data becomes available, which bounds how often the model can realistically be refreshed).

# SECTION 12 - DATA QUALITY ASSESSMENT

Applying the six dimensions from the Data Governance handbook (Part 4) specifically to an ML dataset:

```python
# [Python] A data-quality report
def data_quality_report(df: pd.DataFrame, target: str) -> pd.DataFrame:
    report = pd.DataFrame({
        "dtype": df.dtypes,
        "missing_pct": df.isna().mean() * 100,
        "n_unique": df.nunique(),
        "is_constant": df.nunique() == 1,
    })
    report["completeness_ok"] = report["missing_pct"] < 20      # example threshold, tune per feature's stakes
    return report.sort_values("missing_pct", ascending=False)

print(data_quality_report(df, target="churn"))
```

**Accuracy** (does the value reflect reality - hard to check directly, often inferred from cross-source reconciliation); **completeness** (missing-value rate, above); **consistency** (does a customer's region agree across two source tables); **validity** (do values fall within expected domains - age between 0 and 120); **uniqueness** (no unintended duplicate entities); **timeliness** (is the data fresh enough relative to the prediction frequency, Section 5). A dataset failing badly on any one dimension should be flagged and, ideally, fixed at the source - not silently patched over in a notebook, which just relocates the problem to production later.

# SECTION 13 - DATA PROFILING

```python
# [Python] A reusable profiling function
def profile_dataset(df: pd.DataFrame) -> dict:
    profile = {}
    numeric_cols = df.select_dtypes(include="number").columns
    cat_cols = df.select_dtypes(include=["object", "category"]).columns

    profile["numeric_stats"] = df[numeric_cols].describe().T
    profile["categorical_distributions"] = {c: df[c].value_counts(normalize=True).head(10) for c in cat_cols}
    profile["missingness"] = df.isna().mean().sort_values(ascending=False)
    profile["duplicates"] = df.duplicated().sum()
    profile["cardinality"] = df.nunique().sort_values(ascending=False)
    profile["dtypes"] = df.dtypes

    q1, q3 = df[numeric_cols].quantile(0.25), df[numeric_cols].quantile(0.75)
    iqr = q3 - q1
    profile["outlier_counts"] = ((df[numeric_cols] < q1 - 1.5*iqr) | (df[numeric_cols] > q3 + 1.5*iqr)).sum()
    return profile

report = profile_dataset(df)
```

This function is deliberately the reusable core of the entire module - the same profiling logic reappears, applied to different datasets, in every end-to-end project (Sections 98-105) and the capstone (Section 128). Running it as the very first step of any new project catches a large share of Section 110's most common mistakes before they ever reach a model.

# SECTION 14 - EXPLORATORY DATA ANALYSIS FOR ML

**How EDA for ML differs from ordinary business analytics**: business analytics EDA typically asks "what happened" (revenue trends, regional breakdowns); **ML-focused EDA** asks a narrower, more pointed set of questions specifically aimed at what will make or break a model: **target distribution** (is it imbalanced, Section 18); **feature-target relationships** (which features actually seem informative, Section 16); **feature-feature relationships** (redundancy, multicollinearity, Section 17); **missingness patterns** (Section 31 - is data missing randomly or systematically); **outliers** (Section 35 - genuine signal or noise); **class imbalance** (Section 18, 23); and, critically, **leakage indicators** (Section 19 - does any feature look *suspiciously* predictive, a common leakage red flag). ML-focused EDA is deliberately narrower and more skeptical than general business EDA - every finding is evaluated through the lens of "how does this affect the model I'm about to build," not just "is this interesting."

# SECTION 19 - DATA LEAKAGE

**One of the most detailed sections in this module.**

**Definition**: data leakage occurs whenever information that would not genuinely be available at the real moment of prediction is used, directly or indirectly, to train or evaluate a model - producing performance that looks excellent offline and collapses in production.

| Leakage type | Definition |
|---|---|
| Target leakage | A feature is derived from, or is a close proxy for, the target itself |
| Train-test contamination | Any information from validation/test data influences training |
| Temporal leakage | A feature uses information from after the prediction point in time (Section 20) |
| Feature leakage | A feature encodes the outcome indirectly through an unrelated-looking column |
| Preprocessing leakage | A preprocessing statistic is fit using data outside the training set |
| Aggregation leakage | A group-level aggregate spans across the train/test or time boundary |
| Duplicate leakage | The same entity (or a near-duplicate) appears in both train and test |
| Human-process leakage | A feature reflects a human decision made *because of* the outcome, not before it |

**Representative examples (10 of the requested 30, each following WRONG WORKFLOW -> WHY IT LEAKS -> CORRECT WORKFLOW; extend using this identical structure)**:

1. **Fraud**: *Wrong* - include `was_chargeback_filed` as a feature for predicting `is_fraud`. *Why* - a chargeback is typically filed *after* fraud is confirmed; the feature is a near-restatement of the target. *Correct* - remove it; use only pre-transaction signals (device, velocity, amount).
2. **Churn**: *Wrong* - impute missing `total_charges` using the full dataset's mean before splitting. *Why* - the training-fit statistic now contains information from test-set rows. *Correct* - split first, fit the imputer on training data only (Module 4 Section 17, revisited fully in Section 31).
3. **Credit**: *Wrong* - use `account_closed_date` as a feature. *Why* - accounts close, in part, *because* of default - this column encodes the outcome. *Correct* - remove any post-decision account-status fields entirely.
4. **Healthcare**: *Wrong* - use `discharge_diagnosis` to predict `readmission_risk` computed at admission time. *Why* - the discharge diagnosis isn't known at the moment the prediction must actually be made. *Correct* - restrict every feature to information available strictly at admission.
5. **Energy forecasting**: *Wrong* - compute a rolling average using `.rolling()` without sorting by date first. *Why* - an unsorted rolling window can silently include future values. *Correct* - always `sort_values` by time before any window operation (Module 4 Section 22, 47).
6. **Marketing**: *Wrong* - a "customer responded to campaign" feature computed over the customer's *entire* history, used to predict response to a *specific* past campaign. *Why* - it includes responses to campaigns that happened after the one being predicted. *Correct* - compute strictly using data before each specific campaign's send date.
7. **Telecom**: *Wrong* - `customer_id` included as a raw feature in a model with many rows per customer, then randomly split. *Why* - the same customer's other rows appear in both train and test, letting the model partially "memorize" per-customer patterns rather than generalize. *Correct* - group-based splitting (Section 24).
8. **SaaS**: *Wrong* - a `days_until_cancellation` feature used to predict churn. *Why* - this value is only knowable once the customer has already cancelled - the ultimate target-leakage feature. *Correct* - remove entirely; it's definitionally derived from the outcome.
9. **Supply chain**: *Wrong* - `actual_delivery_date` used as a feature to predict `will_be_late`. *Why* - the actual delivery date is only known after delivery, the very event being predicted. *Correct* - use only planned/scheduled dates and information known at ship time.
10. **Cybersecurity**: *Wrong* - deduplicate a labeled anomaly dataset *after* splitting, allowing near-identical incident records to land in both train and test. *Why* - the model can effectively "look up" the answer for a near-duplicate test row it saw a copy of during training. *Correct* - deduplicate (Module 4 Section 18) before splitting, on a business key, not after.

(Continue this WRONG->WHY->CORRECT format across the module's other domains - recommendation, predictive maintenance, banking, oil & gas - to reach the full 30 examples.)

# SECTION 21 - DATA SPLITTING

**Vocabulary**: **train set** (used to fit model parameters); **validation set** (used to select/tune between candidates); **test set** (used exactly once, at the very end, for a final unbiased performance estimate).

| Split | Typical use |
|---|---|
| 70/30 (train/test) | Small datasets, simple model comparison, cross-validation used for tuning instead of a separate validation set |
| 80/20 (train/test) | The most common default for moderate-to-large datasets |
| 60/20/20 (train/validation/test) | When a genuinely separate validation set is preferred over cross-validation, common with very large datasets or expensive-to-train models |

**Why split ratios are context-dependent, not universal rules**: the right ratio depends on total dataset size (a 60/20/20 split on 500 rows leaves dangerously few test examples; the same ratio on 5 million rows is comfortable), class balance (a rare positive class may need a larger test set just to get enough positive examples to evaluate reliably), and whether cross-validation (Section 27) is being used for tuning (which reduces the need for a separate validation set, since CV effectively re-uses the training data more efficiently). There is no single correct ratio - state the ratio *and* the reasoning behind it explicitly in every project.

# SECTION 26 - TRAINING, VALIDATION, AND TESTING

```
TRAIN       -> LEARN PARAMETERS       (the model fits its internal weights/splits/coefficients here)
VALIDATION  -> SELECT / TUNE           (used to compare candidate models and hyperparameters - Sections 46, 50-51)
TEST        -> FINAL UNBIASED EVALUATION  (touched exactly ONCE, after everything else is finalized)
```

**Why the test set must be touched only once**: every time a modeling decision is made *in response to* test-set performance (picking a model because it scored best on test, tuning a hyperparameter because it improved the test score), the test set stops being an unbiased estimate of production performance - it has effectively become a second validation set, and its number is now optimistically biased. This exact discipline mistake - "test-set contamination" - is common enough to earn its own entry in Section 110's mistake catalog. The professional rule: make every decision (model choice, feature choice, hyperparameters) using train/validation or cross-validation only; compute the test score exactly once, at the very end, and report it as the honest estimate of what to expect in production.

**Outer loop -> model evaluation; inner loop -> hyperparameter tuning.** **Why this reduces optimistic evaluation bias**: if you tune hyperparameters using the *same* cross-validation folds you then report as your final performance estimate, you've implicitly let the test-like folds influence model selection (the same test-set-contamination problem from Section 26, just hidden inside a CV loop) - the reported score is biased optimistically because you picked the hyperparameters that happened to work best on exactly those folds. Nested CV keeps the outer loop's folds completely untouched by the tuning process (which only ever sees the inner loop's folds), giving a genuinely honest estimate of how the *entire tuning procedure* - not just one fixed model - would perform on new data. It's more expensive computationally but is the correct approach whenever you need to both tune hyperparameters *and* report a trustworthy final performance number from the same limited dataset.

**Why the mechanism matters**: MCAR is the easiest case - simple imputation introduces little bias. # SECTION 32 - CATEGORICAL ENCODING

| Method | How it works | When appropriate |
|---|---|---|
| One-hot encoding | One binary column per category | Nominal categories, low-to-moderate cardinality, linear/distance-based models |
| Ordinal encoding | Integer codes respecting a real order | Genuinely ordinal categories (Module 4 Section 15) |
| Label encoding | Arbitrary integer per category, no implied order | Only for tree-based models that can split on arbitrary codes without implying false ordinality; risky for linear models |
| Target encoding (conceptual) | Replace a category with the target's mean for that category | High-cardinality categoricals, but requires careful out-of-fold computation to avoid severe leakage |
| Frequency encoding | Replace a category with its occurrence frequency/count | High-cardinality categoricals where frequency itself is informative |
| Hashing (conceptual) | Map categories to a fixed number of buckets via a hash function | Very high cardinality, memory-constrained settings, some information loss (collisions) accepted |

**Target encoding's leakage risk, explained specifically**: replacing a category with the training-set target mean for that category, if done naively on the *same* rows used to compute it, directly leaks the target into a feature - the fix is to compute it out-of-fold (a different fold's target mean applied to each fold, cross-validation-style) or with added smoothing/noise, never a simple `.groupby().mean()` applied back onto the same rows it was computed from. `handle_unknown="ignore"` on the one-hot encoder is essential for production robustness (Section 78) - without it, a category never seen during training (a brand-new payment method added after deployment) crashes inference entirely rather than degrading gracefully.

# SECTION 40 - FEATURE ENGINEERING WITH SQL

```sql
-- [SQL] Window functions, CTEs, aggregation, and joins - feature engineering pushed to the database
WITH customer_history AS (
    SELECT customer_id, transaction_date, amount,
           ROW_NUMBER() OVER (PARTITION BY customer_id ORDER BY transaction_date DESC) AS rn,
           AVG(amount) OVER (PARTITION BY customer_id ORDER BY transaction_date
                              ROWS BETWEEN 6 PRECEDING AND CURRENT ROW) AS rolling_7txn_avg
    FROM transactions
    WHERE transaction_date < '2026-06-01'    -- the leakage guard, expressed in SQL
)
SELECT customer_id,
       COUNT(*) AS transaction_count,
       SUM(amount) AS total_amount,
       MAX(rolling_7txn_avg) FILTER (WHERE rn = 1) AS latest_rolling_avg
FROM customer_history
GROUP BY customer_id;
```

**When feature engineering belongs in SQL vs. Python**: push aggregation-heavy, large-scale feature computation (customer-level rollups over millions of transactions) to SQL/the warehouse, where the engine is purpose-built for exactly this and avoids pulling raw, unaggregated data into Python memory at all (Module 4 Section 33, 38); keep row-level, model-specific, or highly iterative feature logic (interaction terms, domain-specific ratios you're still actively experimenting with) in Python/Pandas, where iteration is faster and more flexible. A mature production pipeline (Section 119) very often does both - a SQL feature pipeline populating a feature table, with a thinner Python layer handling final model-specific transformations.

# SECTION 41 - FEATURE ENGINEERING PIPELINES

```python
# [Python] ColumnTransformer + Pipeline - the professional standard for bundled, leakage-safe preprocessing
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

numeric_features = ["age", "monthly_charges", "tenure_months"]
categorical_features = ["contract_type", "payment_method"]

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore")),
])

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features),
])
```

**Line by line**: `Pipeline` chains ordered steps, each with a name and a transformer, applying them in sequence and guaranteeing every step is fit only on whatever data is passed to `.fit()` (training data, if called correctly); `ColumnTransformer` applies *different* pipelines to *different column subsets* in parallel, then concatenates the results - exactly what a realistic dataset with both numeric and categorical columns needs. **`FeatureUnion` (conceptually)**: combines the outputs of multiple transformers applied to the *same* input, side by side (e.g. a raw numeric pipeline plus a separately-engineered polynomial-features pipeline, concatenated) - less commonly needed once `ColumnTransformer` covers the typical column-subset case, but useful when multiple genuinely different transformations of the *same* columns should all be included as parallel feature blocks.

This `preprocessor` object is the reusable core the rest of this module builds on - Section 43's first complete pipeline, every end-to-end project (Sections 98-105), and the capstone (Section 128) all wrap a `ColumnTransformer` exactly like this one inside a larger `Pipeline` with a model attached at the end.

# SECTION 42 - SCIKIT-LEARN ESTIMATOR API

| Method | Purpose |
|---|---|
| `fit(X, y)` | Learn parameters from data (model coefficients, tree splits, or a transformer's statistics) |
| `predict(X)` | Produce point predictions (class labels or regression values) |
| `predict_proba(X)` | Produce class probabilities (classifiers only) |
| `transform(X)` | Apply a learned transformation (imputers, scalers, encoders) |
| `fit_transform(X)` | `fit()` followed by `transform()` in one call - convenient, but dangerous to call on anything but training data |
| `score(X, y)` | A default performance metric (accuracy for classifiers, R² for regressors) - convenient but rarely the metric you actually want to report (Section 52-53) |

**Estimator consistency**: every scikit-learn estimator - whether a `SimpleImputer`, a `StandardScaler`, or a `RandomForestClassifier` - implements this same small, consistent interface, which is precisely what makes `Pipeline` and `ColumnTransformer` (Section 41) possible: they can chain arbitrary estimators together because every one of them exposes `fit`/`transform` or `fit`/`predict` in the same predictable shape. This consistency is also why swapping one model for another in a pipeline (Section 46) is a one-line change, not a rewrite.

# SECTION 43 - FIRST COMPLETE ML PIPELINE

```python
# [Python] Data -> Split -> Preprocess -> Train -> Predict -> Evaluate, on a realistic classification dataset
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, roc_auc_score

# 1. DATA
df = pd.read_csv("churn_data.csv")   # assume already cleaned per Module 4's discipline
X = df.drop(columns=["customer_id", "churn"])
y = (df["churn"] == "Yes").astype(int)

# 2. SPLIT
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

# 3. PREPROCESS + 4. TRAIN, bundled in one Pipeline object
model_pipeline = Pipeline([
    ("preprocessor", preprocessor),          # from Section 41
    ("classifier", LogisticRegression(max_iter=1000, random_state=42)),
])
model_pipeline.fit(X_train, y_train)          # fits BOTH preprocessing and the model, correctly, on training data only

# 5. PREDICT
y_pred = model_pipeline.predict(X_test)
y_proba = model_pipeline.predict_proba(X_test)[:, 1]

# 6. EVALUATE
print(classification_report(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_proba))
```

**Every line explained**: loading and immediately separating `X`/`y` (Module 4 Section 53) with ID and target columns removed; `stratify=y` (Section 23) on the split given typical churn class imbalance; wrapping preprocessing *and* the model together in one `Pipeline` so a single `.fit()` call correctly fits the imputer/scaler/encoder on training data and the classifier on the *already-transformed* training data, with no chance of accidentally leaking test data into any preprocessing step; `.predict()` for hard class labels, `.predict_proba()` for the probability needed by threshold-sensitive metrics (Section 55-57); evaluation using both a full classification report and ROC-AUC, never a single accuracy number alone (Section 52-53). This is the minimal but complete shape every subsequent pipeline in this module extends.

# SECTION 46 - MODEL SELECTION

**Selection criteria, deliberately not a ranking**: **problem type** (classification/regression/ranking narrows the candidate family immediately); **dataset size** (very small datasets favor simpler, lower-variance models; very large datasets can support more complex ones); **feature types** (heavily categorical, high-cardinality data often favors tree-based models over linear ones without extensive encoding work); **interpretability** (a regulated credit decision may require an inherently interpretable model - logistic regression, a shallow tree - over an opaque ensemble, Section 66); **latency** (a real-time fraud-scoring API has a hard millisecond budget a large ensemble or deep model may not meet, Section 80); **accuracy** (the metric that matters for the business problem, Sections 52-53, not accuracy generically); **training cost** (compute and engineering time to train and retrain, Section 90); **maintenance cost** (how hard the model is to debug, monitor, and keep current over its lifetime, Section 124).

The professional skill is not "know the best algorithm" but "correctly weigh these criteria against this specific problem's actual constraints" - a judgment call, informed by data, not a lookup table.

# SECTION 49 - MODEL TRAINING

**Vocabulary**: **parameters** (values the model *learns* from data during training - linear regression's coefficients, a neural network's weights); **hyperparameters** (values *you* set before training that control how learning happens - a tree's max depth, a regularization strength, Section 50); **objective function** (the quantity training actually minimizes/maximizes - e.g. log loss for logistic regression, squared error for linear regression); **optimization** (the algorithm that adjusts parameters to improve the objective function - gradient descent and its variants, or a closed-form solution for simple linear regression); **training iterations** (how many passes/updates the optimizer performs before stopping); **randomness** (many algorithms use randomness - random weight initialization, random feature/row subsampling in ensembles - which is why `random_state` (Section 22, 69) matters for reproducibility).

Understanding this vocabulary precisely is what lets you read *any* algorithm's documentation and immediately know which numbers are learned (don't need to be set) versus which numbers you're responsible for choosing (Section 50) - a distinction beginners frequently blur, leading to either leaving important hyperparameters at unsuitable defaults or, the opposite mistake, trying to manually "set" what should have been learned from data.

**What's learned vs. what's selected**: parameters (coefficients, tree split thresholds, network weights) are *learned* automatically by the optimization process during `.fit()`; hyperparameters are *selected* by you, before training, typically via a search procedure (Section 51) guided by validation performance - they are never learned from the training data's loss directly, because doing so would simply pick the hyperparameters that most overfit the training set. This selected/learned distinction is precisely why hyperparameter tuning must always be evaluated on validation data, never training data alone (Section 26, 28) - tuning on the training score alone would relentlessly favor whichever hyperparameters memorize training data best, which is the opposite of what you want.

**`GridSearchCV`** exhaustively tries every combination in a specified grid - guaranteed to find the best combination *within the grid*, but cost grows multiplicatively with the number of hyperparameters and values tried, quickly becoming impractical. **`RandomizedSearchCV`** samples a fixed number of random combinations from specified distributions - often finds a comparably good combination far faster, especially when only a few hyperparameters actually matter much (a well-documented empirical finding in hyperparameter-search research). **Bayesian optimization (conceptually)** - uses the results of previous trials to intelligently choose the next combination to try (rather than grid or pure random sampling), converging on good regions of the search space more efficiently than random search, at the cost of additional complexity; tools like `Optuna` or `scikit-optimize` implement this, worth reaching for once random search's cost becomes the bottleneck.

**Search space, computational cost, validation strategy**: define the search space based on domain knowledge of what reasonable values look like (not an arbitrarily huge range); always budget total computational cost (`n_iter` × `cv folds` × `training time per fit`) before launching a search; always perform the search using cross-validation (never a single validation split) so the chosen hyperparameters aren't overfit to one particular split's quirks.

# SECTION 53 - METRIC SELECTION

**Business costs determine metrics, not the other way around.** **Fraud**: a false negative (missed fraud) often costs far more than a false positive (a legitimate transaction flagged for review) - favoring high **recall**, with precision managed via threshold tuning (Section 55) to keep manual-review volume tractable. **Spam**: a false positive (a real email marked spam) is often more costly than a false negative (a spam email reaching the inbox) - favoring high **precision**. **Healthcare (screening)**: missing a disease (false negative) is frequently far more costly than a false alarm (false positive) - favoring high **sensitivity/recall**, even at some precision cost. **Forecasting**: MAE is easy to explain in business units and robust to outliers; RMSE is preferred when large errors are disproportionately costly (a huge single miss matters more than the average); the choice should follow from how the business actually experiences forecast error, not a default preference.

**The general principle**: before touching a metric, ask "what does a false positive cost, and what does a false negative cost, in this specific business context" (Section 95 formalizes this into an explicit cost matrix) - the answer determines which metric (or custom cost function) genuinely reflects success, and a model optimized for the wrong metric can be technically excellent and still actively harm the business.

# SECTION 55 - THRESHOLD OPTIMIZATION

```python
# [Python]
import numpy as np
from sklearn.metrics import precision_recall_curve

y_proba = model_pipeline.predict_proba(X_test)[:, 1]
precisions, recalls, thresholds = precision_recall_curve(y_test, y_proba)

f1_scores = 2 * (precisions * recalls) / (precisions + recalls + 1e-9)
best_threshold = thresholds[np.argmax(f1_scores[:-1])]

y_pred_custom = (y_proba >= best_threshold).astype(int)   # NOT the default 0.5
```

**Why the classification threshold is not the same thing as the model itself**: a trained classifier's `predict_proba()` output is a continuous score; `predict()`'s default behavior simply applies a 0.5 cutoff to convert that score into a hard label - but 0.5 is an arbitrary default with no inherent business meaning. The *same* trained model can be deployed with a threshold optimized for maximum F1, for a specific recall target ("catch at least 90% of fraud"), or directly for minimum expected business cost (Section 95) - changing the threshold changes the model's *behavior* without retraining anything. Treating the model and its deployed threshold as two separable decisions - one learned, one chosen based on business priorities - is a core professional distinction many beginners never make explicit.

# SECTION 56 - ROC CURVES

```python
# [Python]
from sklearn.metrics import roc_curve, RocCurveDisplay

fpr, tpr, thresholds = roc_curve(y_test, y_proba)
RocCurveDisplay(fpr=fpr, tpr=tpr).plot()
```

**TPR** (True Positive Rate = Recall = `TP/(TP+FN)`), **FPR** (False Positive Rate = `FP/(FP+TN)`) - the ROC curve plots TPR against FPR as the classification threshold sweeps from 1 to 0; **ROC-AUC** summarizes the whole curve as a single number - the probability a randomly chosen positive example is ranked higher than a randomly chosen negative one, independent of any specific threshold.

**Limitations with highly imbalanced datasets**: FPR's denominator (`FP+TN`) is dominated by the (very large) number of true negatives in a heavily imbalanced problem, so even a large absolute number of false positives can produce a deceptively small FPR and an impressively high ROC-AUC, while precision - which a business user actually cares about (Section 53's fraud example) - can simultaneously be poor. This is exactly why Section 57's precision-recall curve is often more informative for rare-event problems (fraud, disease detection) than ROC-AUC alone.

# SECTION 61 - STATISTICAL UNCERTAINTY IN ML PERFORMANCE

Building on Module 7's inferential statistics:

```python
# [Python] Bootstrap confidence interval for a single metric on the test set
import numpy as np

def bootstrap_metric_ci(y_true, y_pred_proba, metric_fn, n_bootstrap=1000, ci=95):
    y_true, y_pred_proba = np.array(y_true), np.array(y_pred_proba)
    scores = []
    n = len(y_true)
    for _ in range(n_bootstrap):
        idx = np.random.choice(n, n, replace=True)                # resample WITH replacement
        scores.append(metric_fn(y_true[idx], y_pred_proba[idx]))
    lower = np.percentile(scores, (100 - ci) / 2)
    upper = np.percentile(scores, 100 - (100 - ci) / 2)
    return np.mean(scores), lower, upper

from sklearn.metrics import roc_auc_score
mean_auc, ci_low, ci_high = bootstrap_metric_ci(y_test, y_proba, roc_auc_score)
print(f"ROC-AUC: {mean_auc:.3f} (95% CI: [{ci_low:.3f}, {ci_high:.3f}])")
```

**Fold variability and repeated evaluation**: cross-validation's fold-to-fold standard deviation (Section 60) is itself a form of uncertainty estimate; `RepeatedStratifiedKFold` (Section 27) with multiple repeats further reduces the noise in that estimate by averaging over many different fold assignments, not just one.

**Why reporting only one score is insufficient**: a single test-set metric is one sample from a distribution of possible outcomes that depends on exactly which rows happened to land in the test set - report it alone, and a colleague (or a future you) has no way to judge whether a 0.02 improvement between two model versions is a real, reproducible gain or noise. A confidence interval (via bootstrap, as above, or cross-validation's spread) turns "model B is better than model A" into a testable, honest claim rather than an assertion based on a single lucky (or unlucky) split.

# SECTION 67 - EXPERIMENT TRACKING FOUNDATIONS

**What should be recorded for every experiment**: **dataset version** (exact snapshot or query used); **features** (the exact feature set and their definitions); **model** (algorithm and library version); **hyperparameters** (the exact configuration used); **random seed** (Section 69); **metrics** (all of them, with confidence intervals where available, not just the headline number); **code version** (a git commit hash); **environment** (dependency versions - Python, scikit-learn, etc.); **experiment ID** (a unique identifier tying all of the above together, so any result can be traced back to exactly how it was produced).

```python
# [Python] A minimal, dependency-free experiment log - the conceptual foundation before a dedicated tool is introduced
import json, uuid, subprocess, datetime

def log_experiment(config: dict, metrics: dict, log_path="experiments.jsonl"):
    record = {
        "experiment_id": str(uuid.uuid4()),
        "timestamp": datetime.datetime.utcnow().isoformat(),
        "git_commit": subprocess.run(["git", "rev-parse", "HEAD"], capture_output=True, text=True).stdout.strip(),
        "config": config,
        "metrics": metrics,
    }
    with open(log_path, "a") as f:
        f.write(json.dumps(record) + "\n")

log_experiment(config={"model": "random_forest", "max_depth": 7}, metrics={"roc_auc": 0.87, "recall": 0.72})
```

# SECTION 69 - RANDOMNESS IN MACHINE LEARNING

**Where randomness enters**: **train/test splitting** (which rows land where, Section 22); **initialization** (random starting weights in many algorithms); **sampling** (bootstrap sampling inside random forests, stochastic gradient descent's mini-batch sampling); **algorithms** (some algorithms have inherently stochastic components beyond initialization); **cross-validation** (fold assignment, Section 27); **hyperparameter search** (`RandomizedSearchCV`'s sampling, Section 51).

**Why `random_state` does not guarantee complete determinism across every environment**: fixing `random_state` controls the *pseudo-random number generator's* sequence within a given library and version - but results can still differ across different library versions (an algorithm's internal implementation can change between releases), different hardware (some parallelized or GPU-accelerated operations are not bit-for-bit deterministic even with a fixed seed, due to floating-point operation ordering), and different operating systems or thread counts (parallel execution order can vary). The professional takeaway: `random_state` is necessary but not sufficient for full reproducibility - pin library versions too (Section 68), and treat small, seed-related result fluctuations across environments as expected rather than alarming, while investigating *large* unexpected differences.

# SECTION 75 - DATA VALIDATION BEFORE TRAINING

```python
# [Python] A pre-training validation gate
def validate_training_data(df: pd.DataFrame, config: dict) -> None:
    errors = []

    missing_cols = set(config["required_columns"]) - set(df.columns)
    if missing_cols:
        errors.append(f"Missing required columns: {missing_cols}")

    for col, expected_dtype in config["dtypes"].items():
        if str(df[col].dtype) != expected_dtype:
            errors.append(f"{col}: expected {expected_dtype}, got {df[col].dtype}")

    missing_rate = df[config["required_columns"]].isna().mean()
    too_sparse = missing_rate[missing_rate > 0.5].index.tolist()
    if too_sparse:
        errors.append(f"Columns exceeding 50% missing: {too_sparse}")

    if config["target_column"] not in df.columns or df[config["target_column"]].isna().any():
        errors.append("Target column missing or contains nulls")

    if df.duplicated(subset=config["id_columns"]).any():
        errors.append("Duplicate records found on ID columns")

    if errors:
        raise DataValidationError("\n".join(errors))

    logger.info("Data validation passed")
```

This function is the training-pipeline equivalent of a pre-flight checklist - **required columns**, **data types**, **missingness** thresholds, **range/category validity** (extended straightforwardly following the same pattern), **duplicate records**, and **target availability** are all checked *before* any training compute is spent, catching problems in seconds rather than after an expensive training run silently produces a model built on broken data. This is a direct, ML-training-specific instance of the Data Governance handbook's Part 9-13 data-quality discipline.

# SECTION 76 - TRAINING PIPELINE DESIGN

```
INGEST -> VALIDATE -> SPLIT -> PREPROCESS -> FEATURE ENGINEER -> TRAIN -> VALIDATE -> EVALUATE -> SAVE MODEL
```

```python
# [Python] The orchestrating training pipeline, tying every prior section together
def run_training_pipeline(config: dict):
    logger.info("1. Ingesting data")
    df = pd.read_parquet(config["data"]["source_path"])

    logger.info("2. Validating data")
    validate_training_data(df, config)                                    # Section 75

    logger.info("3. Splitting data")
    X = df.drop(columns=config["target_column"])
    y = df[config["target_column"]]
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=config["split"]["test_size"],
        stratify=y, random_state=config["split"]["random_state"])          # Sections 21-23

    logger.info("4-5. Preprocessing and feature engineering")
    pipeline = build_pipeline(config)                                        # Sections 30-41

    logger.info("6. Training")
    pipeline.fit(X_train, y_train)

    logger.info("7. Validating / 8. Evaluating")
    metrics = evaluate_model(pipeline, X_test, y_test)                          # Sections 52-59
    logger.info(f"Test metrics: {metrics}")

    logger.info("9. Saving model")
    save_model_artifact(pipeline, metrics, config)                                # Section 77

    log_experiment(config, metrics)                                                # Section 67
    return pipeline, metrics
```

Note this pipeline's `VALIDATE` step appears *twice* in the diagram - once before splitting (raw data quality, Section 75) and once after training (model performance against held-out data, Sections 52-59) - a deliberate, non-redundant distinction between validating that the *data* is trustworthy and validating that the *model* is good.

# SECTION 77 - MODEL ARTIFACTS

```python
# [Python]
import joblib
from datetime import datetime

def save_model_artifact(pipeline, metrics: dict, config: dict, version: str = None):
    version = version or datetime.utcnow().strftime("%Y%m%d_%H%M%S")
    artifact = {
        "pipeline": pipeline,             # the FULL pipeline - preprocessing AND model together (Section 78)
        "metrics": metrics,
        "config": config,
        "version": version,
        "trained_at": datetime.utcnow().isoformat(),
    }
    path = f"models/model_{version}.joblib"
    joblib.dump(artifact, path)
    logger.info(f"Model artifact saved: {path}")
    return path
```

**What to serialize**: not just the model - the **entire fitted pipeline** (preprocessor + model together), since Section 78 requires training and inference preprocessing to match exactly, and the surest way to guarantee that is to serialize them as one inseparable object. **Metadata** (metrics, config, version, training timestamp) travels alongside the model itself, not in a separate, easy-to-lose document. **`joblib`** is preferred over plain `pickle` for scikit-learn objects (more efficient for large NumPy arrays inside fitted estimators). **Pickle risks, conceptually**: a pickled/joblib object executes arbitrary code upon loading if the file has been tampered with - never load a model artifact from an untrusted source, and treat model files with the same security discipline as executable code, not as inert data.

# SECTION 78 - INFERENCE WORKFLOW

```
RAW INPUT -> VALIDATION -> PREPROCESSING -> MODEL -> PREDICTION -> POSTPROCESSING -> BUSINESS DECISION
```

```python
# [Python]
def predict(raw_input: pd.DataFrame, artifact: dict) -> pd.DataFrame:
    validate_inference_schema(raw_input, artifact["config"])                # VALIDATION
    proba = artifact["pipeline"].predict_proba(raw_input)[:, 1]              # PREPROCESSING + MODEL (one call, guaranteed matching)
    threshold = artifact["config"].get("decision_threshold", 0.5)              # POSTPROCESSING
    decision = (proba >= threshold).astype(int)
    return pd.DataFrame({"churn_probability": proba, "flagged_for_outreach": decision})  # BUSINESS DECISION
```

**Why training and inference preprocessing must match exactly**: if training used `StandardScaler` fit on training data, and inference recomputes scaling parameters from *new* incoming data (or applies a different transformation entirely), predictions become meaningless - the model was trained on one numeric scale and is now being asked to interpret a different one. This is precisely why Section 77 serializes the *entire fitted pipeline* rather than the model alone: calling `artifact["pipeline"].predict_proba(raw_input)` on raw, unprocessed input applies the *exact same, already-fitted* preprocessing steps used in training, by construction eliminating this entire class of bug - one of the most consequential and most common production ML failures (Section 111's "model performs well offline but poorly online" case).

# SECTION 79 - BATCH INFERENCE

```python
# [Python]
def run_batch_inference(input_path: str, artifact_path: str, output_path: str, chunk_size: int = 50_000):
    artifact = joblib.load(artifact_path)
    results = []
    for chunk in pd.read_parquet(input_path, chunksize=chunk_size):   # Module 4 Section 38's discipline, applied here
        predictions = predict(chunk, artifact)
        results.append(predictions)
    pd.concat(results).to_parquet(output_path, index=False)
    logger.info(f"Batch inference complete: {output_path}")
```

**Scheduled inference**: batch jobs typically run on a schedule (nightly, weekly) matching how often the business decision they support actually needs refreshing (Section 5's prediction frequency) - orchestrated via the same tools covered in the Data Engineering handbook (Airflow, Part 30-31). **Large datasets**: chunked processing (as above) keeps memory bounded when scoring millions of rows. **Output storage**: predictions are typically written to a table or file the downstream business process (a dashboard, a CRM sync, a retention-team worklist) consumes directly - batch inference's output is itself a dataset with its own quality and freshness expectations (Data Governance handbook Part 27).

# SECTION 86 - DATA DRIFT

```python
# [Python] Population Stability Index (PSI) - a standard drift metric
import numpy as np

def calculate_psi(expected: np.ndarray, actual: np.ndarray, bins=10) -> float:
    breakpoints = np.percentile(expected, np.linspace(0, 100, bins + 1))
    breakpoints[0], breakpoints[-1] = -np.inf, np.inf
    expected_pct = np.histogram(expected, breakpoints)[0] / len(expected)
    actual_pct = np.histogram(actual, breakpoints)[0] / len(actual)
    expected_pct = np.clip(expected_pct, 1e-6, None)   # avoid division by zero / log(0)
    actual_pct = np.clip(actual_pct, 1e-6, None)
    return np.sum((actual_pct - expected_pct) * np.log(actual_pct / expected_pct))

psi = calculate_psi(training_feature_values, production_feature_values)
# Conventional interpretation: PSI < 0.1 stable, 0.1-0.25 moderate shift, > 0.25 significant drift
```

# SECTION 87 - CONCEPT DRIFT

**Definition**: concept drift occurs when the underlying *relationship* between features (X) and target (Y) changes over time, even if the feature distributions themselves look stable - a fundamentally different and often harder-to-detect problem than data drift.

**Why a model can degrade without obvious feature drift**: imagine a fraud model where fraudsters *change tactics* - the same feature values (transaction amount, time of day, merchant category) that used to signal "low risk" now correlate with a new fraud pattern. Every feature's distribution can look identical to training data (no data drift signal at all) while the model's predictions become steadily less accurate, because the *mapping* it learned no longer reflects reality. **Examples**: fraud (adversarial, adapting tactics deliberately); customer behavior (a competitor's new offering shifts what "churn risk" looks like); financial markets (regime changes - a model trained on a low-interest-rate environment can fail entirely once rates shift structurally). Concept drift can only be reliably detected by monitoring *actual model performance against true labels as they arrive* (Section 88) - data-drift monitoring alone is not sufficient to catch it, which is precisely why both layers from Section 85 are needed together.

# SECTION 89 - BUSINESS MONITORING

**Monitored KPIs, by domain**: revenue impact of a pricing or recommendation model; conversion rate for a marketing-response model; fraud losses (both missed fraud and false-positive friction cost) for a fraud model; churn rate itself, as the ultimate outcome a churn model exists to influence; loan approval rate and downstream default rate for a credit model; customer complaints, which can spike even when every technical metric looks fine (e.g. a model that's technically accurate but systematically frustrates a specific customer segment, Section 63).

**Why technically healthy models can still fail business objectives**: a churn model can maintain excellent, stable ROC-AUC indefinitely while churn itself keeps rising - because the *model* was never the actual lever; the retention *team's* follow-through on the model's flagged accounts was. A fraud model's precision/recall can look perfect while fraud losses climb, because fraudsters shifted to an attack vector the model was never designed to catch. Business monitoring exists precisely to catch this gap between "the model is performing as designed" and "the business problem is actually improving" - the two are related but never automatically identical, and only business-KPI monitoring, tied back to Section 2's original objective, can tell them apart.

# SECTION 90 - RETRAINING STRATEGIES

| Strategy | Trigger |
|---|---|
| Scheduled retraining | A fixed calendar cadence (weekly, monthly) regardless of observed drift |
| Trigger-based retraining | A specific event (a major product launch, a known external shift) prompts an ad hoc retrain |
| Drift-based retraining | Automated retraining fires when Section 86's drift metrics cross a defined threshold |
| Performance-based retraining | Automated retraining fires when Section 88's tracked performance drops below a defined threshold |

**Risks of automatic retraining**: a fully automated pipeline can retrain on *already-drifted or corrupted* data without a human ever reviewing it, silently propagating a data-quality problem into a new "official" production model; automatic retraining on a feedback loop where the model's own predictions influence what data gets collected next (e.g. a model that suppresses certain accounts from review, so labels on those accounts stop being generated at all) can create a self-reinforcing bias loop; and retraining without re-validating the *entire* pipeline (not just re-fitting the model) risks silently baking in a leakage or contract-violation bug that manual review would have caught. The professional default: automated retraining triggers should raise an alert and stage a *candidate* model for human review and validation (Section 84's staging concept) rather than auto-promoting directly to production without oversight, at least until a team has strong, demonstrated confidence in the retraining pipeline's own reliability.

# SECTION 91 - MODEL ROLLBACK

```python
# [Python] A minimal rollback pattern
def rollback_to_previous(registry: pd.DataFrame, current_prod_version: str) -> str:
    history = registry[registry["stage"].isin(["production", "archived"])].sort_values("trained_at", ascending=False)
    previous = history[history["version"] != current_prod_version].iloc[0]
    logger.warning(f"Rolling back from {current_prod_version} to {previous['version']}")
    return previous["version"]   # deployment infrastructure then points traffic at this version's artifact
```

**Vocabulary**: **previous model** (the last known-good production version); **versioned artifact** (Section 77, 83 - what makes rollback possible at all - without a retained, versioned prior artifact, there is nothing to roll back *to*); **rollback trigger** (a monitoring alert, Section 85-89, an incident, Section 125, or a manual decision after a bad deployment); **safe deployment** (deploying a new model version in a way that keeps rollback fast and low-risk - e.g. keeping the previous version's artifact and serving infrastructure warm and ready, rather than tearing it down immediately upon promoting a new version). Rollback capability is not optional infrastructure - it is the single most important safety net between "a new model has a subtle, undetected problem" and "the business suffers extended, avoidable harm" before a proper fix can be developed.

# SECTION 95 - COST-SENSITIVE MACHINE LEARNING

```python
# [Python] Expected-cost threshold optimization, replacing a generic F1-optimized threshold with a genuine business-cost-optimized one
cost_matrix = {
    "false_positive": 5,      # e.g. cost of a manual fraud review that turns out to be a false alarm
    "false_negative": 500,    # e.g. cost of an undetected fraudulent transaction
    "true_positive": 5,       # the review still costs something even when correct
    "true_negative": 0,
}

def expected_cost(y_true, y_proba, threshold, cost_matrix):
    y_pred = (y_proba >= threshold).astype(int)
    tp = ((y_pred == 1) & (y_true == 1)).sum()
    fp = ((y_pred == 1) & (y_true == 0)).sum()
    fn = ((y_pred == 0) & (y_true == 1)).sum()
    tn = ((y_pred == 0) & (y_true == 0)).sum()
    return (tp*cost_matrix["true_positive"] + fp*cost_matrix["false_positive"] +
            fn*cost_matrix["false_negative"] + tn*cost_matrix["true_negative"])

thresholds = np.linspace(0.01, 0.99, 99)
costs = [expected_cost(y_test, y_proba, t, cost_matrix) for t in thresholds]
optimal_threshold = thresholds[np.argmin(costs)]
```

**Cost matrix, false-positive cost, false-negative cost, expected cost**: this generalizes Section 55's threshold optimization from a generic statistical criterion (F1) to a genuine business-value criterion - the threshold that minimizes *actual expected dollar cost*, given honestly-estimated per-outcome costs. **Fraud example**: a missed fraudulent transaction (false negative) typically costs far more than one unnecessary manual review (false positive) - the cost matrix should reflect this asymmetry explicitly, rather than treating both error types as equally bad, which F1 implicitly does. **Healthcare example**: a missed diagnosis (false negative) is often far costlier - in both financial and human terms - than a false alarm requiring further (relatively low-cost) testing to rule out, again pushing the cost-optimal threshold toward high recall, exactly as Section 53 argued qualitatively, now made precise and quantitative.

**Simple baselines** (Section 29) are not just a sanity check - they are frequently a legitimate, deployable answer, especially when a complex model's improvement over baseline is marginal relative to its added maintenance burden. **Complexity** buys potential accuracy gains at the cost of: **interpretability** (Section 66, 94 - a complex ensemble is harder to explain and audit); **maintenance** (more moving parts, more that can silently break, more retraining complexity, Section 90); **generalization** risk (a highly complex model is more prone to overfitting subtle, non-generalizable patterns in training data, particularly on smaller datasets).

# SECTION 111 - MACHINE LEARNING TROUBLESHOOTING

Representative set of 12, extensible to the full 50. Each: Symptom -> Diagnosis -> Root causes -> Investigation -> Solution -> Prevention.

**1. Training score high, test score low.** *Diagnosis*: overfitting. *Root causes*: model too complex for the amount of data; insufficient regularization; too few training examples. *Investigation*: plot a learning curve (training vs. validation score across increasing dataset sizes). *Solution*: simplify the model, add regularization, gather more data, or use cross-validation-guided tuning (Section 27, 51). *Prevention*: always monitor train and validation scores together, never train alone.

**2. Cross-validation scores vary heavily across folds.** *Diagnosis*: high variance, possibly a small dataset or a poorly-behaved split. *Root causes*: small sample size, especially for a minority class; a fold-defining variable (like a rare category) unevenly distributed across folds. *Investigation*: check per-fold class balance and sample size (Section 23). *Solution*: use `StratifiedKFold`, increase `n_splits`, or use `RepeatedStratifiedKFold` (Section 27) for a more stable estimate. *Prevention*: always inspect fold-to-fold variance (Section 60), not just the mean.

**3. Model predicts only one class.** *Diagnosis*: severe class imbalance overwhelming the model, or a broken target/feature pipeline. *Root causes*: extreme imbalance with no class weighting; a constant or near-constant feature; a bug zeroing out the minority class during preprocessing. *Investigation*: check `y_train.value_counts()`, check `class_weight` settings, verify preprocessing output shape and values for a minority-class sample row. *Solution*: `class_weight="balanced"`, resampling techniques, or a cost-sensitive threshold (Section 95) rather than relying on `predict()`'s default 0.5 cutoff. *Prevention*: always check the baseline's behavior (Section 44) against the imbalance level before assuming a "real" model's poor behavior is a bug rather than an expected consequence of imbalance.

**4. Precision is low.** *Diagnosis*: too many false positives. *Root causes*: threshold set too low for the actual class balance/cost trade-off; features insufficiently discriminating; genuine label noise inflating apparent false positives. *Investigation*: inspect Section 62's false-positive set for common patterns. *Solution*: raise the threshold (Section 55), add more discriminating features, or investigate/clean noisy labels. *Prevention*: threshold optimization (Section 55, 95) as a standard, not optional, pipeline step.

**5. Recall is low.** *Diagnosis*: too many false negatives - the model is missing real positive cases. *Root causes*: threshold too high; the model genuinely lacks signal for a subset of positive cases (Section 63's segment analysis); severe class imbalance without correction. *Investigation*: inspect Section 62's false-negative set; check segment-level recall. *Solution*: lower the threshold, add class weighting/resampling, or engineer new features targeting the specific missed segment. *Prevention*: always evaluate recall explicitly when false negatives are costly (Section 53), not just overall accuracy.

**6. ROC-AUC high but business results poor.** *Diagnosis*: Section 89's exact scenario - technical health does not guarantee business success. *Root causes*: wrong metric for the actual decision (Section 53); poor calibration if probabilities feed a downstream calculation (Section 58); the model is accurate but the *decision process* built around it isn't acting on its output correctly. *Investigation*: trace the full decision pipeline, not just the model - is the retention team actually calling flagged customers? *Solution*: realign the metric with the true business cost (Section 95), fix calibration, or fix the downstream process, not the model. *Prevention*: business-KPI monitoring (Section 89) from day one, not just technical metrics.

**7. Probability estimates are poorly calibrated.** *Diagnosis*: reliability diagram deviates from the diagonal (Section 58). *Root causes*: tree ensembles' well-known tendency toward overconfident probabilities; training on an artificially rebalanced dataset without correcting probabilities back to the true population rate afterward. *Investigation*: `calibration_curve()`. *Solution*: `CalibratedClassifierCV`, or correct for resampling-induced probability shift explicitly. *Prevention*: check calibration whenever raw probability *values* (not just rankings) feed a downstream decision.

**8. Model performs well offline but poorly online.** *Diagnosis*: Section 78's exact warning - training/inference mismatch, or genuine leakage that inflated offline numbers. *Root causes*: preprocessing recomputed differently at inference than training; a leaked feature available offline but not genuinely available at real prediction time; population shift between training data and live traffic. *Investigation*: compare training-time and inference-time feature values for identical inputs; re-audit every feature against Section 19-20's leakage checklist. *Solution*: serialize and reuse the exact fitted pipeline (Section 77-78); remove any leaked feature and retrain. *Prevention*: the single strongest prevention is Section 78's "predict on raw input through the full serialized pipeline" pattern, never reimplementing preprocessing separately for inference.

**9. Features are missing in production.** *Diagnosis*: an inference request lacks a column the model expects. *Root causes*: an upstream source system change (Data Governance handbook Part 43); a feature-store outage; a genuinely new customer with no history to compute a historical-aggregate feature from. *Investigation*: check Section 75-style validation logs for the specific missing column and its frequency. *Solution*: define and test explicit fallback behavior for missing features (a sensible default, or a documented "cannot score" response) rather than crashing. *Prevention*: input-schema validation (Section 74, 82) at the inference boundary, always.

**10. Data schema changes (upstream).** *Diagnosis*: a previously reliable source column's type, name, or meaning changed. *Root causes*: an uncoordinated upstream deployment (Data Governance handbook Part 42-43's data-contract discussion applies directly). *Investigation*: schema-drift diff between today's and yesterday's source data (Module 4 Section 43-style check). *Solution*: patch the pipeline; establish a data contract with the source team going forward. *Prevention*: automated schema-drift detection as a standing pipeline check, not a reactive one.

**11. Predictions drift over time.** *Diagnosis*: Section 86's data-drift or Section 87's concept-drift signal firing. *Investigation*: PSI and statistical drift tests on key features; compare recent prediction-score distribution to the training-time distribution. *Solution*: depends on cause - retrain (Section 90) if genuine drift; fix a pipeline bug if the "drift" is actually a data-quality regression. *Prevention*: continuous drift monitoring, not periodic manual spot-checks.

**12. Model performance drops sharply right after deployment.** *Diagnosis*: almost always a training/inference mismatch (case 8) or a genuine, sudden population shift (a new product launch, an external shock). *Investigation*: compare the very first production predictions' input feature distributions against training data immediately (don't wait for delayed true labels, Section 88). *Solution*: rollback immediately (Section 91) while investigating, rather than leaving a broken model live. *Prevention*: a staged rollout (a small percentage of traffic first) with tight monitoring before full promotion, directly informed by Section 84's staging concept.

# SECTION 117 - COMPUTATIONAL EFFICIENCY

**How workflow design affects resources**: **memory** (loading an entire large dataset into Pandas at once vs. chunked/SQL-pushed processing, Module 4 Section 38); **CPU** (vectorized operations vs. Python loops, Module 4 Section 39; parallelized cross-validation via `n_jobs=-1`); **GPU** (relevant for gradient-boosted tree libraries with GPU support and, later in the curriculum, deep learning - not typically needed for classical scikit-learn models on tabular data at moderate scale); **runtime** (total wall-clock time for a training run or a hyperparameter search, directly trading off against Section 51's search-space size); **data movement** (transferring large datasets between systems - a database, object storage, a training environment - is often the actual bottleneck, not computation itself).

**Levers**: **vectorization** (Module 4's central performance lesson, applying identically here to feature-engineering code); **batching** (processing data in bounded chunks rather than all at once, Module 4 Section 38, 79's batch-inference pattern); **sampling** (training on a representative subset during early experimentation, only scaling to the full dataset for final candidate models, dramatically speeding up iteration); **parallelization (conceptually)** (cross-validation folds, hyperparameter search trials, and independent feature-engineering steps are all naturally parallelizable - `n_jobs=-1` in scikit-learn is the simplest expression of this). Workflow-level efficiency decisions, made early, often matter more to total project time than any individual model's training speed.

# SECTION 118 - LARGE DATASET WORKFLOW

**How the workflow changes at scale**: **chunked processing** (Module 4 Section 38 - process data in bounded pieces rather than loading everything into memory at once); **SQL preprocessing** (push filtering, joining, and aggregation to the database/warehouse before pulling data into Python - Module 4 Section 33, 40, and this module's Section 40); **Parquet** (Module 4 Section 36 - the default storage format once a dataset outgrows casual CSV handling, both for load speed and for column-selective reads); **sampling** (train on a representative, stratified sample for exploration and early model iteration, validating that findings hold on the full dataset only for the final candidate); **distributed processing (conceptually)** (Spark/Dask, Module 4 Section 63 and the Data Engineering handbook - needed once a dataset genuinely exceeds what one machine can hold or process in reasonable time, a threshold well beyond what most individual ML projects actually hit, but essential awareness for large-scale production systems); **cloud storage (conceptually)** (training data increasingly lives in object storage - S3/GCS/Azure Blob, Data Engineering handbook Part 23 - accessed directly rather than downloaded wholesale first).

The professional judgment call here mirrors Module 4 Section 38-39's guidance directly: profile first, and only reach for the next tier of complexity (chunking, then SQL push-down, then genuinely distributed processing) once the current tier's limits are actually, measurably hit - not preemptively.

# SECTION 119 - ML WORKFLOW WITH SQL

```
SQL -> DATA MART -> PYTHON -> FEATURE PIPELINE -> MODEL -> RESULTS -> DATABASE/DASHBOARD
```

A realistic professional pattern: a **SQL** layer (often a dbt-managed transformation layer, SQL-for-Data-Engineering handbook Part 34) builds a governed **data mart** - a clean, business-validated, conformed table specifically curated for a class of ML problems (e.g. a `customer_features_mart` refreshed nightly); **Python** pulls from this mart (Section 34 of Module 4) rather than raw source tables, inheriting the mart's already-validated quality; the **feature pipeline** (this module's Sections 36-41) applies model-specific transformations on top; the **model** trains and predicts; **results** (predictions, scores) are written back to a **database** for downstream consumption or surfaced directly in a **dashboard** (Power BI, or the Streamlit demo in Section 121). This pattern deliberately keeps heavy, reusable, cross-project aggregation in SQL/the warehouse (governed, tested, shared) while keeping model-specific, still-evolving logic in Python - directly extending Section 40's SQL-vs-Python feature-engineering guidance to the whole project's architecture.

# SECTION 120 - ML WORKFLOW WITH APIs

```
API -> DATA VALIDATION -> FEATURE ENGINEERING -> MODEL -> RESPONSE
```

```python
# [Python] A fuller FastAPI example combining validation, on-the-fly feature engineering, and prediction
from fastapi import FastAPI
from pydantic import BaseModel
import pandas as pd

app = FastAPI()
artifact = joblib.load("models/model_latest.joblib")

class TransactionInput(BaseModel):
    amount: float
    merchant_category: str
    account_age_days: int
    transaction_hour: int

@app.post("/score")
def score_transaction(txn: TransactionInput):
    df = pd.DataFrame([txn.model_dump()])
    df["is_high_amount"] = df["amount"] > 1000          # simple online feature engineering
    df["is_late_night"] = df["transaction_hour"].between(0, 5)
    proba = artifact["pipeline"].predict_proba(df)[:, 1][0]
    return {"fraud_probability": float(proba)}
```

This extends Section 82's API design foundations with the explicit feature-engineering step made visible - a reminder that any feature computed *at inference time* rather than pulled from a pre-computed feature store must be computed using *exactly* the same logic as training-time feature engineering (Section 78's core warning), which is why complex, history-dependent features (rolling averages, RFM) are far better served by a pre-computed feature store than by online recomputation inside the API request path.

# SECTION 125 - ML INCIDENT RESPONSE

**Triggering conditions**: predictions become abnormal (a sudden, sharp shift in prediction distribution, Section 86); the data pipeline feeding the model fails (upstream ingestion or feature-computation breaks entirely, Data Engineering handbook Part 75); model latency spikes (Section 80's operational budget breached); model performance collapses (Section 88, once labels confirm it, or Section 86's drift signals suggest it earlier); a business KPI deteriorates in a way plausibly tied to the model (Section 89).

**A basic ML incident response process**: (1) **detect** - an automated alert fires from monitoring (Sections 85-89); (2) **triage** - assess severity and whether immediate action (rollback, Section 91) is warranted before full investigation; (3) **contain** - roll back to the previous known-good model version if the issue is severe and the cause isn't yet clear, prioritizing stopping ongoing harm over root-causing immediately; (4) **investigate** - trace the issue through Section 111's troubleshooting playbook; (5) **resolve** - fix the underlying cause (retrain, patch a pipeline bug, correct a schema mismatch); (6) **validate** - confirm the fix genuinely resolves the issue before re-promoting; (7) **document** - a full postmortem (Section 126). This process mirrors, and should be directly integrated with, the Data Engineering and Data Governance handbooks' incident-management processes (Data Engineering handbook Part 75; Data Governance handbook Part 33) - an ML incident is very often, at its root, a data or pipeline incident with an ML-specific symptom.

# SECTION 131 - MASTER DECISION TREES

**1. Should this problem use ML?** Enough data? -> No -> use rules/expert judgment. Yes -> Target measurable? -> No -> stop, define the target first. Yes -> Repetitive + actionable decision? -> No -> reconsider. Yes -> proceed (Section 4).

**2. Classification or regression?** Is the target a category (finite set of labels)? -> Classification. Is it a continuous quantity? -> Regression. Is it "when will X happen"? -> Survival/time-to-event. Is it "how should these be ordered"? -> Ranking (Section 3).

**3. Random split or temporal split?** Does time order matter for any feature or for the target's meaning? -> Yes -> temporal split (Section 25). No, but multiple rows per entity? -> Yes -> grouped split (Section 24). Neither -> random split, stratified if imbalanced (Sections 22-23).

**4. Which validation strategy?** Small dataset, need a stable estimate? -> Repeated/Stratified K-Fold (Section 27). Grouped entities? -> GroupKFold/StratifiedGroupKFold. Time-dependent? -> TimeSeriesSplit/walk-forward. Need both tuning AND an honest final score from limited data? -> Nested CV (Section 28).

**5. Which metric?** What does a false positive cost vs. a false negative? -> Asymmetric and known -> cost-sensitive threshold on a ranking metric (ROC-AUC/PR-AUC) (Sections 53, 95). Severely imbalanced target? -> favor PR-AUC over ROC-AUC (Section 57). Regression, outliers matter a lot? -> RMSE. Outliers should not dominate? -> MAE (Section 52).

**6. Which preprocessing?** Scale-sensitive model (linear, distance-based, neural net)? -> Scale features (Section 33). Tree-based model? -> Scaling unnecessary. High-cardinality categorical? -> Frequency/target/hashing encoding, not naive one-hot (Section 32). Missingness plausibly MNAR? -> Add a missingness-indicator feature alongside imputation (Section 31).

**7. Which baseline?** Classification -> majority-class + business baseline. Regression -> mean/median + business baseline. A meaningful existing heuristic exists? -> Always include it as a baseline too (Section 29).

**8. Which model families to try first?** Need interpretability/regulatory approval? -> Start with linear/logistic regression, simple trees. Need maximum accuracy, interpretability not required? -> Gradient-boosted trees, random forests. Very large, unstructured data? -> Defer to later deep-learning-focused modules (Section 46, 96).

**9. Which error analysis?** Classification -> confusion-matrix-driven FP/FN inspection (Section 62). Regression -> residual plots, worst-absolute-error inspection (Section 59, 62). Any imbalance across a business-relevant subgroup? -> Always add segment-level evaluation (Section 63).

**10. When to retrain?** Performance dropped (once labels available)? -> Performance-based retrain (Section 90). Significant drift detected before labels arrive? -> Drift-based retrain, treated as a candidate requiring review (Section 84, 90). Fixed business cadence with no strong drift signal either way? -> Scheduled retrain as a safe default.
