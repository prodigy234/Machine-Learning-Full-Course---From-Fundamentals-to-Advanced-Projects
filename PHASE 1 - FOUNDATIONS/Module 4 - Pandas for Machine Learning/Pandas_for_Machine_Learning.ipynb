# SECTION 1 - WHAT IS PANDAS?

**Beginner explanation**: Pandas is a Python library for working with tables of data - rows and columns, like a spreadsheet or a database table, but manipulated with code instead of clicking cells.

**Technical definition**: Pandas is an open-source data analysis and manipulation library built on top of NumPy, providing two primary labeled data structures - `Series` (1-dimensional) and `DataFrame` (2-dimensional) - along with a large API for reading, cleaning, transforming, aggregating, and writing tabular and time-series data.

**Intuition**: NumPy gives you fast arrays of numbers with no labels; Pandas wraps NumPy arrays with row and column **labels** (an `Index` and column names), so `df["age"]` means something to a human reading the code, and rows can be selected by a meaningful key (a customer ID, a date) instead of only a raw position.

**Why Pandas was created**: before Pandas (created by Wes McKinney in 2008 for financial data analysis), Python had no first-class tool for labeled, heterogeneous tabular data - analysts either fought with raw NumPy arrays (fast, but no labels, and every column has to share one dtype) or left Python entirely for R or Excel. Pandas filled exactly that gap.

**Pandas and NumPy**: every Pandas `Series` is backed by a NumPy array (or, increasingly, other array-like backends) internally - Pandas adds labels, mixed-dtype columns, missing-value handling, and a huge table-oriented API on top of NumPy's raw numerical speed. You will constantly move between them: NumPy for numerical/linear-algebra-heavy work, Pandas for labeled, heterogeneous, tabular work.

**Pandas and tabular data**: a DataFrame is Python's standard in-memory representation of a table - the same mental model as a spreadsheet or a SQL result set.

**Pandas and SQL**: a DataFrame behaves like a SQL result set you can keep transforming in code - `df[df.age > 30]` is a `WHERE` clause; `df.groupby("region").sum()` is a `GROUP BY`. Section 33 gives a full mapping.

**Pandas and Excel**: Pandas can read and write Excel files, but a DataFrame is not "Excel in Python" - it has no cell formulas, no visual formatting, and (critically for ML) far better performance and reproducibility for large or repeated transformations.

**Pandas and Machine Learning**: virtually every ML project in Python passes its data through Pandas at some point between the raw source and the model - Pandas is where you inspect, clean, validate, explore, and engineer features, before finally handing NumPy arrays to scikit-learn (or another framework) for training. This entire module is about doing that handoff professionally.

**Core objects, previewed** (taught in full in Sections 3-4): a **Series** is one labeled column of data; a **DataFrame** is a table of Series sharing a common row **Index** - the row-label axis every DataFrame and Series carries.

**Comparison table**

| | Pandas DataFrame | SQL table | Excel worksheet | NumPy array |
|---|---|---|---|---|
| Structure | Labeled rows + columns, mixed dtypes per column | Rows + typed columns, one type per column | Cells, rows, columns, mostly untyped | N-dimensional, one dtype for the whole array |
| Access | Code (Python) | Query language (SQL) | Mouse/formulas, or VBA | Code (Python) |
| Scale | In-memory (millions of rows, limited by RAM) | Can scale to billions of rows on disk | Practically limited to ~1M rows | In-memory, very fast for numeric-only data |
| Best for | Data cleaning/EDA/feature engineering in code | Large-scale, persistent, concurrent-access storage | Small, human-edited data, ad hoc business use | Pure numerical computation |
| ML fit | The standard pre-model data-prep layer | The standard source-of-record data lives in | Rarely used directly for ML pipelines | The final numeric format models actually train on |

# SECTION 3 - SERIES

**Beginner explanation**: a Series is a single labeled column of data - think of one column from a spreadsheet, with row labels attached.

**Technical definition**: `pandas.Series` is a 1-dimensional labeled array capable of holding any data type, consisting of an array of values and an associated `Index` of labels.

```python
# [Python] Creating Series - the building blocks of every DataFrame column
import pandas as pd

ages = pd.Series([34, 29, 41, 52, 23], name="age")
print(ages)
# 0    34
# 1    29
# 2    41
# 3    52
# 4    23
# Name: age, dtype: int64

# With a custom, meaningful index
revenue = pd.Series([12000, 15500, 9800], index=["Jan", "Feb", "Mar"], name="monthly_revenue")
print(revenue["Feb"])   # 15500 - label-based lookup, not position-based
```

**Anatomy**: `.index` (the row labels - default is `RangeIndex(0, n)` unless specified), `.values` (the underlying NumPy array), `.dtype` (the data type of every element - `int64`, `float64`, `object`, etc.), `.name` (an optional label for the Series itself, which becomes the column name if the Series is added to a DataFrame).

```python
# [Python] Indexing, slicing, filtering, arithmetic, missing values
transaction_amounts = pd.Series([250.0, 89.5, None, 1200.0, 45.0], name="amount")

print(transaction_amounts[1])            # 89.5 - single label
print(transaction_amounts[1:3])          # slice - positions 1 up to (not including) 3
print(transaction_amounts[transaction_amounts > 100])   # Boolean filtering
print(transaction_amounts * 1.1)         # vectorized arithmetic - no loop needed
print(transaction_amounts.isna())        # True where the missing value (None -> NaN) lives
```

Series are the natural shape for an **ML target variable** - `y = df["churn"]` is a Series, not a DataFrame - and this distinction (Series vs single-column DataFrame) matters constantly when preparing `X`/`y` for scikit-learn (Section 53-54).

# SECTION 4 - DATAFRAMES

**Beginner explanation**: a DataFrame is a full table - many labeled columns (each a Series) sharing one set of row labels.

**Technical definition**: `pandas.DataFrame` is a 2-dimensional, size-mutable, potentially heterogeneous tabular data structure with labeled axes (rows and columns).

```python
# [Python] Creating DataFrames from every common source

# From a dictionary of lists - the most common manual-construction pattern
df = pd.DataFrame({
    "customer_id": [101, 102, 103],
    "age": [34, 29, 41],
    "monthly_charges": [59.99, 45.00, 89.50],
})

# From a list of dictionaries ("records") - the shape most APIs return JSON in
records = [
    {"customer_id": 101, "age": 34, "monthly_charges": 59.99},
    {"customer_id": 102, "age": 29, "monthly_charges": 45.00},
]
df_from_records = pd.DataFrame(records)

# From a NumPy array, with explicit column names
import numpy as np
arr = np.array([[101, 34, 59.99], [102, 29, 45.00]])
df_from_array = pd.DataFrame(arr, columns=["customer_id", "age", "monthly_charges"])

# From a list of Series (one per column)
df_from_series = pd.DataFrame({"age": pd.Series([34, 29]), "charges": pd.Series([59.99, 45.00])})
```

**Anatomy**: **rows** are individual observations (one customer, one transaction); **columns** are variables/features (age, monthly_charges); the **index** labels the rows; **values** is the underlying 2D array-like data; **dtypes** is a per-column data type (a DataFrame, unlike a NumPy array, can mix `int64`, `float64`, and `object` columns in one table - this is Pandas' core advantage over raw NumPy for real-world tabular data).

Realistic ML dataset shape used throughout this module: one row per customer/transaction/patient (an **observation**), one column per attribute (a **feature**), exactly the shape scikit-learn expects as input once fully prepared.

# SECTION 9 - INDEXING AND SELECTION

```python
# [Python] [], .loc[], and .iloc[] - the three selection mechanisms, and when to use each
df["age"]                      # [] - column selection by label, returns a Series
df[["age", "income"]]          # [] with a list - column selection, returns a DataFrame

df.loc[0, "age"]               # .loc - LABEL-based: row label 0, column label "age"
df.loc[0:2, ["age", "income"]] # .loc slicing is INCLUSIVE of the end label (0,1,2 all included)
df.loc[df["age"] > 30, "age"]  # .loc combined with a Boolean mask - extremely common in practice

df.iloc[0, 1]                  # .iloc - POSITION-based: first row, second column, by integer position
df.iloc[0:2, [1, 2]]           # .iloc slicing is EXCLUSIVE of the end position (rows 0,1 only)
```

**Label-based vs position-based, the single most important distinction in this section**: `.loc` always refers to the *labels* you assigned (which, if you never called `set_index`, happen to be integers `0, 1, 2...` and look deceptively similar to positions - but are NOT the same mechanism); `.iloc` always refers to *pure integer position*, regardless of what the labels actually are. This distinction becomes critical the moment your index is *not* a simple `0, 1, 2...` range - e.g. after filtering (`df[df.age > 30]`) the remaining rows keep their *original* labels, so `df_filtered.loc[0]` may raise a `KeyError` (no row with label `0` remains) while `df_filtered.iloc[0]` still safely returns the *first remaining row*, whatever its label is. This exact confusion is Section 66's single most common Pandas mistake.

# SECTION 10 - FILTERING DATA

```python
# [Python] Boolean filtering - the core mechanism behind almost every data-cleaning and EDA step in this module
adults = df[df["age"] > 30]

# Multiple conditions - each condition MUST be wrapped in parentheses (see below), combined with & | ~
high_value_seniors = df[(df["age"] > 60) & (df["monthly_charges"] > 80)]
month_to_month_or_no_internet = df[(df["contract_type"] == "Month-to-month") | (df["internet_service"] == "No")]
not_churned = df[~(df["churn"] == "Yes")]

# isin() - filter against a set of allowed values, cleaner than chaining many == with |
selected_regions = df[df["region"].isin(["North", "South", "West"])]

# between() - inclusive range filtering
working_age = df[df["age"].between(18, 65)]

# query() - a readable, string-expression alternative, often faster on large DataFrames
result = df.query("age > 30 and monthly_charges > 80")
```

**Why parentheses are mandatory around each condition**: Python's operator precedence evaluates `&`/`|` *before* comparison operators like `>`/`==` when applied to Pandas objects, unlike the plain Python `and`/`or` keywords (which cannot be used with Series at all, since a Series' truthiness is ambiguous). Writing `df[df.age > 30 & df.income > 50000]` without parentheses silently produces a different, usually wrong, result (or an error) - `df[(df.age > 30) & (df.income > 50000)]` is the required, correct form.

**Comparing the three approaches**: bracket filtering (`df[condition]`) is the most common and universally understood; `isin()` is cleaner than several chained `==`/`|` conditions; `between()` is cleaner than two chained comparisons; `query()` reads closer to SQL's `WHERE` clause and avoids repeating `df[...]` for every condition, and can be measurably faster on large DataFrames since it uses a more efficient evaluation engine internally - a good default once you're comfortable with basic bracket filtering.

# SECTION 11 - SELECTING COLUMNS

```python
# [Python] Single column vs multiple columns
single = df["age"]              # returns a Series
multiple = df[["age", "income"]] # returns a DataFrame, even if you pass just one name in the list: df[["age"]]
```

**Why single-bracket selection returns a Series but double-bracket selection returns a DataFrame**: `df["age"]` passes a single string label - Pandas interprets this as "give me the one column named age" and returns its underlying Series directly. `df[["age", "income"]]` passes a *list* of labels - even a list of one - which Pandas interprets as "give me a table containing these columns," always returning a DataFrame, never unwrapping down to a Series. This is a deliberate, consistent API design: the *type* of the argument (`str` vs `list`) determines the *type* of the result (`Series` vs `DataFrame`), and this distinction matters constantly - many scikit-learn functions expect a 2D `X` (a DataFrame or 2D array) even for a single feature, meaning `df[["age"]]` (DataFrame) is often required where `df["age"]` (Series) will raise a shape error.

# SECTION 12 - ADDING AND REMOVING COLUMNS

```python
# [Python] Creating columns
df["total_charges_per_month"] = df["total_charges"] / df["tenure_months"]   # direct assignment - the common case
df = df.assign(charges_ratio=lambda d: d["monthly_charges"] / d["total_charges"])  # assign() - chainable, functional style

# Renaming
df = df.rename(columns={"cust_id": "customer_id", "mo_charges": "monthly_charges"})

# Dropping
df = df.drop(columns=["unused_column"])
df = df.drop(["unused_column"], axis=1)   # older, equivalent syntax - axis=1 means "columns"
```

**`inplace`, and why it should not be used blindly**: many Pandas methods accept `inplace=True` (e.g. `df.drop(columns=["x"], inplace=True)`), which modifies the DataFrame directly and returns `None` instead of a new object. This seems convenient but is discouraged in modern professional code for three reasons: (1) it breaks method chaining (`df.dropna().reset_index()` reads cleanly; `inplace=True` calls cannot be chained); (2) under Pandas' Copy-on-Write behavior (Section 41), `inplace` operations may silently *not* modify the object you expected them to under certain views/copies, producing confusing bugs; (3) reassignment (`df = df.drop(...)`) makes the data flow explicit and readable - the *default modern-Pandas guidance* is to avoid `inplace=True` entirely and always reassign.

# SECTION 13 - RENAMING COLUMNS

```python
# [Python] A reusable, production-grade column-cleaning function
import re

def clean_column_names(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    df.columns = (
        df.columns
        .str.strip()                              # remove leading/trailing whitespace
        .str.lower()                               # standardize casing
        .str.replace(r"[^\w\s]", "", regex=True)     # strip punctuation like '%', '$', '#'
        .str.replace(r"\s+", "_", regex=True)          # spaces -> underscores
    )
    return df

# Before: ['Customer ID', 'Monthly Charges ($)', 'Contract Type']
# After:  ['customer_id', 'monthly_charges', 'contract_type']
df = clean_column_names(df)
```

**Professional naming standards**: lowercase, `snake_case`, no spaces (spaces force awkward `df["Customer ID"]` bracket access instead of clean `df.customer_id` attribute access), no special characters, and names that are unambiguous rather than generic (`amount` on a table with both transaction amounts and refund amounts is a landmine - prefer `transaction_amount`). Running every new dataset through one reusable cleaning function like the one above, as the very first transformation step, prevents an entire category of downstream bugs (typos when referencing a column that turn into silent `KeyError`s, or worse, silently-created *new* columns from a typo'd assignment).

# SECTION 14 - DATA TYPES

| dtype | Holds | ML relevance |
|---|---|---|
| `int64` | Whole numbers | Numeric features, IDs |
| `float64` | Decimal numbers | Numeric features, especially any column with missing values (NaN forces float) |
| `bool` | True/False | Binary flags, often directly usable as a 0/1 feature |
| `object` | Usually strings, or mixed types | Raw text - almost always needs further processing before ML |
| `string` | Pandas' newer, dedicated string dtype | A more explicit, often more memory-efficient alternative to `object` for text |
| `category` | A fixed, memory-efficient set of labels | Categorical features (Section 15) - directly relevant to encoding for ML |
| `datetime64` | Timestamps | Source for extensive date-based feature engineering (Section 21) |

```python
# [Python] Type conversion
df["customer_id"] = df["customer_id"].astype("int64")
df["signup_date"] = pd.to_datetime(df["signup_date"])
df["gender"] = df["gender"].astype("category")

# convert_dtypes() - lets Pandas infer the "best" modern dtype per column automatically
df = df.convert_dtypes()   # e.g. upgrades int64 columns with missing values to the nullable 'Int64' dtype
```

**Why dtype matters**: **memory** - an `object`-dtype column of repeated category strings can use many times more memory than the equivalent `category` dtype (Section 37); **performance** - numeric operations on `object` columns fall back to slow, per-element Python-level operations instead of fast vectorized NumPy operations; **ML preprocessing** - scikit-learn requires numeric input, so every `object`/`category` column must eventually become a number (Section 49), and the dtype you choose now determines which encoding strategy makes sense later; **data validation** - a `float64` `age` column is a signal something's wrong before you've even looked at values (ages shouldn't need decimals or be a mixed `object` type) - dtype itself is a quick validation signal.

# SECTION 15 - CATEGORICAL DATA

**Nominal categories** have no inherent order (`gender`, `payment_method`, `region`). **Ordinal categories** have a meaningful order (`contract_type`: Month-to-month < One year < Two year; `education_level`: High School < Bachelor's < Master's).

```python
# [Python] The category dtype
df["gender"] = df["gender"].astype("category")

# Ordinal categories - explicitly ordered
contract_order = pd.CategoricalDtype(categories=["Month-to-month", "One year", "Two year"], ordered=True)
df["contract_type"] = df["contract_type"].astype(contract_order)
print(df["contract_type"] < "Two year")   # now a meaningful, orderable comparison
```

**Memory benefits**: `category` dtype stores each unique value once in a lookup table and represents every row as a small integer code pointing into it - on a column like `region` with 5 unique values repeated across a million rows, this can cut memory usage by well over 90% compared to `object` dtype, which stores a full separate Python string per row (Section 37 shows this measured).

**Connection to scikit-learn**: a nominal `category` column is the natural input to scikit-learn's `OneHotEncoder` (which creates one binary column per category, appropriate when there's no order); an ordinal `category` column (with `ordered=True`, as above) is the natural input to `OrdinalEncoder` (which assigns increasing integers respecting the defined order) - deciding nominal vs ordinal correctly *in Pandas*, before encoding, is what makes the correct encoding choice in scikit-learn obvious later (Section 49).

# SECTION 17 - HANDLING MISSING VALUES

```python
# [Python] Dropping
df_dropped_rows = df.dropna()                         # drop any row with ANY missing value - often too aggressive
df_dropped_cols = df.dropna(axis=1, thresh=len(df)*0.5)  # drop columns missing more than 50% of their values

# Filling / imputation
df["age"] = df["age"].fillna(df["age"].mean())          # mean imputation
df["income"] = df["income"].fillna(df["income"].median())  # median imputation - robust to outliers, often preferred
df["region"] = df["region"].fillna(df["region"].mode()[0])   # mode imputation for categoricals
df["notes"] = df["notes"].fillna("unknown")                    # constant imputation
df_sorted = df.sort_values("date")
df_sorted["price"] = df_sorted["price"].ffill()                   # forward fill - carry the last known value forward
df_sorted["price"] = df_sorted["price"].bfill()                     # backward fill - carry the next known value backward
```

**When each is appropriate**: mean imputation suits roughly symmetric numeric distributions with no major outliers; median imputation is safer for skewed distributions (income, transaction amounts) since it's not pulled by extreme values; mode imputation suits categorical columns; constant imputation (`"unknown"`, `0`, `-1`) suits cases where "missing" is itself meaningful information worth preserving as a distinct category, and is often paired with an added indicator column (`df["income_was_missing"] = df["income"].isna()`); forward/backward fill suit ordered, time-based data where the most recent known value is a reasonable estimate (e.g. a sensor reading, a daily stock price).

**Critical: why calculating imputation statistics on the complete dataset before train/test splitting causes data leakage.** If you compute `df["age"].mean()` using *all* rows (including what will become your test set) and use that single mean to fill missing values everywhere, information from the test set (its contribution to that mean) has leaked into the training data's preprocessing - the model is now trained using a statistic that was partly informed by data it should never have seen. This inflates validation/test performance in a way that will not hold up on genuinely new, real-world data. **The correct, production-safe pattern**: split first (Section 53), then compute the mean/median/mode *only from the training set*, and apply that *same, frozen* training-set statistic to fill missing values in both the training set and the test set. This exact discipline - fit only on training data, transform everything with that fitted statistic - is precisely what scikit-learn's `Pipeline` and `SimpleImputer` are built to enforce automatically (Section 54), which is why production ML pipelines generally move imputation out of ad hoc Pandas code and into a properly fit/transform-based preprocessing step once you're past exploratory work.

**Exact duplicates** are byte-identical repeated rows, often caused by a pipeline re-ingesting the same file twice. **Business-key duplicates** are more dangerous and more common: two rows that differ in some columns (e.g. a slightly updated `email` or `last_login`) but share the same real-world identity (`customer_id`) - these will not be caught by `df.duplicated()`'s default (which compares *every* column) and require an explicit `subset` argument targeting the actual business key. **Repeated transactions** (the same purchase logged twice due to a retry) and **duplicate customers** (the same person onboarded twice with slightly different spelling, a Master Data Management problem covered in the Data Governance handbook, Part 38) both distort ML training identically: any duplicated row is effectively over-weighted in whatever the model learns, silently skewing it toward whatever pattern that duplicated record represents, and inflating apparent model confidence on patterns that are really just the same row counted multiple times.

# SECTION 19 - DATA VALIDATION

```python
# [Python] Professional validation functions - the core discipline behind trustworthy ML datasets
def validate_age(df: pd.DataFrame) -> pd.DataFrame:
    violations = df[(df["age"] < 0) | (df["age"] > 120)]
    if not violations.empty:
        print(f"WARNING: {len(violations)} rows have an impossible age")
    return violations

def validate_amount_non_negative(df: pd.DataFrame, col: str) -> pd.DataFrame:
    return df[df[col] < 0]

def validate_allowed_categories(df: pd.DataFrame, col: str, allowed: set) -> pd.DataFrame:
    return df[~df[col].isin(allowed)]

def validate_dates(df: pd.DataFrame, col: str) -> pd.DataFrame:
    parsed = pd.to_datetime(df[col], errors="coerce")
    return df[parsed.isna() & df[col].notna()]   # rows that had a value but failed to parse as a date

def validate_target_classes(df: pd.DataFrame, target: str, expected: set) -> pd.DataFrame:
    return df[~df[target].isin(expected)]

# Usage - a validation suite run before any data is considered "trusted"
age_violations = validate_age(df)
amount_violations = validate_amount_non_negative(df, "transaction_amount")
gender_violations = validate_allowed_categories(df, "gender", {"Male", "Female", "Other"})
target_violations = validate_target_classes(df, "churn", {"Yes", "No"})
```

**Introducing schema validation**: each function above checks one specific rule; a professional workflow runs the *whole set* together and summarizes the results before any modeling begins - this is exactly the idea behind dedicated schema-validation libraries (like Pandera or Great Expectations, covered in depth in the Data Governance handbook) which let you declare a full schema (column names, types, allowed ranges, nullability) once and validate a DataFrame against it in a single call. Understanding *why* schema validation matters - and being able to hand-roll it in plain Pandas as above - is the right foundation before reaching for one of those libraries.

# SECTION 22 - TIME-BASED DATA

```python
# [Python] Sorting, resampling, rolling windows, shifting - the toolkit for time-series feature engineering
df = df.sort_values("transaction_date")   # ALWAYS sort chronologically before any time-based operation

# Resampling - aggregate irregular timestamps into a regular frequency
daily_totals = df.set_index("transaction_date")["amount"].resample("D").sum()
monthly_avg = df.set_index("transaction_date")["amount"].resample("ME").mean()

# Rolling windows - a moving calculation over the last N rows/periods
df["rolling_7day_avg"] = df.set_index("transaction_date")["amount"].rolling("7D").mean().values

# Shifting - access a prior row's value, the foundation of lag features (Section 47)
df["previous_amount"] = df.groupby("customer_id")["amount"].shift(1)

# Time-based grouping
monthly_by_customer = df.groupby([df["transaction_date"].dt.to_period("M"), "customer_id"])["amount"].sum()
```

**How these support time-series ML**: `resample()` turns irregular event data (transactions happening at arbitrary times) into a regular series (one value per day/week/month) that time-series models and rolling-feature calculations expect; `rolling()` computes moving statistics (a 7-day average spend) that capture recent trend, a staple feature in fraud and demand-forecasting models; `shift()` looks backward in time to build lag features.

**Warning about future information leakage**: every one of these operations must be computed using **only data available up to and including the point in time each row represents** - a rolling average that accidentally includes *future* transactions (e.g. because the data wasn't sorted first, or a window was defined incorrectly) hands the model information it would never have access to in live production, producing exactly the same class of data leakage covered in depth in Section 52. Always sort by time first, always double-check a rolling/shift operation's direction, and always mentally ask "would I actually know this value at prediction time in production?"

# SECTION 26 - GROUPBY

**This is a major section.** `groupby()` is the single most powerful and most frequently used Pandas operation in professional data work - it implements the "split-apply-combine" pattern: split the data into groups by some key, apply a computation to each group independently, then combine the results back into one structure.

```python
# [Python] Basic aggregation
revenue_by_region = df.groupby("region")["revenue"].sum()
avg_salary_by_dept = df.groupby("department")["salary"].mean()
churn_by_contract = df.groupby("contract_type")["churn"].apply(lambda s: (s == "Yes").mean())

# Multiple aggregations at once
summary = df.groupby("region")["revenue"].agg(["sum", "mean", "count", "std"])

# Named aggregation - clean, explicit column naming for multiple metrics across multiple columns
summary_named = df.groupby("region").agg(
    total_revenue=("revenue", "sum"),
    avg_revenue=("revenue", "mean"),
    n_customers=("customer_id", "nunique"),
)

# groupby + transform - see Section 28 for the full treatment
df["region_avg_revenue"] = df.groupby("region")["revenue"].transform("mean")

# groupby + filter - keep only groups meeting a condition
large_regions = df.groupby("region").filter(lambda g: len(g) > 100)

# Multi-level grouping
by_region_and_contract = df.groupby(["region", "contract_type"])["revenue"].sum()
fraud_rate_by_merchant_category = df.groupby("merchant_category")["is_fraud"].mean()
```

**Line by line, on the named-aggregation example**: `df.groupby("region")` splits the DataFrame into one sub-DataFrame per distinct region; `.agg(total_revenue=("revenue", "sum"), ...)` applies `sum` to the `revenue` column of each group and names the result column `total_revenue`, `mean` to `revenue` named `avg_revenue`, and `nunique` to `customer_id` named `n_customers`, all in one call; the result is one row per region with three clearly-named summary columns - this named-aggregation syntax is the modern, professional standard, dramatically more readable than older approaches that produced ambiguous multi-index column names.

**Realistic business examples used throughout this module**: revenue by region, average salary by department, churn rate by contract type, fraud rate by merchant category - every one of these is a single `groupby().agg()` call, and every one is a genuinely common real-world analytical question.

# SECTION 27 - AGGREGATION

```python
# [Python] Core aggregation functions
df.groupby("region")["revenue"].sum()
df.groupby("region")["revenue"].mean()
df.groupby("region")["revenue"].median()
df.groupby("region")["revenue"].min()
df.groupby("region")["revenue"].max()
df.groupby("region")["revenue"].count()   # counts NON-NULL values
df.groupby("region").size()                # counts ALL rows per group, including those with nulls in other columns
df.groupby("region")["revenue"].std()
df.groupby("region")["revenue"].var()

# agg() with multiple functions and multiple columns at once
df.groupby("region").agg({"revenue": ["sum", "mean"], "customer_id": "nunique"})
```

# SECTION 28 - GROUPBY TRANSFORM

**The difference between `aggregate()` and `transform()`**: `aggregate()` (or `agg()`) collapses each group down to a single summary row - the *output has fewer rows than the input*. `transform()` computes the same kind of group-level statistic but *broadcasts it back to every original row* - the *output has exactly the same number of rows as the input*, making it ideal for adding a new feature column without losing row-level granularity.

```python
# [Python] Group mean, group standardization, and group-level features via transform()

# Group mean, broadcast back to every row
df["region_avg_revenue"] = df.groupby("region")["revenue"].transform("mean")

# Group standardization (z-score within each group) - a genuinely powerful feature-engineering pattern
group_mean = df.groupby("region")["revenue"].transform("mean")
group_std = df.groupby("region")["revenue"].transform("std")
df["revenue_zscore_within_region"] = (df["revenue"] - group_mean) / group_std

# Customer spending relative to their region's average spending - directly usable as an ML feature
df["spend_vs_region_avg"] = df["monthly_charges"] / df.groupby("region")["monthly_charges"].transform("mean")
```

**Why `transform` is so useful for feature engineering**: it lets you compare an individual observation to its *own group's* context, which is often far more predictive than the raw value alone - a customer spending $200/month means something different in a region where the average is $50 versus a region where the average is $180, and `spend_vs_region_avg` captures exactly that relative signal in a single feature, at row-level granularity ready to feed directly into a model, unlike `agg()` which would only give you the region-level averages themselves.

# SECTION 29 - APPLY VS MAP VS ELEMENTWISE OPERATIONS

```python
# [Python] map() - element-wise, Series only, great for simple value substitution/lookup
mapping = {"Yes": 1, "No": 0}
df["churn_binary"] = df["churn"].map(mapping)

# apply() - Series or DataFrame, runs an arbitrary function per element, row, or column
df["age_group"] = df["age"].apply(lambda x: "senior" if x >= 60 else "adult")
df["full_name"] = df.apply(lambda row: f"{row['first_name']} {row['last_name']}", axis=1)   # row-wise, axis=1

# Vectorized operation - the preferred approach whenever a built-in vectorized alternative exists
df["age_group_vectorized"] = np.where(df["age"] >= 60, "senior", "adult")   # equivalent to the apply() above, much faster
```

```python
# [Python] SLOW vs EFFICIENT, measured
import time

# Slow: apply() with a Python function, called once per row
start = time.time()
df["doubled_slow"] = df["amount"].apply(lambda x: x * 2)
print(f"apply: {time.time() - start:.4f}s")

# Efficient: vectorized arithmetic, operates on the whole column via compiled NumPy code at once
start = time.time()
df["doubled_fast"] = df["amount"] * 2
print(f"vectorized: {time.time() - start:.4f}s")
# On a large DataFrame, the vectorized version is commonly 10-100x faster
```

**When vectorized operations should be preferred**: essentially always, whenever one exists - `apply()` and `map()` still execute a Python-level function call per element/row under the hood (even though they're more convenient to write than a raw `for` loop), while true vectorized operations (`+`, `*`, `np.where`, `.str.*`, `.dt.*`, built-in aggregation functions) push the entire computation down into compiled C/NumPy code operating on the whole array at once. Reach for `apply()` only when the logic is genuinely too complex to express with vectorized building blocks (e.g. calling an external function per row that has no vectorized equivalent) - and even then, check whether `np.vectorize`, `.map()` with a dict, or restructuring the logic into vectorized pieces can eliminate the need entirely. This performance gap is measured concretely in Section 67's performance lab.

# SECTION 30 - MERGING DATAFRAMES

```python
# [Python] merge() - the DataFrame equivalent of a SQL JOIN
customers = pd.DataFrame({"customer_id": [1, 2, 3], "region": ["North", "South", "North"]})
transactions = pd.DataFrame({"customer_id": [1, 1, 2, 4], "amount": [100, 50, 75, 30]})

inner = pd.merge(customers, transactions, on="customer_id", how="inner")   # only matching customer_ids in BOTH
left = pd.merge(customers, transactions, on="customer_id", how="left")     # all customers, unmatched transactions -> NaN
right = pd.merge(customers, transactions, on="customer_id", how="right")   # all transactions, unmatched customers -> NaN
outer = pd.merge(customers, transactions, on="customer_id", how="outer")   # everything from both sides
cross = pd.merge(customers, transactions, how="cross")                      # every combination of both tables (rare)
```

**Key vocabulary**: a **primary key** uniquely identifies a row in its own table (`customer_id` in `customers`); a **foreign key** in another table (`customer_id` in `transactions`) references that primary key; the **join key** is the column(s) merge actually matches on (specified via `on=`, or `left_on=`/`right_on=` if the column names differ between the two tables).

**Realistic tables used throughout this module**: `customers`, `transactions`, `accounts`, `products` - merging `customers` with `transactions` on `customer_id` to build customer-level features (Section 73) is one of the most common operations in this entire module.

**Direct connection to SQL**: `pd.merge(a, b, on="key", how="left")` is exactly `SELECT * FROM a LEFT JOIN b ON a.key = b.key` - Section 33 gives the complete SQL-to-Pandas mapping, and understanding merge deeply here is what makes that mapping click immediately.

# SECTION 31 - JOINING DATA

```python
# [Python] join() - a convenience wrapper around merge(), joining primarily on the INDEX
customers_indexed = customers.set_index("customer_id")
accounts_indexed = accounts.set_index("customer_id")

joined = customers_indexed.join(accounts_indexed, how="left")   # joins on the shared index by default
```

**`merge()` vs `join()`**: `merge()` is the general-purpose tool, joining on any specified column(s) from either table (`on=`, `left_on=`/`right_on=`); `join()` is a more convenient shorthand specifically for joining on the **index** (or a mix of index and columns) - functionally, `join()` calls `merge()` internally under the hood. In practice, most professional code defaults to `merge()` for its explicitness (the join key is always visible in the call) and reserves `join()` for the specific, common case of combining several DataFrames that already share a meaningful index (e.g. several feature tables all indexed by `customer_id`, joined together in a chain).

# SECTION 32 - CONCATENATING DATA

```python
# [Python] pd.concat() - stacking DataFrames, not matching on keys
jan_data = pd.DataFrame({"customer_id": [1, 2], "revenue": [100, 150]})
feb_data = pd.DataFrame({"customer_id": [3, 4], "revenue": [200, 90]})

# Row-wise concatenation - stacking two months of data with the SAME columns into one longer table
all_data = pd.concat([jan_data, feb_data], axis=0, ignore_index=True)

# Column-wise concatenation - combining two tables with the SAME rows but DIFFERENT columns, aligned by index
demographics = pd.DataFrame({"age": [34, 29]}, index=[1, 2])
spend = pd.DataFrame({"total_spend": [500, 300]}, index=[1, 2])
combined = pd.concat([demographics, spend], axis=1)
```

**When concatenation is appropriate vs merge**: use `concat()` when you're simply *stacking* datasets that share the same structure (12 monthly CSV files with identical columns, appended into one year of data) - there's no "matching" logic involved, just alignment by position/index. Use `merge()` when you're *combining different information about the same entities* based on a shared key (customer demographics + their transaction history) - this is a fundamentally different operation, and using `concat(axis=1)` when you actually need a keyed `merge()` is a common mistake that silently misaligns rows if the two DataFrames' indexes don't correspond to the same entities in the same order.

# SECTION 33 - SQL VS PANDAS

| SQL | Pandas equivalent |
|---|---|
| `SELECT col1, col2 FROM t` | `df[["col1", "col2"]]` |
| `SELECT * FROM t WHERE age > 30` | `df[df["age"] > 30]` |
| `SELECT * FROM t WHERE age > 30 AND region = 'North'` | `df[(df["age"] > 30) & (df["region"] == "North")]` |
| `SELECT DISTINCT region FROM t` | `df["region"].unique()` |
| `SELECT region, COUNT(*) FROM t GROUP BY region` | `df.groupby("region").size()` |
| `SELECT region, SUM(revenue) FROM t GROUP BY region` | `df.groupby("region")["revenue"].sum()` |
| `SELECT region, SUM(revenue) FROM t GROUP BY region HAVING SUM(revenue) > 1000` | `df.groupby("region")["revenue"].sum().loc[lambda s: s > 1000]` |
| `SELECT * FROM t ORDER BY age DESC` | `df.sort_values("age", ascending=False)` |
| `SELECT a.*, b.amount FROM a JOIN b ON a.id = b.id` | `pd.merge(a, b, on="id", how="inner")` |
| `SELECT AVG(revenue) FROM t` | `df["revenue"].mean()` |
| `CASE WHEN age >= 60 THEN 'senior' ELSE 'adult' END` | `np.where(df["age"] >= 60, "senior", "adult")` |

**Where SQL should be preferred over Pandas**: when the data already lives in a database, especially when it's too large to comfortably fit in memory - SQL pushes filtering, joining, and aggregation down to the database engine, which is purpose-built for exactly that at scale, and only pulls the (usually much smaller) *result* into Pandas via `read_sql()` (Section 34). A common and important professional pattern: do as much filtering/aggregation as possible in SQL first, and use Pandas for the remaining row-level cleaning, feature engineering, and ML-specific preparation that SQL is more awkward for.

# SECTION 36 - PARQUET

**Columnar storage**: Parquet stores each column's values contiguously on disk (rather than row-by-row, as CSV does), so a query/load that only needs 3 of 50 columns physically reads only those columns' data. **Compression**: because similar values are stored together within a column, Parquet compresses dramatically better than row-oriented formats. **Schema**: Parquet embeds its schema (column names and types) directly in the file - no re-inferring types on every load, unlike CSV. **Performance**: the combination of columnar storage, compression, and embedded schema makes Parquet reads/writes both faster and smaller than CSV, often by a large margin at real dataset sizes.

| | CSV | Parquet |
|---|---|---|
| Human-readable | Yes | No |
| Schema embedded | No (inferred on every read) | Yes |
| Compression | None natively | Built-in, efficient |
| Column-selective reads | No (must parse every column) | Yes (reads only requested columns) |
| Typical file size | Larger | Often 5-10x smaller for the same data |
| Load speed at scale | Slower | Faster |

```python
# [Python]
df.to_parquet("customers.parquet", index=False)
df = pd.read_parquet("customers.parquet", columns=["customer_id", "age", "monthly_charges"])   # column-selective read
```

**Why Parquet matters in modern Data Engineering and ML**: it is the default interchange and storage format for data lakes and lakehouses (Data Engineering handbook, Part 15) precisely because of these properties - faster iteration when developing feature engineering code, lower storage cost at scale, and a schema you can trust without re-validating types on every load. Any ML dataset expected to be reloaded repeatedly (rather than a one-off CSV export) should be stored as Parquet as a default professional habit.

# SECTION 38 - LARGE DATASETS

**Pandas' fundamental limitation**: it is an in-memory tool - a DataFrame (plus the overhead of any intermediate copies your code creates) must fit within available RAM, which becomes the hard ceiling as datasets grow into the tens of millions of rows or more.

```python
# [Python] Chunked reading - process a file too large to load all at once, piece by piece
total_revenue = 0
for chunk in pd.read_csv("huge_transactions.csv", chunksize=500_000):
    chunk = chunk[chunk["amount"] > 0]          # filter early, within each chunk
    total_revenue += chunk["amount"].sum()       # accumulate a running result instead of holding everything in memory
```

**The full toolkit for working within Pandas' limits**: **`chunksize`** (as above) processes a file in bounded pieces, never holding the whole thing in memory at once; **selecting only needed columns** (`usecols`, Section 7) reduces memory before it's even loaded; **filtering early** (as in the chunk loop above) discards irrelevant rows as soon as possible rather than carrying them through every subsequent step; **efficient dtypes** (Section 37) shrink whatever does stay in memory; **Parquet** (Section 36) both loads faster and supports column-selective reads natively; **database aggregation** (Section 33-34) pushes the heavy lifting to a system built for out-of-core data entirely, pulling back only a small, already-aggregated result. When even these techniques aren't enough, the right move is not to fight Pandas harder but to reach for a genuinely distributed/out-of-core tool (Section 63) - Spark, Dask, or Polars' lazy execution mode (Section 62).

# SECTION 39 - PANDAS PERFORMANCE

```python
# [Python] The performance hierarchy, from slowest to fastest, measured on the same task
import time

df_large = pd.DataFrame({"amount": np.random.rand(200_000) * 1000})

# 1. Plain Python loop with .iloc - by far the slowest, avoid entirely
start = time.time()
result = []
for i in range(len(df_large)):
    result.append(df_large.iloc[i]["amount"] * 1.1)
print(f"loop + iloc: {time.time()-start:.3f}s")

# 2. iterrows() - still slow, one Python-level Series object created PER ROW
start = time.time()
result = [row["amount"] * 1.1 for _, row in df_large.iterrows()]
print(f"iterrows: {time.time()-start:.3f}s")

# 3. itertuples() - faster than iterrows(), avoids Series overhead, still a Python-level loop
start = time.time()
result = [row.amount * 1.1 for row in df_large.itertuples()]
print(f"itertuples: {time.time()-start:.3f}s")

# 4. apply() - still a per-row Python function call under the hood
start = time.time()
result = df_large["amount"].apply(lambda x: x * 1.1)
print(f"apply: {time.time()-start:.3f}s")

# 5. Vectorization - by far the fastest, operates on the whole column at once in compiled code
start = time.time()
result = df_large["amount"] * 1.1
print(f"vectorized: {time.time()-start:.3f}s")
```

**Typical relative results** (exact numbers vary by machine, but the *ordering* is highly consistent): `loop+iloc` and `iterrows()` are commonly 50-200x slower than vectorization; `itertuples()` is meaningfully faster than `iterrows()` (no per-row Series construction overhead) but still far behind vectorization; `apply()` sits roughly between `itertuples()` and full vectorization depending on the operation.

**Why `iterrows()` is generally not preferred for large numerical transformations**: it constructs a full, type-inferred Series object for every single row, which is expensive - and worse, that per-row Series often loses the original per-column dtypes (mixing int and float columns in one row can silently upcast everything to `float64` within that row's Series), a subtle correctness trap on top of the performance cost. The professional default, in order of preference, is: vectorized operation first, `.str`/`.dt` accessor or `np.where`/`np.select` second, `apply()` third (only when no vectorized path exists), `itertuples()` fourth (only when you genuinely need row-by-row imperative logic apply() can't express), and `iterrows()`/manual index loops essentially never in performance-sensitive code.

# SECTION 40 - COPY VS VIEW

```python
# [Python] The chained-assignment trap
subset = df[df["age"] > 30]           # this MAY be a view or a copy - Pandas doesn't guarantee which
subset["flag"] = True                  # this can raise SettingWithCopyWarning - is it modifying df or a temporary copy?

# The safe, explicit fix
subset = df[df["age"] > 30].copy()      # .copy() makes the intent unambiguous - a fully independent new DataFrame
subset["flag"] = True                     # now unambiguously safe
```

**`SettingWithCopyWarning`**: Pandas raises this warning when you assign into a DataFrame that *might* be a view into another DataFrame, because it genuinely cannot always tell whether your assignment will affect the original data or a disconnected temporary object - a correctness ambiguity, not just a style nitpick. **Why explicit copying makes code clearer**: calling `.copy()` immediately after any filter/selection you intend to modify removes all ambiguity - both for Pandas' internal behavior and for the next human reading your code, who no longer has to reason about whether `subset` and `df` are secretly linked. The professional default: call `.copy()` explicitly any time you filter or select and then plan to assign new columns into the result, rather than hoping Pandas' internal view/copy heuristics happen to do what you want.

# SECTION 41 - COPY-ON-WRITE

**Why it exists**: historically, Pandas' view-vs-copy behavior (Section 40) was inconsistent and hard to predict - sometimes a slice returned a view (modifying it silently changed the original), sometimes a copy (it didn't), depending on internal details most users couldn't reasonably anticipate. **Copy-on-Write (CoW)**, the default behavior in modern Pandas (2.x+), makes this fully consistent: every DataFrame/Series you get back from an operation behaves *as if* it were an independent copy - reading is cheap and shared internally for efficiency, but the moment you *write* to it, Pandas transparently makes a real, private copy first, so your modification can never unexpectedly leak back into some other DataFrame you didn't mean to touch.

```python
# [Python] Copy-on-Write behavior, version-appropriate example
pd.set_option("mode.copy_on_write", True)   # the default in modern Pandas; shown explicitly for clarity

df = pd.DataFrame({"age": [25, 35, 45]})
subset = df[df["age"] > 30]        # under CoW, this is safe and unambiguous
subset.loc[:, "flag"] = True        # modifies subset only - df is NEVER silently affected, unlike pre-CoW behavior

print(df.columns.tolist())          # ['age'] - unaffected
print(subset.columns.tolist())      # ['age', 'flag']
```

**How it affects views/copies and assignment behavior**: under CoW, `SettingWithCopyWarning` becomes far less common and far less ambiguous in meaning, because the underlying uncertainty it used to warn about is largely eliminated - assignments behave predictably regardless of whether Pandas happened to give you a view or a copy internally. **Memory implications**: CoW is efficient because the actual copy only happens lazily, at the moment a write occurs (not eagerly every time you slice) - reading and passing DataFrames around remains cheap; only genuine modification triggers the memory cost of copying. **Why explicit `.copy()` calls (Section 40) are still good practice even under CoW**: they document *intent* clearly for human readers, and remain fully correct and harmless regardless of the underlying CoW implementation details - a habit worth keeping even as the underlying mechanics that originally motivated it have improved.

# SECTION 46 - WINDOW OPERATIONS

```python
# [Python]
df = df.sort_values("date")

# rolling() - a moving window of a FIXED size, sliding across the data
df["moving_avg_7"] = df["sales"].rolling(window=7).mean()

# expanding() - a window that GROWS to include everything up to the current row (cumulative)
df["cumulative_avg"] = df["sales"].expanding().mean()

# ewm() - exponentially weighted, giving more influence to RECENT observations than a flat rolling average does
df["ewm_avg"] = df["sales"].ewm(span=7).mean()
```

**Use cases and connection to time-series ML**: a **moving average** (`rolling`) smooths out day-to-day noise to reveal underlying trend, and is itself a common ML feature (recent average spend, recent average transaction size); a **cumulative metric** (`expanding`) is useful for running totals (total spend to date) that grow monotonically over a customer's history; an **exponentially weighted value** (`ewm`) is useful when recent behavior should matter more than older behavior - common in fraud detection (a sudden change from a customer's *recent* typical pattern matters more than their pattern from a year ago) and in financial time series. # SECTION 47 - LAG AND LEAD FEATURES

```python
# [Python] shift() - the foundation of lag features
df = df.sort_values(["customer_id", "date"])

df["lag_1"] = df.groupby("customer_id")["amount"].shift(1)     # the previous transaction's amount
df["lag_7"] = df.groupby("customer_id")["amount"].shift(7)      # the amount 7 transactions ago
df["lag_30"] = df.groupby("customer_id")["amount"].shift(30)     # the amount 30 transactions ago
```

**Explaining future leakage very carefully**: `shift(1)` looks *backward* (the previous row's value) - this is safe. `shift(-1)` looks *forward* (the *next* row's value) - using this to build a "feature" is a textbook example of **future information leakage**: at the moment you'd actually need to make a prediction in production, the next transaction hasn't happened yet, so a feature built from `shift(-1)` simply cannot exist in real deployment, even though it's trivially computable when you have the complete historical dataset sitting in front of you during training. A model trained with such a feature will show implausibly excellent validation performance and then fail in production, because the feature that made it "work" never actually exists at prediction time.

```python
# [Python] Correct chronological splitting for time-dependent ML - never a random shuffle
df = df.sort_values("date")
split_date = "2026-06-01"
train = df[df["date"] < split_date]
test = df[df["date"] >= split_date]
# Every lag/rolling feature in `train` must only ever reference data also within `train`'s own time range -
# a feature computed using the full dataset BEFORE this split, then split afterward, has already leaked
# test-period information into training (see Section 52 for the general principle in full).
```

# SECTION 48 - FEATURE ENGINEERING WITH PANDAS

**This is one of the most important sections in this module.**

```python
# [Python] The core feature-engineering patterns, all directly reusable

# Ratios
df["profit_margin"] = (df["revenue"] - df["cost"]) / df["revenue"]
df["conversion_rate"] = df["conversions"] / df["visits"]

# Differences
df["price_change"] = df["current_price"] - df["previous_price"]

# Percentages
df["pct_of_total_spend"] = df["category_spend"] / df["total_spend"] * 100

# Bins - turning a continuous variable into a categorical one
df["age_bucket"] = pd.cut(df["age"], bins=[0, 18, 35, 60, 120], labels=["minor", "young_adult", "adult", "senior"])
df["income_quartile"] = pd.qcut(df["income"], q=4, labels=["Q1", "Q2", "Q3", "Q4"])

# Interaction features - capturing a relationship between two variables a model can't infer on its own
df["age_x_income"] = df["age"] * df["income"]

# Date features (Section 21)
df["customer_tenure_days"] = (pd.Timestamp.now() - df["signup_date"]).dt.days

# Aggregated / frequency features (Sections 26-28)
df["transaction_frequency"] = df.groupby("customer_id")["transaction_id"].transform("count")
df["avg_order_value"] = df.groupby("customer_id")["amount"].transform("mean")

# Rolling / lag features (Sections 46-47)
df["rolling_7day_spend"] = df.groupby("customer_id")["amount"].transform(lambda s: s.rolling(7, min_periods=1).sum())
```

# SECTION 50 - NUMERICAL FEATURE ENGINEERING

```python
# [Python]
# Ratios (see also Section 48)
df["debt_to_income"] = df["total_debt"] / df["income"]

# Logarithmic transformation - conceptually: compresses right-skewed distributions (income, transaction amounts)
df["log_income"] = np.log1p(df["income"])   # log1p = log(1+x), safely handles zero values unlike plain log(x)

# Binning (see also Section 48)
df["charges_bucket"] = pd.cut(df["monthly_charges"], bins=5)

# Preparing for scaling - Pandas gets the RAW values correct and clean; scikit-learn's scalers (Section 54) do the actual scaling
# (StandardScaler, MinMaxScaler are fit on training data only, exactly like imputation in Section 17)

# Interaction variables (see also Section 48)
df["tenure_x_charges"] = df["tenure_months"] * df["monthly_charges"]

# Clipping - capping extreme values without deleting the rows entirely
df["income_clipped"] = df["income"].clip(lower=df["income"].quantile(0.01), upper=df["income"].quantile(0.99))
```

**Why log transformation matters conceptually**: many real-world numeric features (income, transaction amounts, house prices) are strongly right-skewed - a small number of very large values stretch the distribution and can disproportionately influence certain models (particularly linear models and distance-based methods). A log transform compresses that skew toward a more symmetric, closer-to-normal shape, often improving model performance and stability - **this is feature engineering, not the same operation as scikit-learn's `StandardScaler`**, which centers and rescales (mean 0, unit variance) without addressing skew shape at all; the two are frequently used together (log-transform first, then scale). **Clipping** is a gentler alternative to deleting outliers (Section 51) entirely - it caps extreme values at a reasonable boundary while keeping every row and its other feature values intact.

# SECTION 51 - OUTLIER DETECTION WITH PANDAS

```python
# [Python] IQR (Interquartile Range) method - the most common, robust general-purpose approach
Q1 = df["income"].quantile(0.25)
Q3 = df["income"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers_iqr = df[(df["income"] < lower_bound) | (df["income"] > upper_bound)]

# Percentile thresholds - a simpler, direct approach
lower_pct = df["income"].quantile(0.01)
upper_pct = df["income"].quantile(0.99)
outliers_pct = df[(df["income"] < lower_pct) | (df["income"] > upper_pct)]

# Z-score concept - how many standard deviations a value is from the mean
mean, std = df["income"].mean(), df["income"].std()
df["income_zscore"] = (df["income"] - mean) / std
outliers_z = df[df["income_zscore"].abs() > 3]
```

**Do not automatically delete every outlier.** An extreme value can be one of several very different things, each requiring a different response: a **genuine extreme observation** (a legitimately very high-earning customer - real, valuable signal, deleting it discards true information about your population); a **data-entry error** (an age of `999`, clearly wrong - correcting or dropping this is appropriate); **fraud** (an unusually large or unusual transaction is often exactly the pattern a fraud model exists to detect - deleting fraud-indicative outliers as "noise" would be actively counterproductive, removing the very signal the model needs to learn); a **rare event** (a legitimate but uncommon pattern that a model genuinely should learn to recognize, not discard as noise). **Connection to ML**: before deleting or capping any outlier, ask what it represents and what the model is actually meant to predict - for a fraud, medical-risk, or anomaly-detection model, extreme values are frequently the entire point of the exercise, not noise to be cleaned away.

# SECTION 52 - DATA LEAKAGE

**A dedicated, detailed chapter** - this is one of the most consequential mistakes in applied ML, and Pandas code is very often where it silently happens.

| Leakage type | Definition | Realistic example |
|---|---|---|
| Target leakage | A feature is itself derived from (or a proxy for) the target, and wouldn't be known before the target is determined | Using `"was_refunded"` to predict `"is_fraud"` when refunds only happen *after* fraud is confirmed |
| Train/test contamination | Information from the test set influences training in any way | Fitting a scaler or imputer on the full dataset before splitting (Section 17) |
| Temporal leakage | A feature uses information from after the prediction point in time | A `shift(-1)` "lag" feature (Section 47), or a rolling average computed without sorting by date first |
| Preprocessing leakage | Any preprocessing statistic (mean, encoding, scaling parameters) computed using data outside the training set | Computing a category's target-encoding average using the full dataset, not just training rows |
| Aggregation leakage | A customer/entity-level aggregate feature is computed using data that spans across the train/test time boundary | `df.groupby("customer_id")["amount"].mean()` computed over ALL transactions, then used as a feature for predicting an EARLIER transaction |
| Duplicate leakage | The same real-world entity (or a near-duplicate row) appears in both the training and test sets | A customer's data appears twice due to an unresolved duplicate (Section 18), once landing in train and once in test |

**Worked example - aggregation leakage, explained in full**: suppose you're predicting whether a customer will churn, and you engineer a feature `avg_monthly_spend = df.groupby("customer_id")["amount"].transform("mean")`, computed by averaging **every** transaction that customer ever made - including transactions that happened *after* the point in time you're supposedly making the prediction from. This average now implicitly "knows" about the customer's future spending behavior (including, potentially, behavior that happened only because they hadn't churned yet) - a form of temporal leakage layered inside an aggregation feature. **Why this produces unrealistic model performance**: the model appears to perform extremely well during validation, because the leaked feature is trivially, almost circularly, informative - but in true production deployment, at the actual moment a prediction is needed, that future information simply does not exist yet, and the model's real-world performance collapses far below what validation suggested.

**The general defense, restated one final time**: (1) always split by time (or a stable, leakage-safe random split) *before* computing any statistic; (2) fit every preprocessing statistic - mean, encoding map, scaler parameters - on the training set only; (3) for any feature involving aggregation or a rolling/lag window, explicitly verify it only uses information available strictly before the relevant prediction point; (4) treat every new engineered feature with a moment of suspicion: "would I actually have this exact value, computed this exact way, at the real moment I'd need to make this prediction in production?" If the honest answer is no, the feature leaks.

# SECTION 55 - DATA PREPARATION PIPELINE

```
Raw Data -> Validation -> Cleaning -> Feature Engineering -> Split -> Preprocessing -> Model
```

| Stage | Belongs in |
|---|---|
| Raw data ingestion | Pandas (`read_csv`/`read_sql`/`read_parquet`, Section 7, 34) |
| Validation | Pandas (Section 19) - schema, ranges, allowed categories, checked BEFORE any transformation |
| Cleaning | Pandas (Sections 12-18, 20) - column names, dtypes, duplicates - structural, not statistics-dependent |
| Feature engineering | Pandas (Sections 48-50) - ratios, dates, categorical prep - mostly structural, domain-driven logic |
| **Split** | scikit-learn's `train_test_split` (Section 53) - **the dividing line** |
| Preprocessing (imputation, scaling, encoding) | scikit-learn `Pipeline`/`ColumnTransformer` - fit on training data ONLY (Section 17, 52, 54) |
| Model | scikit-learn (or another ML framework) |

**The dividing line, explained**: everything *before* the train/test split is safe to do once, on the whole dataset, in Pandas - because it doesn't compute any statistic that could leak information between train and test (renaming a column, parsing a date, or computing a ratio between two columns *within the same row* carries no cross-row information). Everything that computes a statistic *across rows* (a mean for imputation, a category's frequency for encoding, a scaler's mean/std) must happen *after* the split, fit on training data only - which is exactly why scikit-learn's `Pipeline` object exists: it packages these fit-dependent steps together with the model itself, so the discipline is enforced by the tool rather than remembered by the engineer every single time.

# SECTION 62 - PANDAS + POLARS

**A neutral technical introduction.** Polars is a newer DataFrame library, written in Rust, designed from the ground up for performance and (optionally) lazy, query-optimized execution.

| | Pandas | Polars |
|---|---|---|
| Syntax | `df[df.age > 30]`, `.loc`/`.iloc` | Similar conceptually but a distinct API (`pl.col("age") > 30`) - not a drop-in replacement |
| Performance | Single-threaded by default for most operations | Multi-threaded by default; often substantially faster, especially on large data |
| Lazy execution | No - operations run immediately (eager) | Yes, optionally - `pl.scan_csv()` builds a query plan, optimized and executed only when `.collect()` is called |
| Memory | Higher overhead per operation, more intermediate copies historically | Generally more memory-efficient |
| Ecosystem | Massive - the default choice most tutorials, libraries, and Stack Overflow answers assume | Smaller but rapidly growing; not every library integrates with it as seamlessly yet |
| Compatibility | The long-standing standard; scikit-learn and most ML tooling assume Pandas/NumPy input | Growing but not universal support - often converted to Pandas or NumPy before reaching scikit-learn anyway |

**When a professional might consider Polars**: genuinely large datasets (tens of millions of rows and beyond) where Pandas' single-threaded performance becomes a real bottleneck; a pipeline that benefits from lazy, query-plan-optimized execution (similar in spirit to how a SQL query planner works, Section 33); a greenfield project with no legacy dependency on Pandas-specific APIs. This module remains Pandas-focused deliberately, because Pandas is still the default, most broadly compatible tool across the ML ecosystem this curriculum builds toward (Module 5 onward) - understanding Polars' *existence and trade-offs* is valuable professional awareness, not a reason to abandon the tool this entire module has taught.

**Where Pandas becomes inappropriate for very large-scale distributed processing**: once a dataset genuinely exceeds what fits comfortably in one machine's memory (Section 38), or a pipeline needs to run across a genuinely distributed cluster of machines for both storage and compute, Pandas is the wrong tool - not because it's poorly built, but because it was never designed for that scale. **Connections to the right tools at that point**: **Spark** (via PySpark, covered in the Data Engineering handbook) is the standard choice for large-scale distributed batch/streaming processing; **Dask** offers a Pandas-like API that scales out across a cluster, often an easier migration path for teams with existing Pandas code; **Polars** (Section 62) scales further on a single machine before distribution becomes necessary at all; **SQL / cloud data warehouses** (Section 33-34) should handle the heaviest aggregation work directly at the source rather than pulling raw data into Pandas at all. The professional judgment call - "is this still a Pandas job, or has it outgrown Pandas" - is itself an important skill, informed directly by everything in Sections 37-39.

# SECTION 65 - PANDAS DEBUGGING

Each entry: Cause -> Example -> Error -> Diagnosis -> Solution -> Prevention.

**`KeyError`** - *Cause*: referencing a column/index label that doesn't exist. *Example*: `df["Age"]` when the actual column is `"age"`. *Error*: `KeyError: 'Age'`. *Diagnosis*: `print(df.columns.tolist())` to see exact names. *Solution*: fix the typo/casing. *Prevention*: Section 13's column-cleaning function, applied immediately on load.

**`TypeError`** - *Cause*: an operation applied to an incompatible dtype. *Example*: `df["amount"] + "10"` when `amount` is numeric and `"10"` is a string. *Error*: `TypeError: unsupported operand type(s)`. *Diagnosis*: `df.dtypes` to check actual column types. *Solution*: `pd.to_numeric()` or `astype()` the offending column first. *Prevention*: check dtypes immediately after load (Section 6), before any arithmetic.

**`ValueError`** - *Cause*: a value that can't be converted/parsed as requested. *Example*: `pd.to_datetime(df["date"])` when some values are `"not_a_date"`. *Error*: `ValueError: time data "not_a_date" doesn't match format`. *Diagnosis*: inspect unique values in the offending column. *Solution*: `pd.to_datetime(df["date"], errors="coerce")` to turn unparseable values into `NaT` instead of crashing. *Prevention*: validate date columns (Section 19) before parsing at scale.

**`IndexError`** - *Cause*: `.iloc[]` referencing a position beyond the DataFrame's actual size. *Example*: `df.iloc[10000]` on a 500-row DataFrame. *Error*: `IndexError: single positional indexer is out-of-bounds`. *Diagnosis*: `len(df)` or `df.shape[0]`. *Solution*: bound-check before indexing, or use `.loc` with an actual label if that's really what's intended. *Prevention*: prefer `.loc` when working with meaningful labels; reserve `.iloc` for genuinely position-based logic where bounds are already known to be safe.

**`AttributeError`** - *Cause*: calling a Series/DataFrame method that doesn't exist, or calling a Series-only method (like `.str`) on a DataFrame, or vice versa. *Example*: `df.str.lower()` (should be `df["col"].str.lower()`). *Error*: `AttributeError: 'DataFrame' object has no attribute 'str'`. *Diagnosis*: check whether the object is a Series or DataFrame (`type(x)`). *Solution*: select the specific column first. *Prevention*: be deliberate about which object type a variable holds at each step - a common source of confusion after a groupby/filter changes an object's shape unexpectedly.

**Unexpected NaNs** - *Cause*: an arithmetic operation, merge, or reindex introduced NaN where none was expected. *Diagnosis*: `df.isna().sum()` immediately after each suspect step, not just at the end. *Solution*: trace back to the introducing step (often a `merge()` with `how="left"` producing NaN for unmatched rows, Section 30). *Prevention*: check row counts and NaN counts before and after every merge/join.

**Duplicate columns after merge** - *Cause*: both DataFrames had a same-named column that wasn't the join key. *Example*: both `customers` and `transactions` have a `date` column. *Error*: no error, but the result has `date_x` and `date_y`, silently confusing. *Diagnosis*: `df.columns.tolist()` right after the merge. *Solution*: rename before merging, or use the `suffixes=` parameter deliberately. *Prevention*: inspect both tables' columns before merging.

**Wrong dtype after load** - *Cause*: CSV loading, everything is inferred and a numeric column with a stray non-numeric value (e.g. `"N/A"`) becomes `object`. *Diagnosis*: `df.dtypes` immediately after load. *Solution*: `pd.to_numeric(df["col"], errors="coerce")`. *Prevention*: specify `dtype=`/`na_values=` explicitly at `read_csv()` time (Section 7).

**Incorrect merge (wrong key)** - *Cause*: merging on a column that looks like the right key but isn't unique/isn't the true business key. *Diagnosis*: row-count explosion (below). *Solution*: verify the actual key with `df[key].nunique()` vs `len(df)` before merging. *Prevention*: always sanity-check cardinality on both sides of a join before trusting the result.

**Row-count explosion after merge** - *Cause*: the join key isn't unique on one (or both) sides, so each match produces multiple output rows (a many-to-many join). *Example*: merging `customers` (1 row per customer) with `transactions` (many rows per customer) on `customer_id` naturally, correctly, multiplies rows - but merging two tables that were *both* supposed to be one-row-per-customer, where one accidentally has duplicates, silently and incorrectly multiplies rows too. *Diagnosis*: compare `len(result)` against the expected row count before and after. *Solution*: deduplicate (Section 18) before merging if duplication wasn't intended. *Prevention*: always check `len(result)` immediately after any merge - an unexpected row count is often the very first sign of a "row-count explosion" bug.

**Missing columns** - *Cause*: a column was dropped or renamed upstream, or a filter accidentally removed the only rows that had a value in a sparse column. *Diagnosis*: compare `df.columns` before/after each pipeline step. *Solution*: trace back to the introducing step. *Prevention*: assert expected columns exist at key pipeline checkpoints (Section 75).

**Shape mismatch (feeding scikit-learn)** - *Cause*: `X` has a different number of rows than `y`, often because a filter was applied to one but not the other. *Error*: a scikit-learn `ValueError` about inconsistent sample counts. *Diagnosis*: `X.shape[0]` vs `y.shape[0]`. *Solution*: always derive `X` and `y` from the *same* filtered DataFrame, not two separately-filtered ones. *Prevention*: filter the full DataFrame first, then split into `X`/`y` from that single filtered result (Section 53).

**Chained assignment** - *Cause*: writing `df[df.age > 30]["flag"] = True` in one chained expression. *Error*: `SettingWithCopyWarning`, and the assignment silently may not persist. *Diagnosis*: recognize the chained-bracket pattern. *Solution*: split into two steps with an explicit `.copy()` (Section 40). *Prevention*: never chain a filter and an assignment in a single expression.

**Date parsing problems** - *Cause*: mixed date formats within one column (`"2026-01-05"` and `"01/05/2026"` both present). *Diagnosis*: `pd.to_datetime(df["date"], errors="coerce").isna().sum()` reveals how many failed to parse. *Solution*: `format=` if one consistent format is expected, or a custom per-row parsing fallback for genuinely mixed sources. *Prevention*: validate date formats at ingestion (Section 19), and push back on sources with inconsistent formats rather than working around it indefinitely downstream.

# SECTION 67 - PANDAS PERFORMANCE LAB

```python
# [Python] Benchmark experiments - the exact comparisons a professional should be able to reproduce and explain
import time

def benchmark(label, fn):
    start = time.time()
    result = fn()
    print(f"{label}: {time.time() - start:.4f}s")
    return result

n = 500_000
big_df = pd.DataFrame({"amount": np.random.rand(n) * 1000, "category": np.random.choice(["A", "B", "C"], n)})

# 1. Vectorization vs apply()
benchmark("vectorized *1.1", lambda: big_df["amount"] * 1.1)
benchmark("apply() *1.1", lambda: big_df["amount"].apply(lambda x: x * 1.1))

# 2. object vs category dtype - memory
big_df["category_obj"] = big_df["category"].astype("object")
big_df["category_cat"] = big_df["category"].astype("category")
print("object memory:", big_df["category_obj"].memory_usage(deep=True))
print("category memory:", big_df["category_cat"].memory_usage(deep=True))

# 3. full loading vs chunksize (conceptual timing on a real large file)
# benchmark("full load", lambda: pd.read_csv("huge.csv"))
# benchmark("chunked load", lambda: sum(chunk["amount"].sum() for chunk in pd.read_csv("huge.csv", chunksize=100_000)))

# 4. unnecessary copies vs efficient in-place-safe operations
benchmark("with unnecessary intermediate copies", lambda: big_df.copy().copy()["amount"] * 2)
benchmark("no unnecessary copies", lambda: big_df["amount"] * 2)
```

**Why performance differs, explained**: vectorized operations execute in compiled C/NumPy code across the whole column at once, avoiding Python's per-element interpreter overhead entirely (Section 39); `category` dtype stores repeated strings once, referenced by small integer codes, versus `object` dtype storing a full separate Python string object per row (Section 15, 37); chunked reading trades a small amount of loop overhead for keeping peak memory usage bounded, which matters enormously (or is entirely irrelevant) depending on whether the file fits in RAM to begin with (Section 38); avoiding unnecessary `.copy()` calls avoids redundant memory allocation and copying work that adds up on large DataFrames, though (per Section 40-41) *deliberate*, necessary copies for correctness are always worth their cost. Running this exact benchmark script (or your own variant) against your own real datasets, and being able to explain *why* each result comes out the way it does, is a genuinely valuable professional habit, not just an academic exercise - it's what lets you make informed, evidence-based decisions instead of guessing when a real pipeline is slow.

```python
# [Python] Customer-level fraud features, built as of a specific reference date (as-of-date discipline, Section 52)
as_of_date = pd.Timestamp("2026-08-01")
history = txn_df[txn_df["transaction_date"] < as_of_date]   # STRICTLY before the reference date - the leakage guard

fraud_features = history.groupby("customer_id").agg(
    transaction_count=("transaction_id", "count"),
    total_transaction_amount=("amount", "sum"),
    average_transaction_amount=("amount", "mean"),
    maximum_transaction_amount=("amount", "max"),
    unique_merchants=("merchant_category", "nunique"),
)

last_7_days = history[history["transaction_date"] >= as_of_date - pd.Timedelta(days=7)]
fraud_features["transactions_last_7_days"] = last_7_days.groupby("customer_id")["transaction_id"].count()

last_30_days = history[history["transaction_date"] >= as_of_date - pd.Timedelta(days=30)]
fraud_features["transactions_last_30_days"] = last_30_days.groupby("customer_id")["transaction_id"].count()

time_diffs = history.sort_values(["customer_id", "transaction_date"]).groupby("customer_id")["transaction_date"].diff()
fraud_features["average_time_between_transactions_days"] = time_diffs.dt.days.groupby(history["customer_id"]).mean()

fraud_rate_by_merchant = history.groupby("merchant_category")["is_fraud"].mean()
history_with_rate = history.merge(fraud_rate_by_merchant.rename("merchant_fraud_rate"), on="merchant_category")
fraud_features["avg_fraud_rate_by_merchant"] = history_with_rate.groupby("customer_id")["merchant_fraud_rate"].mean()

fraud_features["unusual_transaction_indicator"] = (
    fraud_features["maximum_transaction_amount"] > fraud_features["average_transaction_amount"] * 5
).astype(int)

fraud_features = fraud_features.fillna(0).reset_index()
```

**Which features could create leakage depending on the prediction timestamp, examined explicitly**: `transaction_count`, `total_transaction_amount`, `average_transaction_amount`, `maximum_transaction_amount`, and `unique_merchants` are all safe **only because** they're computed from `history` - the pre-filtered slice strictly before `as_of_date`; computing them from the full `txn_df` instead (including transactions after the reference date) would leak future behavior directly. `transactions_last_7_days`/`30_days` are similarly safe only because they're windowed *backward* from `as_of_date`, never forward. `avg_fraud_rate_by_merchant` deserves special scrutiny: computing a merchant's fraud rate using *all* historical transactions (including ones that happen after the specific customer's transaction being scored) is a subtle aggregation-leakage risk (Section 52's table) - a fully rigorous production pipeline would compute this merchant fraud rate using *only* transactions before each individual transaction's own timestamp, not one global as-of-date, since merchant risk itself evolves over time; the simplified single-`as_of_date` version shown here is appropriate for a training-time feature table but would need this refinement for point-in-time-correct backtesting across many prediction dates. This exact tension - between "one reference date is often good enough for a first version" and "true point-in-time correctness requires per-row, not just per-batch, temporal discipline" - is one of the most important lessons a working data scientist or ML engineer learns from real fraud-modeling experience, and Section 52's leakage taxonomy is the vocabulary for reasoning about it precisely.

# SECTION 80 - MASTER CAPSTONE: "End-to-End Machine Learning Data Preparation Platform"

```
raw_data -> load -> inspect_schema -> validate -> clean -> handle_missing -> detect_duplicates
    -> analyze_outliers -> engineer_features -> detect_leakage -> prepare_X_y -> export_processed_data
        (+ eda_report, logging throughout, tests, professional project structure)
```

```python
# [Python] The platform's orchestrating entry point, tying every module together
import logging
from pathlib import Path
import pandas as pd

logging.basicConfig(level=logging.INFO, format="%(asctime)s %(levelname)s %(message)s")
logger = logging.getLogger("ml_data_prep_platform")

def run_pipeline(raw_path: Path, config: dict) -> tuple[pd.DataFrame, pd.Series]:
    logger.info("1. Loading raw data")
    df = pd.read_csv(raw_path)

    logger.info("2. Inspecting schema")
    logger.info(f"Shape: {df.shape}, dtypes: {df.dtypes.to_dict()}")

    logger.info("3. Validating")
    missing_required = set(config["required_columns"]) - set(df.columns)
    if missing_required:
        raise ValueError(f"Missing required columns: {missing_required}")

    logger.info("4. Cleaning")
    df = clean_column_names(df)

    logger.info("5. Handling missing values")
    for col, strategy in config["imputation_strategy"].items():
        if strategy == "median":
            df[col] = df[col].fillna(df[col].median())

    logger.info("6. Detecting duplicates")
    before = len(df)
    df = df.drop_duplicates(subset=config["business_key"])
    logger.info(f"Removed {before - len(df)} duplicate rows")

    logger.info("7. Analyzing outliers")
    for col in config["numeric_columns"]:
        q1, q3 = df[col].quantile(0.25), df[col].quantile(0.75)
        iqr = q3 - q1
        n_outliers = ((df[col] < q1 - 1.5*iqr) | (df[col] > q3 + 1.5*iqr)).sum()
        logger.info(f"{col}: {n_outliers} outliers detected (not automatically removed)")

    logger.info("8-9. Engineering features")
    df["tenure_bucket"] = pd.cut(df["tenure_months"], bins=[0, 12, 36, 100], labels=["new", "established", "loyal"])

    logger.info("10. Detecting leakage")
    for col in config.get("known_leakage_columns", []):
        if col in df.columns:
            logger.warning(f"Dropping known leakage column: {col}")
            df = df.drop(columns=[col])

    logger.info("11. Preparing X and y")
    feature_cols = [c for c in df.columns if c not in config["id_columns"] + [config["target_column"]]]
    X, y = df[feature_cols], df[config["target_column"]]

    logger.info("12. Exporting processed data")
    df.to_parquet(config["output_path"], index=False)

    logger.info("13. Pipeline complete")
    return X, y
```
